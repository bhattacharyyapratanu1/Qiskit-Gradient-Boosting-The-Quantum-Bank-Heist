```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report
)

# Separate features and target
X = train.drop(columns=["y", "duration"], errors="ignore")
y = train["y"].map({"no": 0, "yes": 1})

# Remove customer_id if it exists in the training sheet
X = X.drop(columns=["customer_id"], errors="ignore")

# Identify numerical and categorical columns
numeric_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_cols = X.select_dtypes(exclude=["int64", "float64"]).columns.tolist()

# Preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_cols),
    ("cat", categorical_pipeline, categorical_cols)
])

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000, class_weight="balanced"
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42
    ),
    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    )
}

# Use stratified cross-validation for a small training dataset
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

results = []

for name, model in models.items():
    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X,
        y,
        cv=cv,
        scoring=["accuracy", "precision", "recall", "f1", "roc_auc"]
    )

    results.append({
        "Model": name,
        "Accuracy": scores["test_accuracy"].mean(),
        "Precision": scores["test_precision"].mean(),
        "Recall": scores["test_recall"].mean(),
        "F1": scores["test_f1"].mean(),
        "ROC_AUC": scores["test_roc_auc"].mean()
    })

results_df = pd.DataFrame(results).sort_values(
    by="ROC_AUC", ascending=False
)

display(results_df)
```

```python
best_model_name = results_df.iloc[0]["Model"]
best_model = models[best_model_name]

final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", best_model)
])

final_model.fit(X, y)

# Prepare test features using the same feature columns
X_test = test.drop(
    columns=["y", "duration", "customer_id"],
    errors="ignore"
)

X_test = X_test[X.columns]

# Predict subscription probabilities
test_probabilities = final_model.predict_proba(X_test)[:, 1]

predictions = test.copy()
predictions["predicted_probability"] = test_probabilities

print("Selected model:", best_model_name)
display(predictions.head())
```

```python

import os
import pandas as pd

BASE_PATH = "/content/drive/MyDrive/ATMIQ_HACKATHON"

required_features = {
    "age", "job", "marital", "education", "default",
    "balance", "housing", "loan", "contact", "day",
    "month", "campaign", "pdays", "previous", "poutcome"
}

found_datasets = []

print("========== SEARCHING ALL DATASETS ==========")

for root, dirs, files in os.walk(BASE_PATH):
    for file in files:
        path = os.path.join(root, file)

        if file.lower().endswith(".csv"):
            try:
                df = pd.read_csv(path)
                found_datasets.append((path, "CSV", df))
            except Exception as e:
                print("CSV read error:", path, e)

        elif file.lower().endswith((".xlsx", ".xls")):
            try:
                xls = pd.ExcelFile(path)

                for sheet in xls.sheet_names:
                    df = pd.read_excel(path, sheet_name=sheet)
                    found_datasets.append(
                        (path + " | " + sheet, "EXCEL", df)
                    )
            except Exception as e:
                print("Excel read error:", path, e)

print("\n========== DATASET INSPECTION ==========")

source_candidates = []

for path, file_type, df in found_datasets:
    cols = {
        str(c).strip().lower()
        for c in df.columns
    }

    matches = required_features.issubset(cols)
    has_id = "customer_id" in cols
    has_target = "y" in cols

    print("\nLOCATION:", path)
    print("TYPE:", file_type)
    print("SHAPE:", df.shape)
    print("COLUMNS:", df.columns.tolist())
    print("HAS 15 FEATURES:", matches)
    print("HAS CUSTOMER ID:", has_id)
    print("HAS TARGET y:", has_target)

    if matches and has_id:
        source_candidates.append((path, df))

print("\n========== FINAL RESULT ==========")
print("Customer-level source candidates:", len(source_candidates))

for path, df in source_candidates:
    print("\nPOSSIBLE SOURCE:", path)
    display(df.head(3))

# Specifically inspect the Stage 2 data
print("\n========== STAGE 2 DATA ==========")

for path, file_type, df in found_datasets:
    if "Stage2_Data" in path:
        print("Location:", path)
        print("Columns:", df.columns.tolist())
        display(df.head(10))
```

```python

# ============================================================
# ATMIQ HACKATHON - COMPLETE REMAINING PIPELINE
# CODE 12 TO CODE 20
# ============================================================

import os
import itertools
import numpy as np
import pandas as pd
import joblib
from scipy.optimize import minimize

RESULTS = BASE + "/Results"
os.makedirs(RESULTS, exist_ok=True)

print("=" * 65)
print("ATMIQ HACKATHON - FINAL PIPELINE")
print("=" * 65)


# ============================================================
# CODE 12 - STAGE 1 TOP 10 PREDICTIONS
# ============================================================

print("\nCODE 12: TOP 10 PREDICTIONS")

prediction_path = RESULTS + "/stage1_predictions.csv"

if not os.path.exists(prediction_path):
    raise FileNotFoundError(
        "Existing stage1_predictions.csv not found."
    )

pred = pd.read_csv(prediction_path)

prob_candidates = [
    "predicted_probability",
    "probability",
    "probability_yes",
    "p_yes",
    "subscription_probability",
    "y_probability"
]

prob_col = next(
    (c for c in prob_candidates if c in pred.columns),
    None
)

if prob_col is None:
    raise ValueError(
        "Prediction probability column not found. "
        "Available columns: " + str(pred.columns.tolist())
    )

if len(pred) != 100:
    raise ValueError(
        f"Expected 100 predictions, found {len(pred)}."
    )

pred["predicted_probability"] = pd.to_numeric(
    pred[prob_col], errors="raise"
)

if not pred["predicted_probability"].between(0, 1).all():
    raise ValueError("Invalid probability values.")

if "test_row" not in pred.columns:
    if "Unnamed: 0" in pred.columns:
        pred = pred.rename(columns={"Unnamed: 0": "test_row"})
    else:
        # Valid only if the prediction file preserves original
        # Stage1_Test row order.
        pred["test_row"] = range(100)

top10 = (
    pred.sort_values(
        "predicted_probability",
        ascending=False,
        kind="stable"
    )
    .head(10)
    .copy()
    .reset_index(drop=True)
)

top10.insert(0, "rank", range(1, 11))

top10.to_csv(
    RESULTS + "/top10_stage1.csv",
    index=False
)

display(top10)


# ============================================================
# CODE 13 - VERIFIED CUSTOMER ID MAPPING
# ============================================================

print("\nCODE 13: CUSTOMER ID MAPPING")

if "customer_id" in top10.columns:
    print("Customer IDs already present in predictions.")

else:
    mapping_path = RESULTS + "/customer_id_mapping.csv"

    if not os.path.exists(mapping_path):
        top10.to_csv(
            RESULTS + "/stage1_top10_unmapped.csv",
            index=False
        )

        print("\nMAPPING NOT YET VERIFIED.")
        print("Saved Top 10 without fabricated IDs.")
        print("Required mapping file:")
        print(mapping_path)
        print("\nRequired columns:")
        print("test_row, customer_id")
        print("\nThe test_row must refer to the original")
        print("zero-based Stage1_Test row.")
        raise RuntimeError(
            "Pipeline safely stopped: official customer mapping "
            "is required before Stage 2."
        )

    mapping = pd.read_csv(mapping_path)

    required = {"test_row", "customer_id"}

    if not required.issubset(mapping.columns):
        raise ValueError(
            "Mapping file must contain test_row and customer_id."
        )

    if mapping["test_row"].duplicated().any():
        raise ValueError("Duplicate test_row in mapping.")

    if mapping["customer_id"].duplicated().any():
        raise ValueError("Duplicate customer IDs in mapping.")

    top10 = top10.merge(
        mapping[["test_row", "customer_id"]],
        on="test_row",
        how="left",
        validate="one_to_one"
    )

    if top10["customer_id"].isna().any():
        raise ValueError(
            "Some Top 10 customers have no verified ID."
        )

print("\nVERIFIED TOP 10:")
display(top10)


# ============================================================
# CODE 14 - EXPECTED NET VALUE
# ============================================================

print("\nCODE 14: EXPECTED NET VALUE")

top10 = top10.merge(
    stage2,
    on="customer_id",
    how="left",
    validate="one_to_one"
)

required_columns = [
    "contact_cost",
    "expected_deposit_value",
    "capacity_units"
]

if top10[required_columns].isna().any().any():
    raise ValueError(
        "Some customer IDs are missing Stage2_Data information."
    )

top10["expected_net_value"] = (
    top10["predicted_probability"]
    * top10["expected_deposit_value"]
    - top10["contact_cost"]
)

top10.to_csv(
    RESULTS + "/stage2_candidate_pool_verified.csv",
    index=False
)

print("EXPECTED NET VALUES:")
display(
    top10[
        [
            "rank",
            "customer_id",
            "predicted_probability",
            "contact_cost",
            "expected_deposit_value",
            "capacity_units",
            "expected_net_value"
        ]
    ]
)


# ============================================================
# CODE 15 - CLASSICAL EXACT OPTIMIZATION
# ============================================================

print("\nCODE 15: CLASSICAL OPTIMIZATION")

N = 10
K = 6
BUDGET = 25
CAPACITY = 22

best_selection = None
best_value = -float("inf")

feasible_count = 0

for combo in itertools.combinations(range(N), K):

    selected = top10.iloc[list(combo)]

    total_cost = selected["contact_cost"].sum()
    total_capacity = selected["capacity_units"].sum()
    total_value = selected["expected_net_value"].sum()

    if total_cost <= BUDGET and total_capacity <= CAPACITY:

        feasible_count += 1

        if total_value > best_value:
            best_value = total_value
            best_selection = selected.copy()

if best_selection is None:
    raise ValueError(
        "No feasible six-customer combination found."
    )

best_selection = best_selection.sort_values("rank")

print("Feasible combinations:", feasible_count)
print("Best expected net value:", best_value)

best_selection.to_csv(
    RESULTS + "/classical_optimal_selection.csv",
    index=False
)

display(best_selection)


# ============================================================
# CODE 16 - CONSTRAINT VERIFICATION
# ============================================================

print("\nCODE 16: CONSTRAINT VERIFICATION")

total_selected = len(best_selection)
total_cost = best_selection["contact_cost"].sum()
total_capacity = best_selection["capacity_units"].sum()
total_net_value = best_selection["expected_net_value"].sum()

verification = {
    "Selected customers": total_selected,
    "Required customers": K,
    "Total contact cost": total_cost,
    "Budget limit": BUDGET,
    "Total capacity": total_capacity,
    "Capacity limit": CAPACITY,
    "Expected net value": total_net_value,
    "Exact selection": total_selected == K,
    "Budget satisfied": total_cost <= BUDGET,
    "Capacity satisfied": total_capacity <= CAPACITY
}

for key, value in verification.items():
    print(key, ":", value)

if not all([
    total_selected == K,
    total_cost <= BUDGET,
    total_capacity <= CAPACITY
]):
    raise AssertionError("Optimization constraints violated!")


# ============================================================
# CODE 17 - QUANTUM QAOA STATEVECTOR SIMULATION
# ============================================================

print("\nCODE 17: QUANTUM QAOA SIMULATION")

# Ten qubits represent the ten candidate customers.
# Penalty terms enforce cardinality, budget and capacity.

from scipy.optimize import minimize

n = 10
dimension = 2 ** n

values = top10["expected_net_value"].to_numpy(dtype=float)
costs = top10["contact_cost"].to_numpy(dtype=float)
capacities = top10["capacity_units"].to_numpy(dtype=float)

states = np.arange(dimension)
bits = (
    (states[:, None] >> np.arange(n)) & 1
).astype(float)

number_selected = bits.sum(axis=1)
state_cost = bits @ costs
state_capacity = bits @ capacities
state_value = bits @ values

penalty = 1000.0

energy = (
    -state_value
    + penalty * (number_selected - K) ** 2
    + penalty * np.maximum(0, state_cost - BUDGET) ** 2
    + penalty * np.maximum(0, state_capacity - CAPACITY) ** 2
)

# Initial equal superposition
initial_state = np.ones(dimension, dtype=complex) / np.sqrt(dimension)

zero_indices = [
    np.where((states & (1 << q)) == 0)[0]
    for q in range(n)
]

one_indices = [
    idx + (1 << q) for q, idx in enumerate(zero_indices)
]

def qaoa_state(params):

    gamma, beta = params

    psi = initial_state.copy()

    # Cost Hamiltonian
    psi *= np.exp(-1j * gamma * energy)

    # X mixer
    cos_beta = np.cos(beta)
    sin_beta = -1j * np.sin(beta)

    for q in range(n):

        i0 = zero_indices[q]
        i1 = one_indices[q]

        a0 = psi[i0].copy()
        a1 = psi[i1].copy()

        psi[i0] = cos_beta * a0 + sin_beta * a1
        psi[i1] = sin_beta * a0 + cos_beta * a1

    return psi


def qaoa_objective(params):

    psi = qaoa_state(params)
    probabilities = np.abs(psi) ** 2

    return float(np.dot(probabilities, energy))


# Optimize QAOA angles
qaoa_result = minimize(
    qaoa_objective,
    x0=np.array([0.01, 0.5]),
    method="Nelder-Mead",
    options={"maxiter": 500}
)

psi = qaoa_state(qaoa_result.x)
qaoa_probabilities = np.abs(psi) ** 2

# Select the most probable feasible QAOA state
feasible_mask = (
    (number_selected == K)
    & (state_cost <= BUDGET)
    & (state_capacity <= CAPACITY)
)

if not feasible_mask.any():
    raise ValueError("No feasible QAOA states found.")

feasible_indices = np.where(feasible_mask)[0]

best_qaoa_state = feasible_indices[
    np.argmax(qaoa_probabilities[feasible_indices])
]

qaoa_selection = bits[best_qaoa_state].astype(int)

qaoa_selected = top10.loc[
    qaoa_selection == 1
].copy()

qaoa_value = qaoa_selected["expected_net_value"].sum()

qaoa_selected.to_csv(
    RESULTS + "/qaoa_selection.csv",
    index=False
)

print("QAOA optimization completed.")
print("QAOA expected net value:", qaoa_value)
print("QAOA selected customers:")
display(qaoa_selected)


# ============================================================
# CODE 18 - CLASSICAL VS QUANTUM COMPARISON
# ============================================================

print("\nCODE 18: CLASSICAL VS QUANTUM")

comparison = pd.DataFrame([
    {
        "Method": "Classical Exact Enumeration",
        "Selected": len(best_selection),
        "Expected Net Value": total_net_value,
        "Total Cost": total_cost,
        "Total Capacity": total_capacity
    },
    {
        "Method": "QAOA Statevector Simulation",
        "Selected": len(qaoa_selected),
        "Expected Net Value": qaoa_value,
        "Total Cost": qaoa_selected["contact_cost"].sum(),
        "Total Capacity": qaoa_selected["capacity_units"].sum()
    }
])

comparison.to_csv(
    RESULTS + "/classical_quantum_comparison.csv",
    index=False
)

display(comparison)

print(
    "\nQAOA optimality gap:",
    total_net_value - qaoa_value
)


# ============================================================
# CODE 19 - STAGE 3 PROBABILITY
# ============================================================

print("\nCODE 19: STAGE 3 PROBABILITY")

# Probability that at least 3 out of 6 customers subscribe.
# Assumes independent customer outcomes.

probabilities = best_selection[
    "predicted_probability"
].to_numpy(dtype=float)

dp = np.zeros(len(probabilities) + 1)
dp[0] = 1.0

for p in probabilities:

    updated = np.zeros_like(dp)

    for j in range(len(probabilities) + 1):

        updated[j] += dp[j] * (1 - p)

        if j + 1 < len(dp):
            updated[j + 1] += dp[j] * p

    dp = updated

prob_at_least_3 = dp[3:].sum()

print("Selected customer probabilities:")
print(probabilities)

print("\nProbability of at least 3 successes:")
print(f"{prob_at_least_3:.6f}")

print(f"Percentage: {prob_at_least_3 * 100:.2f}%")

stage3 = pd.DataFrame({
    "successful_customers": range(7),
    "probability": dp[:7]
})

stage3.to_csv(
    RESULTS + "/stage3_probability_distribution.csv",
    index=False
)

print("\nProbability distribution:")
display(stage3)


# ============================================================
# CODE 20 - FINAL HACKATHON RESULTS
# ============================================================

print("\nCODE 20: FINAL RESULTS")

final_report = pd.DataFrame([
    ["Top 10 candidates", len(top10)],
    ["Selected customers", total_selected],
    ["Contact budget used", total_cost],
    ["Budget limit", BUDGET],
    ["Capacity used", total_capacity],
    ["Capacity limit", CAPACITY],
    ["Classical expected net value", total_net_value],
    ["QAOA expected net value", qaoa_value],
    ["Probability of at least 3 successes", prob_at_least_3],
    ["Probability percentage", prob_at_least_3 * 100],
    ["Feasible combinations", feasible_count]
], columns=["Metric", "Result"])

final_report.to_csv(
    RESULTS + "/ATMIQ_FINAL_REPORT.csv",
    index=False
)

display(final_report)

print("\n" + "=" * 65)
print("ATMIQ PIPELINE FINISHED")
print("=" * 65)

print("\nRESULTS SAVED IN:")
print(RESULTS)

print("\nGenerated files:")
for f in sorted(os.listdir(RESULTS)):
    if f.endswith(".csv"):
        print(" -", f)
```

```python

# CELL 7: CANDIDATE POOL AND PROVISIONAL MAPPING AUDIT

import os
import pandas as pd
import numpy as np

stage2 = pd.read_csv(
    os.path.join(BASE, "stage2_data.csv")
) if "stage2" not in globals() else stage2.copy()

candidate_path = os.path.join(
    RESULTS, "stage2_candidate_pool.csv"
)

provisional_path = os.path.join(
    RESULTS, "stage1_predictions_provisional.csv"
)

if not os.path.isfile(candidate_path):
    raise FileNotFoundError(candidate_path)

if not os.path.isfile(provisional_path):
    raise FileNotFoundError(provisional_path)

candidate = pd.read_csv(candidate_path)
provisional = pd.read_csv(provisional_path)

financial_cols = [
    "customer_id",
    "contact_cost",
    "expected_deposit_value",
    "capacity_units",
    "customer_segment"
]

required_cols = financial_cols

for col in required_cols:
    if col not in candidate.columns or col not in stage2.columns:
        raise ValueError(f"Missing financial column: {col}")

# Check duplicate IDs
print("CANDIDATE RECORDS:", len(candidate))
print("UNIQUE CANDIDATE IDs:", candidate.customer_id.nunique())
print("DUPLICATE CANDIDATE IDs:",
      candidate.customer_id.duplicated().sum())

# Check whether all candidate IDs exist in Stage 2
stage2_ids = set(stage2.customer_id.astype(str))
candidate_ids = set(candidate.customer_id.astype(str))

missing_ids = sorted(candidate_ids - stage2_ids)

print("\nCANDIDATE IDs NOT IN ORIGINAL STAGE 2:", missing_ids)

# Verify financial values against original Stage 2
comparison = candidate[financial_cols].merge(
    stage2[financial_cols],
    on="customer_id",
    how="left",
    suffixes=("_candidate", "_original"),
    indicator=True,
    validate="one_to_one"
)

for col in financial_cols[1:]:
    comparison[f"{col}_matches"] = (
        comparison[f"{col}_candidate"]
        == comparison[f"{col}_original"]
    )

print("\nFINANCIAL RECORD VALIDATION:")
print("IDs found in Stage 2:",
      (comparison["_merge"] == "both").sum())

for col in financial_cols[1:]:
    print(
        col,
        "matches:",
        int(comparison[f"{col}_matches"].sum())
    )

# Inspect provisional IDs
if "customer_id" not in provisional.columns:
    raise ValueError("Provisional file has no customer_id")

if "mapping_status" not in provisional.columns:
    raise ValueError("Provisional file has no mapping_status")

provisional_ids = set(
    provisional.customer_id.dropna().astype(str)
)

print("\nPROVISIONAL ID COUNT:", len(provisional_ids))
print("PROVISIONAL IDs IN CANDIDATE POOL:",
      len(provisional_ids & candidate_ids))
print("PROVISIONAL IDs IN STAGE 2:",
      len(provisional_ids & stage2_ids))

print("\nMAPPING STATUS COUNTS:")
print(provisional.mapping_status.value_counts(dropna=False))

# Save audit
audit = {
    "candidate_records": len(candidate),
    "unique_candidate_ids": candidate.customer_id.nunique(),
    "duplicate_candidate_ids": int(
        candidate.customer_id.duplicated().sum()
    ),
    "candidate_ids_missing_from_stage2": len(missing_ids),
    "provisional_ids": len(provisional_ids),
    "provisional_ids_in_candidate_pool": len(
        provisional_ids & candidate_ids
    ),
    "provisional_ids_in_stage2": len(
        provisional_ids & stage2_ids
    ),
    "provisional_mapping_verified": False
}

audit_df = pd.DataFrame(
    list(audit.items()),
    columns=["check", "result"]
)

display(audit_df)

audit_df.to_csv(
    os.path.join(RESULTS, "stage2_candidate_audit.csv"),
    index=False
)

print("\nAUDIT SAVED: stage2_candidate_audit.csv")
print("\nNo customer mapping has been declared official.")
```

```python

# CELL 75: EXACT CUSTOMER-TO-TEST-ROW MATCHING

import pandas as pd
import os

WORKBOOK = (
    "/content/drive/MyDrive/ATMIQ_HACKATHON/"
    "Dataset/bank_campaign_participant_release.xlsx"
)

test = pd.read_excel(WORKBOOK, sheet_name="Stage1_Test")
stage2 = pd.read_excel(WORKBOOK, sheet_name="Stage2_Data")

# Original Stage 1 test features
feature_cols = list(test.columns)

print("=" * 70)
print("FEATURE MATCHING INVESTIGATION")
print("=" * 70)

print("Stage1_Test columns:", feature_cols)
print("Stage2_Data columns:", list(stage2.columns))

# Verify that all test features exist in Stage 2
missing = [c for c in feature_cols if c not in stage2.columns]

if missing:
    print("\nMISSING FEATURES:", missing)

else:
    print("\nAll test features exist in Stage2_Data.")

    # Check duplicate feature combinations
    test_duplicates = test.duplicated(
        subset=feature_cols
    ).sum()

    stage2_duplicates = stage2.duplicated(
        subset=feature_cols
    ).sum()

    print("\nDuplicate test feature combinations:", test_duplicates)
    print("Duplicate Stage 2 feature combinations:", stage2_duplicates)

    # Preserve original test row numbers
    test_indexed = test.copy()
    test_indexed["test_row"] = range(len(test_indexed))

    # Exact match using all original features
    mapping = test_indexed.merge(
        stage2[["customer_id"] + feature_cols],
        on=feature_cols,
        how="left",
        indicator=True
    )

    # Match diagnostics
    print("\n" + "=" * 70)
    print("MATCHING RESULTS")
    print("=" * 70)

    print("Test records:", len(test))
    print("Matched records:", (mapping["_merge"] == "both").sum())
    print("Unmatched records:", (mapping["_merge"] == "left_only").sum())
    print("Total mapping rows:", len(mapping))

    # Check whether every test row has exactly one customer ID
    counts = mapping.groupby("test_row")["customer_id"].nunique()

    print("\nRows with exactly one matching ID:", (counts == 1).sum())
    print("Rows with zero matching IDs:", (counts == 0).sum())
    print("Rows with multiple matching IDs:", (counts > 1).sum())

    if (
        len(mapping) == len(test)
        and (counts == 1).all()
        and mapping["customer_id"].notna().all()
    ):
        print("\nVERIFIED ONE-TO-ONE FEATURE MATCH FOUND")

        final_mapping = mapping[
            ["test_row", "customer_id"]
        ].sort_values("test_row")

        display(final_mapping)

        output = (
            "/content/drive/MyDrive/ATMIQ_HACKATHON/"
            "Results/verified_test_customer_mapping.csv"
        )

        final_mapping.to_csv(output, index=False)

        print("\nSaved mapping:", output)

    else:
        print("\nONE-TO-ONE MAPPING NOT ESTABLISHED.")
        print("Do not use this mapping for optimization.")
```

```python

# CELL 81 - FULL DATASET MEMBERSHIP TEST (SAFER VERSION)

import pandas as pd
from collections import Counter

WORKBOOK = "/content/drive/MyDrive/ATMIQ_HACKATHON/Dataset/bank_campaign_participant_release.xlsx"
BANK_CSV = "/content/drive/MyDrive/ATMIQ_HACKATHON/Dataset/bank.csv"

train = pd.read_excel(WORKBOOK, sheet_name="Stage1_Train")
test = pd.read_excel(WORKBOOK, sheet_name="Stage1_Test")
source = pd.read_csv(BANK_CSV, sep=";", keep_default_na=False)

# Exclude identifiers; compare only columns present in both datasets.
excluded = {"customer_id", "id"}

features = [
    col for col in test.columns
    if col in source.columns and col.lower() not in excluded
]

missing_columns = [
    col for col in test.columns
    if col not in source.columns and col.lower() not in excluded
]

print("=" * 70)
print("FULL DATASET MEMBERSHIP TEST")
print("=" * 70)

print("Original bank.csv rows:", len(source))
print("Official training rows:", len(train))
print("Official test rows:", len(test))
print("Common comparison features:", len(features))
print("Missing non-ID columns:", missing_columns)

if not features:
    raise ValueError("No common feature columns available.")

def normalize_records(df):
    result = df[features].copy()

    numeric_cols = [
        "age", "balance", "day", "campaign",
        "pdays", "previous"
    ]

    for col in numeric_cols:
        if col in result.columns:
            result[col] = pd.to_numeric(
                result[col], errors="coerce"
            )

    for col in result.columns:
        if col not in numeric_cols:
            result[col] = result[col].astype(str).str.strip()

    return result

official = normalize_records(train)
original = normalize_records(source)

source_records = Counter(map(tuple, original.to_numpy()))
official_records = Counter(map(tuple, official.to_numpy()))

matched_count = sum(
    (source_records & official_records).values()
)

unmatched_count = len(official) - matched_count

# Row-level membership display
source_record_set = set(source_records)

matches = official.apply(
    lambda row: tuple(row) in source_record_set,
    axis=1
)

print("\nRESULTS")
print("Matching training records:", matched_count)
print("Training records NOT found:", unmatched_count)
print("Match percentage:", round(
    matched_count / len(official) * 100, 4
) if len(official) else 0)

print("\nFirst 10 unmatched training records:")
display(train.loc[~matches].head(10))

print("\nFirst 10 matched training records:")
display(train.loc[matches].head(10))

print("\nCELL 81 COMPLETED")
```

```python

# CELL 82: VERIFY CUSTOMER ID ASSIGNMENT USING RANDOM_STATE=42

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

WORKBOOK = "/content/drive/MyDrive/ATMIQ_HACKATHON/Dataset/bank_campaign_participant_release.xlsx"

train = pd.read_excel(WORKBOOK, sheet_name="Stage1_Train")
stage2 = pd.read_excel(WORKBOOK, sheet_name="Stage2_Data")

# Recreate the documented 300/100 split of 400 records
train_positions, test_positions = train_test_split(
    np.arange(400),
    test_size=0.25,
    random_state=42
)

# IDs that would result if C0001-C0400 were assigned
# before the train/test split
predicted_train_ids = [
    f"C{i+1:04d}" for i in train_positions
]

actual_train_ids = train["customer_id"].astype(str).tolist()

# Compare row by row
matches = [
    actual == predicted
    for actual, predicted in zip(
        actual_train_ids,
        predicted_train_ids
    )
]

print("=" * 70)
print("CUSTOMER ID RECONSTRUCTION TEST")
print("=" * 70)

print("Training rows:", len(actual_train_ids))
print("Exact row-position matches:", sum(matches))
print("Total rows:", len(matches))
print("Match percentage:", round(100 * sum(matches) / len(matches), 2), "%")

print("\nFirst 15 comparisons:")

comparison = pd.DataFrame({
    "workbook_customer_id": actual_train_ids[:15],
    "predicted_customer_id": predicted_train_ids[:15],
    "match": matches[:15]
})

display(comparison)

# Also compare the complete ID sets
print("\nComplete ID-set comparison:")
print(
    "Same ID set:",
    set(actual_train_ids) == set(predicted_train_ids)
)

print("\nCELL 82 COMPLETED")
```

```python

# =============================================================================
# CELL 83 - TRAINING CUSTOMER ID TO ORIGINAL BANK CSV ROW MAPPING
# DUPLICATE-AWARE VERSION
# =============================================================================

import pandas as pd
import numpy as np
import os
from collections import defaultdict

WORKBOOK = "/content/drive/MyDrive/ATMIQ_HACKATHON/Dataset/bank_campaign_participant_release.xlsx"
BANK_CSV = "/content/drive/MyDrive/ATMIQ_HACKATHON/Dataset/bank.csv"

train = pd.read_excel(WORKBOOK, sheet_name="Stage1_Train")
test = pd.read_excel(WORKBOOK, sheet_name="Stage1_Test")
bank_source = pd.read_csv(BANK_CSV, sep=";", keep_default_na=False)

# Only compare actual features present in both datasets.
excluded = {"customer_id", "id"}

features = [
    col for col in test.columns
    if col in bank_source.columns
    and col.lower() not in excluded
]

missing = [
    col for col in test.columns
    if col not in bank_source.columns
    and col.lower() not in excluded
]

print("=" * 75)
print("CELL 83 - SOURCE ROW MAPPING AUDIT")
print("=" * 75)

print("Training rows:", len(train))
print("Original bank rows:", len(bank_source))
print("Comparison features:", len(features))
print("Missing non-ID features:", missing)

if not features:
    raise ValueError("No common feature columns found.")

numeric_cols = [
    "age", "balance", "day", "campaign",
    "pdays", "previous"
]

def normalize(df):
    result = df[features].copy()

    for col in numeric_cols:
        if col in result.columns:
            result[col] = pd.to_numeric(
                result[col], errors="coerce"
            )

    for col in result.columns:
        if col not in numeric_cols:
            result[col] = (
                result[col].astype(str).str.strip()
            )

    return result

source = normalize(bank_source)
training = normalize(train)

# Build lookup preserving every matching original row position.
lookup = defaultdict(list)

for position, record in enumerate(
    source.itertuples(index=False, name=None)
):
    lookup[record].append(position)

mapping_rows = []

for train_position, record in enumerate(
    training.itertuples(index=False, name=None)
):

    matches = lookup.get(record, [])

    if len(matches) == 1:
        status = "UNIQUE_MATCH"
        recovered_position = matches[0]

    elif len(matches) > 1:
        status = "AMBIGUOUS_DUPLICATE"
        recovered_position = None

    else:
        status = "NO_MATCH"
        recovered_position = None

    mapping_rows.append({
        "customer_id": train.iloc[train_position]["customer_id"]
            if "customer_id" in train.columns else None,
        "training_position": train_position,
        "original_bank_csv_row": recovered_position,
        "candidate_source_rows": matches,
        "candidate_count": len(matches),
        "mapping_status": status
    })

mapping = pd.DataFrame(mapping_rows)

# ---------------------------------------------------------------------------
# SUMMARY
# ---------------------------------------------------------------------------

unique_count = int(
    (mapping["mapping_status"] == "UNIQUE_MATCH").sum()
)

ambiguous_count = int(
    (mapping["mapping_status"] == "AMBIGUOUS_DUPLICATE").sum()
)

unmatched_count = int(
    (mapping["mapping_status"] == "NO_MATCH").sum()
)

print("\n1. MAPPING SUMMARY")
print("Unique mappings:", unique_count)
print("Ambiguous mappings:", ambiguous_count)
print("Unmatched records:", unmatched_count)

print("\n2. FIRST 30 MAPPINGS")
display(mapping.head(30))

print("\n3. MAPPING STATUS COUNTS")
display(mapping["mapping_status"].value_counts())

# ---------------------------------------------------------------------------
# SAVE INVESTIGATION RESULTS
# ---------------------------------------------------------------------------

output_dir = (
    "/content/drive/MyDrive/ATMIQ_HACKATHON/Results"
)

os.makedirs(output_dir, exist_ok=True)

output = os.path.join(
    output_dir,
    "cell83_source_row_mapping.csv"
)

mapping.to_csv(output, index=False)

print("\n4. EXPORT")
print("Saved:", output)

print("\n" + "=" * 75)
print("CELL 83 COMPLETED")
print("=" * 75)
```

```python

# ============================================================
# ATMIQ CELL 83: EXACT SOURCE RECORD RECONSTRUCTION
# ============================================================

import pandas as pd
import numpy as np
import os

# 1. Identify common columns
features = [
    col for col in train.columns
    if col != "customer_id" and col in bank_source.columns
]

print("COMMON FEATURES:", len(features))
print(features)

# 2. Numeric columns
numeric_cols = [
    "age", "balance", "day",
    "campaign", "pdays", "previous"
]

# 3. Normalize records consistently
def normalize_value(value, col):
    if pd.isna(value):
        return None

    if col in numeric_cols:
        return float(value)

    return str(value).strip()

def make_signature(row, columns):
    return tuple(
        normalize_value(row[col], col)
        for col in columns
    )

# 4. Build source dataset lookup
source_lookup = {}

for position in range(len(bank_source)):
    record = make_signature(
        bank_source.iloc[position],
        features
    )

    source_lookup.setdefault(record, []).append(position)

# 5. Match all 300 training records
results = []

for _, row in train.iterrows():

    signature = make_signature(row, features)
    positions = source_lookup.get(signature, [])

    results.append({
        "customer_id": row["customer_id"],
        "source_positions": positions,
        "number_of_matches": len(positions),
        "original_bank_csv_row": (
            positions[0] if len(positions) == 1 else None
        )
    })

mapping = pd.DataFrame(results)

# 6. Print results
print("\n" + "=" * 60)
print("CELL 83 RESULTS")
print("=" * 60)

print("Training records:", len(mapping))
print(
    "Records found:",
    (mapping["number_of_matches"] > 0).sum()
)
print(
    "Records not found:",
    (mapping["number_of_matches"] == 0).sum()
)
print(
    "Uniquely matched:",
    (mapping["number_of_matches"] == 1).sum()
)
print(
    "Ambiguous matches:",
    (mapping["number_of_matches"] > 1).sum()
)

print("\nFIRST 30 MATCHES:")
display(mapping.head(30))

# 7. Save results
output = (
    "/content/drive/MyDrive/ATMIQ_HACKATHON/"
    "Results/cell83_source_row_mapping.csv"
)

os.makedirs(os.path.dirname(output), exist_ok=True)
mapping.to_csv(output, index=False)

print("\nSaved:", output)
print("\nCELL 83 COMPLETED")
```

```python

# ============================================================
# ATMIQ CELL 85: CORRECTED TEST MEMBERSHIP MATCHING
# Handles missing target column y
# ============================================================

import pandas as pd
import numpy as np

WORKBOOK = (
    "/content/drive/MyDrive/ATMIQ_HACKATHON/"
    "Dataset/bank_campaign_participant_release.xlsx"
)

test = pd.read_excel(
    WORKBOOK,
    sheet_name="Stage1_Test"
)

# Only use columns actually present in BOTH datasets
features = [
    c for c in test.columns
    if c in bank_source.columns
    and c != "customer_id"
]

numeric_cols = [
    "age", "balance", "day",
    "campaign", "pdays", "previous",
    "duration"
]

print("TEST COLUMNS:", list(test.columns))
print("\nMATCHING FEATURES:", features)
print("NUMBER OF FEATURES:", len(features))

def signature(row):
    values = []

    for col in features:
        value = row[col]

        if pd.isna(value):
            values.append(None)

        elif col in numeric_cols:
            values.append(float(value))

        else:
            values.append(str(value).strip())

    return tuple(values)

# Build source lookup
source_lookup = {}

for pos in range(len(bank_source)):
    key = signature(bank_source.iloc[pos])
    source_lookup.setdefault(key, []).append(pos)

# Match test records
test_results = []

for i in range(len(test)):
    key = signature(test.iloc[i])
    positions = source_lookup.get(key, [])

    test_results.append({
        "test_row": i,
        "source_positions": positions,
        "match_count": len(positions),
        "status": (
            "UNIQUE" if len(positions) == 1
            else "NOT_FOUND" if len(positions) == 0
            else "AMBIGUOUS"
        )
    })

test_mapping = pd.DataFrame(test_results)

print("\n" + "=" * 55)
print("CELL 85: TEST MEMBERSHIP RESULTS")
print("=" * 55)

print("Total test rows:", len(test_mapping))
print("Unique matches:",
      (test_mapping.match_count == 1).sum())
print("Unmatched:",
      (test_mapping.match_count == 0).sum())
print("Ambiguous:",
      (test_mapping.match_count > 1).sum())

print("\nFIRST 30 RESULTS:")
display(test_mapping.head(30))

# Save results
output = (
    "/content/drive/MyDrive/ATMIQ_HACKATHON/"
    "Results/cell85_test_membership.csv"
)

test_mapping.to_csv(output, index=False)

print("\nSaved:", output)
print("CELL 85 COMPLETED")
```

```python

# ============================================================
# ATMIQ CELL 89: ALTERNATIVE SAMPLING INVESTIGATION
# ============================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# Use positional row numbers consistently
source = bank_source.reset_index(drop=True)

actual_positions = (
    set(mapping["original_bank_csv_row"].astype(int))
    |
    set(
        test_mapping["source_positions"].apply(
            lambda x: int(x[0])
        )
    )
)

n = len(source)
indices = np.arange(n)

candidates = {}

# Method 1: Pandas sample
candidates["PANDAS_SAMPLE"] = set(
    source.sample(n=400, random_state=42).index
)

# Method 2: sklearn train_size
a, b = train_test_split(
    indices, train_size=400, random_state=42
)
candidates["SKLEARN_TRAIN_SIZE"] = set(a)

# Method 3: sklearn test_size
a, b = train_test_split(
    indices, test_size=400, random_state=42
)
candidates["SKLEARN_TEST_SIZE"] = set(b)

# Method 4: Stratified train_size
if "y" in source.columns:
    a, b = train_test_split(
        indices,
        train_size=400,
        random_state=42,
        stratify=source["y"]
    )
    candidates["STRATIFIED_TRAIN_SIZE"] = set(a)

    # Method 5: Stratified test_size
    a, b = train_test_split(
        indices,
        test_size=400,
        random_state=42,
        stratify=source["y"]
    )
    candidates["STRATIFIED_TEST_SIZE"] = set(b)

# Compare each candidate against actual 400 source records
report = []

for name, candidate in candidates.items():
    overlap = actual_positions & candidate

    report.append({
        "method": name,
        "candidate_size": len(candidate),
        "actual_overlap": len(overlap),
        "missing_actual_rows": len(actual_positions - candidate),
        "extra_candidate_rows": len(candidate - actual_positions),
        "exact_match": candidate == actual_positions
    })

results89 = pd.DataFrame(report)

print("=" * 65)
print("CELL 89: SAMPLING METHOD COMPARISON")
print("=" * 65)

display(results89)

print("\nACTUAL SOURCE ROW COUNT:", len(actual_positions))
print("SOURCE DATASET ROW COUNT:", n)

print("\nACTUAL TARGET DISTRIBUTION:")
display(
    source.iloc[sorted(actual_positions)]["y"].value_counts()
)

print("\nCELL 89 COMPLETED")
```

```python

# ============================================================
# ATMIQ CELL 90: STRATIFIED SAMPLING INVESTIGATION
# ============================================================

from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np

source = bank_source.reset_index(drop=True)
indices = np.arange(len(source))

# Recreate the best candidate
candidate_train, candidate_test = train_test_split(
    indices,
    train_size=400,
    random_state=42,
    stratify=source["y"]
)

candidate_set = set(candidate_train)
actual_set = actual_positions

overlap = actual_set & candidate_set
missing = actual_set - candidate_set
extra = candidate_set - actual_set

print("=" * 65)
print("CELL 90: STRATIFIED SAMPLE DIAGNOSTICS")
print("=" * 65)

# 1. Compare target distributions
print("\n1. ORIGINAL DATASET TARGET COUNTS")
display(source["y"].value_counts())

print("\n2. OFFICIAL 400 RECORD TARGET COUNTS")
display(source.iloc[sorted(actual_set)]["y"].value_counts())

print("\n3. STRATIFIED CANDIDATE TARGET COUNTS")
display(source.iloc[sorted(candidate_set)]["y"].value_counts())

# 2. Compare overlap by target
print("\n4. OVERLAP BY TARGET")

for label in sorted(source["y"].dropna().unique()):
    actual_label = {
        i for i in actual_set
        if source.iloc[i]["y"] == label
    }

    candidate_label = {
        i for i in candidate_set
        if source.iloc[i]["y"] == label
    }

    print(
        label,
        "| official:", len(actual_label),
        "| candidate:", len(candidate_label),
        "| overlap:", len(actual_label & candidate_label),
        "| missing:", len(actual_label - candidate_label),
        "| extra:", len(candidate_label - actual_label)
    )

# 3. Inspect differing source records
print("\n5. FIRST 20 MISSING OFFICIAL SOURCE ROWS")
display(
    source.iloc[sorted(missing)].head(20)
)

print("\n6. FIRST 20 EXTRA CANDIDATE SOURCE ROWS")
display(
    source.iloc[sorted(extra)].head(20)
)

# 4. Check whether candidate ordering resembles official IDs
print("\n7. CANDIDATE SUMMARY")
print("Official sample size:", len(actual_set))
print("Candidate sample size:", len(candidate_set))
print("Overlap:", len(overlap))
print("Missing:", len(missing))
print("Extra:", len(extra))
print("Exact match:", actual_set == candidate_set)

print("\nCELL 90 COMPLETED")
```

```python

# ============================================================
# ATMIQ CELL 91: OPTIMIZED RANDOM STATE INVESTIGATION
# ============================================================

from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np
import time

source = bank_source.reset_index(drop=True)

indices = np.arange(len(source))
labels = source["y"].to_numpy()

actual_set = set(map(int, actual_positions))

assert len(actual_set) == 400
assert min(actual_set) >= 0
assert max(actual_set) < len(source)

# Convert actual sample to a Boolean mask
actual_mask = np.zeros(len(source), dtype=bool)
actual_mask[list(actual_set)] = True

# Class masks
yes_mask = labels == "yes"
no_mask = ~yes_mask

actual_yes_count = int(np.sum(actual_mask & yes_mask))
actual_no_count = int(np.sum(actual_mask & no_mask))

print("=" * 70)
print("CELL 91: OPTIMIZED RANDOM STATE INVESTIGATION")
print("=" * 70)

print("Actual sample size:", len(actual_set))
print("Actual yes records:", actual_yes_count)
print("Actual no records:", actual_no_count)

results = []

start_time = time.time()

# Search seeds
for seed in range(10001):

    candidate_train, _ = train_test_split(
        indices,
        train_size=400,
        random_state=seed,
        stratify=labels
    )

    candidate_mask = np.zeros(len(source), dtype=bool)
    candidate_mask[candidate_train] = True

    total_overlap = int(
        np.sum(actual_mask & candidate_mask)
    )

    yes_overlap = int(
        np.sum(actual_mask & candidate_mask & yes_mask)
    )

    no_overlap = int(
        np.sum(actual_mask & candidate_mask & no_mask)
    )

    exact_match = total_overlap == 400

    results.append((
        seed,
        total_overlap,
        no_overlap,
        yes_overlap,
        exact_match
    ))

    if exact_match:
        print("\nEXACT MATCH FOUND AT SEED:", seed)
        break

results_df = pd.DataFrame(
    results,
    columns=[
        "seed",
        "total_overlap",
        "no_overlap",
        "yes_overlap",
        "exact_match"
    ]
)

elapsed = time.time() - start_time

print("\n1. BEST 15 RANDOM STATES")

display(
    results_df.sort_values(
        ["total_overlap", "yes_overlap"],
        ascending=False
    ).head(15)
)

print("\n2. EXACT MATCH COUNT")
print(int(results_df["exact_match"].sum()))

print("\n3. BEST YES-CLASS OVERLAP")
print(results_df["yes_overlap"].max())

print("\n4. BEST NO-CLASS OVERLAP")
print(results_df["no_overlap"].max())

print("\n5. RANDOM STATE 42")
display(results_df[results_df["seed"] == 42])

print("\n6. NUMBER OF SEEDS WITH PERFECT NO-CLASS MATCH")
print(int((results_df["no_overlap"] == actual_no_count).sum()))

print("\n7. EXECUTION TIME")
print(round(elapsed, 2), "seconds")

print("\n" + "=" * 70)
print("CELL 91 COMPLETED")
print("=" * 70)
```

```python

# ============================================================
# ATMIQ CELL 92: INDEPENDENT CLASS SAMPLING INVESTIGATION
# ============================================================

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

source = bank_source.reset_index(drop=True)
actual_set = set(map(int, actual_positions))

# Separate official positions by target
actual_no = {
    i for i in actual_set if source.iloc[i]["y"] == "no"
}
actual_yes = {
    i for i in actual_set if source.iloc[i]["y"] == "yes"
}

class_positions = {
    label: np.where(source["y"].to_numpy() == label)[0]
    for label in ["no", "yes"]
}

sample_sizes = {
    "no": len(actual_no),
    "yes": len(actual_yes)
}

print("=" * 70)
print("CELL 92: INDEPENDENT CLASS SAMPLING INVESTIGATION")
print("=" * 70)

results = []

def evaluate(method, label, candidate):
    actual = actual_no if label == "no" else actual_yes
    candidate = set(map(int, candidate))

    results.append({
        "method": method,
        "class": label,
        "official_count": len(actual),
        "candidate_count": len(candidate),
        "overlap": len(actual & candidate),
        "missing": len(actual - candidate),
        "extra": len(candidate - actual),
        "exact_match": actual == candidate
    })

# Method 1: Independent sklearn sampling
for label in ["no", "yes"]:

    positions = class_positions[label]

    selected, _ = train_test_split(
        positions,
        train_size=sample_sizes[label],
        random_state=42
    )

    evaluate("SKLEARN_SEPARATE_42", label, selected)

# Method 2: Independent pandas sampling
for label in ["no", "yes"]:

    positions = class_positions[label]

    selected = pd.Series(positions).sample(
        n=sample_sizes[label],
        random_state=42
    ).to_numpy()

    evaluate("PANDAS_SEPARATE_42", label, selected)

# Method 3: Independent NumPy RandomState sampling
for label in ["no", "yes"]:

    positions = class_positions[label]

    selected = np.random.RandomState(42).choice(
        positions,
        size=sample_sizes[label],
        replace=False
    )

    evaluate("NUMPY_RANDOMSTATE_42", label, selected)

# Method 4: Independent NumPy default_rng sampling
for label in ["no", "yes"]:

    positions = class_positions[label]

    selected = np.random.default_rng(42).choice(
        positions,
        size=sample_sizes[label],
        replace=False
    )

    evaluate("NUMPY_DEFAULT_RNG_42", label, selected)

results_df = pd.DataFrame(results)

print("\n1. INDEPENDENT SAMPLING RESULTS")
display(results_df)

print("\n2. EXACT CLASS MATCHES")
display(
    results_df[results_df["exact_match"] == True]
)

print("\n3. BEST OVERLAP BY METHOD")
display(
    results_df.sort_values(
        ["overlap", "exact_match"],
        ascending=False
    )
)

print("\n4. OFFICIAL CLASS COUNTS")
print("no:", len(actual_no))
print("yes:", len(actual_yes))

print("\n5. SOURCE CLASS COUNTS")
print(source["y"].value_counts())

print("\nCELL 92 COMPLETED")
```

```python

# ============================================================
# ATMIQ CELL 93: FINAL SAMPLE RECONSTRUCTION VALIDATION
# ============================================================

import pandas as pd
import numpy as np

source = bank_source.reset_index(drop=True)
actual_set = set(map(int, actual_positions))

print("=" * 70)
print("CELL 93: FINAL SAMPLE RECONSTRUCTION")
print("=" * 70)

# Independently sample each target class
no_positions = pd.Series(
    np.where(source["y"].to_numpy() == "no")[0]
).sample(n=354, random_state=42)

yes_positions = pd.Series(
    np.where(source["y"].to_numpy() == "yes")[0]
).sample(n=46, random_state=42)

# Preserve the order within each sampled class
reconstructed_positions = pd.concat(
    [no_positions, yes_positions],
    ignore_index=True
).astype(int).tolist()

reconstructed_set = set(reconstructed_positions)

# Verify membership
overlap = actual_set & reconstructed_set
missing = actual_set - reconstructed_set
extra = reconstructed_set - actual_set

print("\n1. SAMPLE SIZE")
print("Official:", len(actual_set))
print("Reconstructed:", len(reconstructed_set))

print("\n2. EXACT MATCH VALIDATION")
print("Overlap:", len(overlap))
print("Missing:", len(missing))
print("Extra:", len(extra))
print("Exact match:", actual_set == reconstructed_set)

assert len(reconstructed_set) == 400
assert actual_set == reconstructed_set

# Create reconstructed source mapping
reconstructed_sample = source.iloc[
    reconstructed_positions
].copy()

reconstructed_sample.insert(
    0, "source_position", reconstructed_positions
)

print("\n3. RECONSTRUCTED TARGET DISTRIBUTION")
display(reconstructed_sample["y"].value_counts())

print("\n4. FIRST 10 RECONSTRUCTED SOURCE RECORDS")
display(reconstructed_sample.head(10))

# Save the verified membership reconstruction
reconstructed_sample.to_csv(
    "/content/drive/MyDrive/ATMIQ_HACKATHON/Results/"
    "verified_400_sample_reconstruction.csv",
    index=False
)

print("\n5. FILE SAVED")
print("verified_400_sample_reconstruction.csv")

print("\nCELL 93 COMPLETED")
```

```python

# ============================================================
# ATMIQ CELL 96: CUSTOMER ID ORDER RECONSTRUCTION
# ============================================================

import pandas as pd
import numpy as np
import ast

print("=" * 75)
print("CELL 96: CUSTOMER ID ORDER RECONSTRUCTION")
print("=" * 75)

# 1. Convert source positions into integers
def extract_position(value):
    if isinstance(value, (list, tuple, np.ndarray)):
        return int(value[0]) if len(value) == 1 else None

    parsed = ast.literal_eval(str(value))

    if isinstance(parsed, (list, tuple)) and len(parsed) == 1:
        return int(parsed[0])

    return None

train_map = mapping.copy()
test_map = test_mapping.copy()

train_map["source_position"] = (
    train_map["source_positions"].apply(extract_position)
)

test_map["source_position"] = (
    test_map["source_positions"].apply(extract_position)
)

# 2. Reconstruct the sample ordering
sample_order = pd.DataFrame({
    "source_position": reconstructed_positions,
    "sample_order": range(len(reconstructed_positions))
})

# 3. Attach sample ordering to known training IDs
train_order = train_map.merge(
    sample_order,
    on="source_position",
    how="left",
    validate="one_to_one"
)

train_order["customer_number"] = (
    train_order["customer_id"].str.extract(r"(\d+)")[0].astype(int)
)

# 4. Attach sample ordering to test records
test_order = test_map.merge(
    sample_order,
    on="source_position",
    how="left",
    validate="one_to_one"
)

print("\n1. MAPPING INTEGRITY")

print("Training mappings:", len(train_order))
print("Training unique positions:",
      train_order["source_position"].nunique())

print("Test mappings:", len(test_order))
print("Test unique positions:",
      test_order["source_position"].nunique())

print("Training positions in reconstructed sample:",
      train_order["sample_order"].notna().sum())

print("Test positions in reconstructed sample:",
      test_order["sample_order"].notna().sum())

# 5. Check whether customer IDs follow sample ordering
print("\n2. CUSTOMER ID VS SAMPLE ORDER")

train_order["expected_id_if_sequential"] = (
    train_order["sample_order"] + 1
)

train_order["id_difference"] = (
    train_order["customer_number"] -
    train_order["expected_id_if_sequential"]
)

print("Exact sequential matches:",
      (train_order["id_difference"] == 0).sum())

print("Correlation:",
      train_order[["customer_number", "sample_order"]].corr().iloc[0, 1])

print("\n3. FIRST 20 TRAINING COMPARISONS")

display(
    train_order[
        ["customer_id", "source_position",
         "sample_order", "id_difference"]
    ].head(20)
)

# 6. Investigate whether IDs increase with sample order
sorted_order = train_order.sort_values("sample_order")

print("\n4. ORDERING CONSISTENCY")

print("Customer ID increases between consecutive sample positions:",
      (sorted_order["customer_number"].diff().dropna() > 0).sum())

print("Customer ID decreases between consecutive sample positions:",
      (sorted_order["customer_number"].diff().dropna() < 0).sum())

# 7. Check whether all test source positions belong to official sample
print("\n5. TEST SAMPLE MEMBERSHIP")

print("Test positions inside official sample:",
      set(test_order["source_position"]) <= actual_set)

print("Test rows with missing sample order:",
      test_order["sample_order"].isna().sum())

print("\n6. FIRST 20 TEST ORDER COMPARISONS")

display(
    test_order[
        ["test_row", "source_position", "sample_order"]
    ].head(20)
)

print("\n7. SAMPLE ORDER DISTRIBUTION")

display(
    train_order["sample_order"].describe()
)

print("\nCELL 96 COMPLETED")
```

```python

import pandas as pd
import numpy as np

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    log_loss
)

print("=" * 75)
print("CELL 100: STAGE 1 MODEL COMPARISON")
print("=" * 75)

file_path = (
    "/content/drive/MyDrive/ATMIQ_HACKATHON/"
    "Dataset/bank_campaign_participant_release.xlsx"
)

train = pd.read_excel(file_path, sheet_name="Stage1_Train")

# Features and target
X = train.drop(columns=["customer_id", "y"])
y = train["y"].map({"no": 0, "yes": 1})

numeric_cols = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_cols = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("\n1. DATASET")
print("Training records:", len(X))
print("Positive cases:", int(y.sum()))
print("Negative cases:", int((y == 0).sum()))
print("Numeric features:", len(numeric_cols))
print("Categorical features:", len(categorical_cols))

# Preprocessing
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_cols),
    ("cat", categorical_transformer, categorical_cols)
])

# Models
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=500,
        class_weight="balanced",
        min_samples_leaf=3,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=500,
        class_weight="balanced",
        min_samples_leaf=3,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.03,
        max_depth=2,
        random_state=42
    )
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []
cv_predictions = {}

print("\n2. CROSS-VALIDATION RESULTS")

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    probabilities = cross_val_predict(
        pipeline,
        X,
        y,
        cv=cv,
        method="predict_proba",
        n_jobs=1
    )[:, 1]

    cv_predictions[name] = probabilities

    results.append({
        "model": name,
        "ROC_AUC": roc_auc_score(y, probabilities),
        "Average_Precision": average_precision_score(
            y, probabilities
        ),
        "Brier_Score": brier_score_loss(y, probabilities),
        "Log_Loss": log_loss(y, probabilities)
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "ROC_AUC",
    ascending=False
).reset_index(drop=True)

print("\n3. MODEL PERFORMANCE TABLE")
print(results_df.to_string(index=False))

print("\n4. BEST ROC-AUC MODEL")
print(results_df.iloc[0].to_string())

print("\n5. CROSS-VALIDATION CONFIGURATION")
print("Folds: 5")
print("Shuffle: True")
print("Random seed: 42")
print("Customer ID excluded: True")
print("Test data used during training: False")

print("\n" + "=" * 75)
print("CELL 100 COMPLETED")
print("=" * 75)
```

```python

import os
import pandas as pd
import numpy as np

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    brier_score_loss,
    log_loss
)

print("=" * 75)
print("CELL 101: FINAL STAGE 1 MODEL AND TEST PREDICTIONS")
print("=" * 75)

base = "/content/drive/MyDrive/ATMIQ_HACKATHON"
file_path = base + "/Dataset/bank_campaign_participant_release.xlsx"

train = pd.read_excel(file_path, sheet_name="Stage1_Train")
test = pd.read_excel(file_path, sheet_name="Stage1_Test")

# 1. Prepare data
X = train.drop(columns=["customer_id", "y"])
y = train["y"].map({"no": 0, "yes": 1})

X_test = test.drop(columns=["customer_id"])

# Ensure identical feature order
X_test = X_test[X.columns]

numeric_cols = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_cols = X.select_dtypes(
    include=["object"]
).columns.tolist()

# 2. Preprocessing
preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_cols),

    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]), categorical_cols)
])

# 3. Selected model
gb_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.03,
    max_depth=2,
    random_state=42
)

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", gb_model)
])

# 4. Out-of-fold validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

oof_prob = cross_val_predict(
    pipeline,
    X,
    y,
    cv=cv,
    method="predict_proba",
    n_jobs=1
)[:, 1]

prevalence = y.mean()
baseline_brier = prevalence * (1 - prevalence)

print("\n1. OUT-OF-FOLD VALIDATION")
print("ROC-AUC:", round(roc_auc_score(y, oof_prob), 6))
print("Average Precision:",
      round(average_precision_score(y, oof_prob), 6))
print("Brier Score:", round(brier_score_loss(y, oof_prob), 6))
print("Log Loss:", round(log_loss(y, oof_prob), 6))

print("\n2. BASELINE COMPARISON")
print("Positive prevalence:", round(prevalence, 6))
print("Constant probability Brier:", round(baseline_brier, 6))
print("Gradient Boosting Brier:",
      round(brier_score_loss(y, oof_prob), 6))

# 5. Top 10% out-of-fold performance
oof_df = pd.DataFrame({
    "actual": y.to_numpy(),
    "oof_probability": oof_prob
})

oof_df = oof_df.sort_values(
    "oof_probability",
    ascending=False
).reset_index(drop=True)

top_count = int(np.ceil(len(oof_df) * 0.10))
top_10 = oof_df.head(top_count)

print("\n3. TOP 10% OUT-OF-FOLD PERFORMANCE")
print("Selected validation records:", top_count)
print("Actual subscribers:", int(top_10["actual"].sum()))
print("Top 10% precision:",
      round(top_10["actual"].mean(), 6))
print("Overall subscriber rate:", round(prevalence, 6))

# 6. Train final model on all 300 rows
print("\n4. FINAL MODEL TRAINING")

pipeline.fit(X, y)

print("Training records:", len(X))
print("Test records:", len(X_test))
print("Features:", X.shape[1])

# 7. Predict test probabilities
test_prob = pipeline.predict_proba(X_test)[:, 1]

predictions = pd.DataFrame({
    "customer_id": test["customer_id"],
    "predicted_probability": test_prob
})

# Official tie-breaking rule
predictions = predictions.sort_values(
    ["predicted_probability", "customer_id"],
    ascending=[False, True],
    kind="mergesort"
).reset_index(drop=True)

predictions.insert(
    0,
    "rank",
    range(1, len(predictions) + 1)
)

# 8. Save results
results_dir = base + "/Results"
os.makedirs(results_dir, exist_ok=True)

all_path = results_dir + "/stage1_test_predictions.csv"
top_path = results_dir + "/stage1_top10_candidates.csv"

predictions.to_csv(all_path, index=False)
predictions.head(10).to_csv(top_path, index=False)

print("\n5. TOP 10 OFFICIAL TEST CANDIDATES")
print(predictions.head(10).to_string(index=False))

print("\n6. PREDICTION DISTRIBUTION")
print(predictions["predicted_probability"].describe())

print("\n7. FILES SAVED")
print(all_path)
print(top_path)

print("\n8. INTEGRITY CHECKS")
print("Predictions:", len(predictions))
print("Unique customer IDs:", predictions["customer_id"].nunique())
print("Missing probabilities:",
      predictions["predicted_probability"].isna().sum())
print("Probability range valid:",
      predictions["predicted_probability"].between(0, 1).all())

print("\n" + "=" * 75)
print("CELL 101 COMPLETED")
print("=" * 75)
```

```python

import os
import pandas as pd
import numpy as np

print("=" * 75)
print("CELL 102: STAGE 1 - STAGE 2 CUSTOMER INTEGRATION")
print("=" * 75)

base = "/content/drive/MyDrive/ATMIQ_HACKATHON"

pred_path = base + "/Results/stage1_test_predictions.csv"
workbook_path = (
    base + "/Dataset/bank_campaign_participant_release.xlsx"
)

# 1. Load files
predictions = pd.read_csv(pred_path)

stage2 = pd.read_excel(
    workbook_path,
    sheet_name="Stage2_Data"
)

train = pd.read_excel(
    workbook_path,
    sheet_name="Stage1_Train"
)

test = pd.read_excel(
    workbook_path,
    sheet_name="Stage1_Test"
)

# 2. Extract official top 10
top10 = predictions.head(10).copy()

print("\n1. TOP 10 CUSTOMER IDS")
print(top10[["rank", "customer_id"]].to_string(index=False))

# 3. Verify membership
top10_ids = set(top10["customer_id"])
stage2_ids = set(stage2["customer_id"])
train_ids = set(train["customer_id"])
test_ids = set(test["customer_id"])

print("\n2. MEMBERSHIP VERIFICATION")
print("Top 10 unique IDs:", len(top10_ids))
print("Top 10 IDs in Stage2:", len(top10_ids & stage2_ids))
print("Top 10 IDs in Stage1_Test:", len(top10_ids & test_ids))
print("Unexpected training IDs:", len(top10_ids & train_ids))

assert len(top10) == 10
assert len(top10_ids) == 10
assert top10_ids.issubset(stage2_ids)
assert top10_ids.issubset(test_ids)
assert len(top10_ids & train_ids) == 0

# 4. Merge
candidate_data = top10.merge(
    stage2,
    on="customer_id",
    how="left",
    validate="one_to_one",
    indicator=True
)

print("\n3. MERGE VERIFICATION")
print(candidate_data["_merge"].value_counts().to_dict())

assert (candidate_data["_merge"] == "both").all()
assert len(candidate_data) == 10

candidate_data = candidate_data.drop(columns=["_merge"])

# 5. Calculate expected net value
candidate_data["expected_net_value"] = (
    candidate_data["predicted_probability"]
    * candidate_data["expected_deposit_value"]
    - candidate_data["contact_cost"]
)

# 6. Display candidate economics
candidate_data = candidate_data.sort_values(
    "rank"
).reset_index(drop=True)

display_columns = [
    "rank",
    "customer_id",
    "predicted_probability",
    "contact_cost",
    "expected_deposit_value",
    "capacity_units",
    "customer_segment",
    "expected_net_value"
]

print("\n4. STAGE 2 CANDIDATE ECONOMICS")
print(
    candidate_data[display_columns].to_string(
        index=False
    )
)

# 7. Summary
print("\n5. ECONOMIC SUMMARY")
print("Total candidate contact cost:",
      candidate_data["contact_cost"].sum())

print("Total candidate capacity:",
      candidate_data["capacity_units"].sum())

print("Total candidate expected net value:",
      round(candidate_data["expected_net_value"].sum(), 4))

print("Positive expected net value candidates:",
      (candidate_data["expected_net_value"] > 0).sum())

# 8. Save
results_dir = base + "/Results"
os.makedirs(results_dir, exist_ok=True)

output_path = (
    results_dir + "/stage2_top10_candidate_economics.csv"
)

candidate_data.to_csv(output_path, index=False)

print("\n6. SAVED FILE")
print(output_path)

print("\n" + "=" * 75)
print("CELL 102 COMPLETED")
print("=" * 75)
```

```python

import os
import itertools
import pandas as pd
import numpy as np

print("=" * 75)
print("CELL 103: EXHAUSTIVE STAGE 2 CLASSICAL OPTIMIZATION")
print("=" * 75)

base = "/content/drive/MyDrive/ATMIQ_HACKATHON"
results_dir = base + "/Results"

input_path = results_dir + "/stage2_top10_candidate_economics.csv"

df = pd.read_csv(input_path)

# Official constraints from README
N = 10
K = 6
BUDGET = 25
CAPACITY = 22

# 1. Input validation
print("\n1. INPUT VALIDATION")

assert len(df) == N
assert df["customer_id"].nunique() == N
assert df["predicted_probability"].between(0, 1).all()

print("Candidates:", len(df))
print("Required selection:", K)
print("Budget:", BUDGET)
print("Capacity:", CAPACITY)

# 2. Enumerate all combinations
all_combinations = list(itertools.combinations(range(N), K))

print("\n2. SEARCH SPACE")
print("Total possible combinations:", len(all_combinations))

results = []

for combination in all_combinations:

    selected = df.iloc[list(combination)]

    total_cost = int(selected["contact_cost"].sum())
    total_capacity = int(selected["capacity_units"].sum())

    total_net_value = float(
        selected["expected_net_value"].sum()
    )

    total_expected_deposit = float(
        (
            selected["predicted_probability"]
            * selected["expected_deposit_value"]
        ).sum()
    )

    feasible = (
        total_cost <= BUDGET
        and total_capacity <= CAPACITY
        and len(selected) == K
    )

    results.append({
        "customer_ids": ",".join(
            selected["customer_id"].tolist()
        ),
        "candidate_ranks": ",".join(
            selected["rank"].astype(str).tolist()
        ),
        "number_selected": len(selected),
        "total_contact_cost": total_cost,
        "total_capacity": total_capacity,
        "expected_deposit_contribution": total_expected_deposit,
        "expected_net_value": total_net_value,
        "feasible": feasible
    })

combos = pd.DataFrame(results)

# 3. Feasibility analysis
feasible = combos[combos["feasible"]].copy()

print("\n3. FEASIBILITY RESULTS")
print("Total combinations:", len(combos))
print("Feasible combinations:", len(feasible))
print("Infeasible combinations:", len(combos) - len(feasible))

assert len(combos) == 210
assert len(feasible) > 0

# 4. Rank feasible combinations
feasible = feasible.sort_values(
    ["expected_net_value", "customer_ids"],
    ascending=[False, True]
).reset_index(drop=True)

print("\n4. TOP 10 FEASIBLE COMBINATIONS")

print(
    feasible.head(10).to_string(index=False)
)

# 5. Identify optimal solution
best_value = feasible["expected_net_value"].max()

optimal_solutions = feasible[
    np.isclose(
        feasible["expected_net_value"],
        best_value,
        rtol=1e-10,
        atol=1e-10
    )
].copy()

best = feasible.iloc[0]

print("\n5. OPTIMAL SOLUTION")

print("Selected customers:", best["customer_ids"])
print("Candidate ranks:", best["candidate_ranks"])
print("Customers selected:", best["number_selected"])
print("Total contact cost:", best["total_contact_cost"])
print("Total capacity:", best["total_capacity"])

print(
    "Expected deposit contribution:",
    round(best["expected_deposit_contribution"], 6)
)

print(
    "Maximum expected net value:",
    round(best["expected_net_value"], 6)
)

print("Number of equally optimal solutions:", len(optimal_solutions))

# 6. Detailed selected customers
selected_ids = best["customer_ids"].split(",")

selected_customers = df[
    df["customer_id"].isin(selected_ids)
].copy()

selected_customers = selected_customers.sort_values("rank")

print("\n6. SELECTED CUSTOMER DETAILS")

columns = [
    "rank",
    "customer_id",
    "predicted_probability",
    "contact_cost",
    "expected_deposit_value",
    "capacity_units",
    "customer_segment",
    "expected_net_value"
]

print(selected_customers[columns].to_string(index=False))

# 7. Final constraint assertions
print("\n7. FINAL CONSTRAINT VERIFICATION")

assert len(selected_customers) == K
assert selected_customers["contact_cost"].sum() <= BUDGET
assert selected_customers["capacity_units"].sum() <= CAPACITY

print("Exactly six customers:", True)
print("Budget satisfied:", True)
print("Capacity satisfied:", True)
print("Objective maximized over feasible combinations:", True)

# 8. Save results
feasible_path = results_dir + "/stage2_all_feasible_combinations.csv"
optimal_path = results_dir + "/stage2_optimal_six_customers.csv"

combos.to_csv(
    results_dir + "/stage2_all_210_combinations.csv",
    index=False
)

feasible.to_csv(feasible_path, index=False)

selected_customers.to_csv(
    optimal_path,
    index=False
)

print("\n8. FILES SAVED")
print(results_dir + "/stage2_all_210_combinations.csv")
print(feasible_path)
print(optimal_path)

print("\n" + "=" * 75)
print("CELL 103 COMPLETED")
print("=" * 75)
```

```python

import os
import pandas as pd
import numpy as np

print("=" * 75)
print("CELL 104: INDEPENDENT STAGE 2 OPTIMUM VERIFICATION")
print("=" * 75)

base = "/content/drive/MyDrive/ATMIQ_HACKATHON"
dataset_path = base + "/Dataset/bank_campaign_participant_release.xlsx"
results_dir = base + "/Results"

# Load original official data and saved results
stage2 = pd.read_excel(
    dataset_path,
    sheet_name="Stage2_Data"
)

predictions = pd.read_csv(
    results_dir + "/stage1_test_predictions.csv"
)

saved_selection = pd.read_csv(
    results_dir + "/stage2_optimal_six_customers.csv"
)

# Official optimal IDs from exhaustive search
optimal_ids = [
    "C0081",
    "C0031",
    "C0079",
    "C0327",
    "C0170",
    "C0349"
]

# 1. Independently reconstruct selected customer economics
print("\n1. REBUILDING SELECTION FROM ORIGINAL DATA")

selected = pd.DataFrame({"customer_id": optimal_ids})

selected = selected.merge(
    predictions[["customer_id", "predicted_probability"]],
    on="customer_id",
    how="left",
    validate="one_to_one"
)

selected = selected.merge(
    stage2,
    on="customer_id",
    how="left",
    validate="one_to_one"
)

assert len(selected) == 6
assert selected["predicted_probability"].notna().all()
assert selected["contact_cost"].notna().all()
assert selected["expected_deposit_value"].notna().all()
assert selected["capacity_units"].notna().all()

selected["expected_deposit_contribution"] = (
    selected["predicted_probability"]
    * selected["expected_deposit_value"]
)

selected["expected_net_value"] = (
    selected["expected_deposit_contribution"]
    - selected["contact_cost"]
)

# 2. Calculate totals independently
total_cost = int(selected["contact_cost"].sum())
total_capacity = int(selected["capacity_units"].sum())
total_deposit_contribution = float(
    selected["expected_deposit_contribution"].sum()
)
total_net_value = float(selected["expected_net_value"].sum())

print("\n2. INDEPENDENTLY RECALCULATED CUSTOMER DETAILS")

columns = [
    "customer_id",
    "predicted_probability",
    "contact_cost",
    "expected_deposit_value",
    "capacity_units",
    "customer_segment",
    "expected_deposit_contribution",
    "expected_net_value"
]

print(selected[columns].to_string(index=False))

# 3. Verify official constraints
print("\n3. OFFICIAL CONSTRAINT VERIFICATION")

print("Selected customers:", len(selected))
print("Total contact cost:", total_cost, "/ 25")
print("Total capacity:", total_capacity, "/ 22")
print("Expected deposit contribution:", round(total_deposit_contribution, 6))
print("Expected net value:", round(total_net_value, 6))

assert len(selected) == 6
assert total_cost <= 25
assert total_capacity <= 22

# 4. Compare with Cell 103 result
print("\n4. COMPARISON WITH CELL 103")

cell103_ids = set(saved_selection["customer_id"])
cell104_ids = set(selected["customer_id"])

same_customers = cell103_ids == cell104_ids

cell103_value = float(saved_selection["expected_net_value"].sum())

print("Same six customer IDs:", same_customers)
print("Cell 103 expected net value:", round(cell103_value, 6))
print("Cell 104 expected net value:", round(total_net_value, 6))
print(
    "Objective values match:",
    np.isclose(cell103_value, total_net_value, atol=1e-8)
)

assert same_customers
assert np.isclose(cell103_value, total_net_value, atol=1e-8)

# 5. Save clean final Stage 2 selection
selected["selection_status"] = "SELECTED"

submission_columns = [
    "customer_id",
    "predicted_probability",
    "contact_cost",
    "expected_deposit_value",
    "capacity_units",
    "customer_segment",
    "expected_deposit_contribution",
    "expected_net_value",
    "selection_status"
]

final_selection = selected[submission_columns].copy()

final_path = (
    results_dir + "/STAGE2_FINAL_OPTIMAL_SELECTION.csv"
)

final_selection.to_csv(final_path, index=False)

# 6. Save a compact summary
summary = pd.DataFrame([{
    "selected_customers": 6,
    "customer_ids": ",".join(optimal_ids),
    "total_contact_cost": total_cost,
    "budget_limit": 25,
    "total_capacity": total_capacity,
    "capacity_limit": 22,
    "expected_deposit_contribution": total_deposit_contribution,
    "expected_net_value": total_net_value,
    "all_constraints_satisfied": True,
    "independent_verification_passed": True
}])

summary_path = (
    results_dir + "/STAGE2_FINAL_OPTIMIZATION_SUMMARY.csv"
)

summary.to_csv(summary_path, index=False)

print("\n5. FINAL FILES SAVED")
print(final_path)
print(summary_path)

print("\n" + "=" * 75)
print("CELL 104 COMPLETED")
print("STAGE 2 INDEPENDENT VERIFICATION PASSED")
print("=" * 75)
```

```python

import os
import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 105: STAGE 3 - PROBABILITY OF AT LEAST 3 SUBSCRIPTIONS")
print("=" * 75)

base = "/content/drive/MyDrive/ATMIQ_HACKATHON"
results_dir = base + "/Results"

# 1. Load the independently verified six customers
selection_path = (
    results_dir + "/STAGE2_FINAL_OPTIMAL_SELECTION.csv"
)

selected = pd.read_csv(selection_path)

assert len(selected) == 6
assert selected["customer_id"].nunique() == 6

probabilities = selected[
    "predicted_probability"
].to_numpy(dtype=float)

assert np.all((probabilities >= 0) & (probabilities <= 1))

print("\n1. SELECTED CUSTOMER PROBABILITIES")
print(
    selected[
        ["customer_id", "predicted_probability"]
    ].to_string(index=False)
)

# 2. Exact Poisson-binomial calculation using dynamic programming
# distribution[k] = probability of exactly k subscriptions

distribution = np.array([1.0])

for p in probabilities:
    updated = np.zeros(len(distribution) + 1)

    updated[:-1] += distribution * (1 - p)
    updated[1:] += distribution * p

    distribution = updated

# 3. Display probability for exactly 0 through 6 subscriptions
print("\n2. PROBABILITY DISTRIBUTION")

distribution_table = pd.DataFrame({
    "number_of_subscribers": np.arange(7),
    "probability": distribution,
    "percentage": distribution * 100
})

print(
    distribution_table.to_string(
        index=False,
        formatters={
            "probability": "{:.8f}".format,
            "percentage": "{:.4f}%".format
        }
    )
)

# 4. Calculate Stage 3 target
probability_at_least_3 = float(distribution[3:].sum())
probability_below_3 = float(distribution[:3].sum())

expected_subscribers = float(probabilities.sum())

print("\n3. STAGE 3 SUCCESS PROBABILITY")

print(
    "Probability of at least 3 subscriptions:",
    round(probability_at_least_3, 8)
)

print(
    "Percentage chance of at least 3 subscriptions:",
    round(probability_at_least_3 * 100, 4),
    "%"
)

print(
    "Probability of fewer than 3 subscriptions:",
    round(probability_below_3 * 100, 4),
    "%"
)

print(
    "Expected number of subscribers:",
    round(expected_subscribers, 6),
    "out of 6"
)

# 5. Monte Carlo cross-check
rng = np.random.default_rng(42)
simulations = 1_000_000

random_outcomes = (
    rng.random((simulations, len(probabilities)))
    < probabilities
)

subscriber_counts = random_outcomes.sum(axis=1)

monte_carlo_probability = float(
    np.mean(subscriber_counts >= 3)
)

print("\n4. MONTE CARLO CROSS-CHECK")
print("Simulations:", simulations)
print(
    "Monte Carlo probability:",
    round(monte_carlo_probability * 100, 4),
    "%"
)

print(
    "Absolute difference from exact result:",
    round(
        abs(monte_carlo_probability - probability_at_least_3),
        8
    )
)

# 6. Save detailed results
distribution_path = (
    results_dir + "/STAGE3_SUBSCRIPTION_PROBABILITY_DISTRIBUTION.csv"
)

summary_path = (
    results_dir + "/STAGE3_SUCCESS_PROBABILITY_SUMMARY.csv"
)

distribution_table.to_csv(distribution_path, index=False)

summary = pd.DataFrame([{
    "selected_customer_count": 6,
    "success_threshold": 3,
    "expected_subscribers": expected_subscribers,
    "probability_at_least_3": probability_at_least_3,
    "probability_at_least_3_percent": probability_at_least_3 * 100,
    "probability_fewer_than_3": probability_below_3,
    "monte_carlo_probability": monte_carlo_probability,
    "monte_carlo_simulations": simulations,
    "method": "Exact Poisson-binomial",
    "assumption": "Independent customer subscription outcomes"
}])

summary.to_csv(summary_path, index=False)

print("\n5. FILES SAVED")
print(distribution_path)
print(summary_path)

print("\n6. IMPORTANT INTERPRETATION")
print(
    "This is a model-based estimate conditional on the six "
    "predicted probabilities and the independence assumption."
)
print(
    "It is not a guarantee of actual campaign outcomes. "
    "Prediction calibration and dependence between customers "
    "may affect real-world accuracy."
)

print("\n" + "=" * 75)
print("CELL 105 COMPLETED")
print("=" * 75)
```

```python

# ============================================================
# CELL 105: STAGE 1 — AUTHORITATIVE PREDICTION AUDIT
# Read-only verification. Does not modify existing results.
# ============================================================

import os
import numpy as np
import pandas as pd

BASE_DIR = "/content/drive/MyDrive/ATMIQ_HACKATHON"
DATA_DIR = os.path.join(BASE_DIR, "Dataset")
RESULTS_DIR = os.path.join(BASE_DIR, "Results")

WORKBOOK = os.path.join(
    DATA_DIR, "bank_campaign_participant_release.xlsx"
)

PRED_PATH = os.path.join(
    RESULTS_DIR, "stage1_test_predictions.csv"
)

TOP10_PATH = os.path.join(
    RESULTS_DIR, "stage1_top10_candidates.csv"
)

print("=" * 75)
print("STAGE 1 — AUTHORITATIVE PREDICTION AUDIT")
print("=" * 75)

# 1. Load official test data and saved predictions
official_test = pd.read_excel(
    WORKBOOK, sheet_name="Stage1_Test"
)

pred = pd.read_csv(PRED_PATH)
top10 = pd.read_csv(TOP10_PATH)

print("\n1. FILE SHAPES")
print("Official test:", official_test.shape)
print("Saved predictions:", pred.shape)
print("Saved top 10:", top10.shape)

# 2. Check customer ID integrity
official_ids = set(official_test["customer_id"])
prediction_ids = set(pred["customer_id"])

print("\n2. CUSTOMER ID INTEGRITY")
print("Official test IDs:", len(official_ids))
print("Prediction IDs:", len(prediction_ids))
print("Duplicate official IDs:",
      official_test["customer_id"].duplicated().sum())
print("Duplicate prediction IDs:",
      pred["customer_id"].duplicated().sum())
print("Missing prediction IDs:",
      len(official_ids - prediction_ids))
print("Unexpected prediction IDs:",
      len(prediction_ids - official_ids))
print("Exact ID match:",
      official_ids == prediction_ids)

# 3. Check prediction probabilities
print("\n3. PROBABILITY VALIDATION")

prob_ok = (
    pred["predicted_probability"].notna().all()
    and np.isfinite(pred["predicted_probability"]).all()
    and pred["predicted_probability"].between(0, 1).all()
)

print("All probabilities valid:", prob_ok)
print("Minimum probability:",
      pred["predicted_probability"].min())
print("Maximum probability:",
      pred["predicted_probability"].max())

# 4. Independently reconstruct ranking
reconstructed = pred.sort_values(
    "predicted_probability",
    ascending=False,
    kind="stable"
).reset_index(drop=True)

reconstructed["expected_rank"] = (
    np.arange(1, len(reconstructed) + 1)
)

if "rank" in pred.columns:
    rank_match = np.array_equal(
        pred.sort_values("rank")["customer_id"].to_numpy(),
        reconstructed["customer_id"].to_numpy()
    )
    print("\n4. RANKING VALIDATION")
    print("Saved ranking matches probability ranking:",
          rank_match)
else:
    rank_match = False
    print("\n4. RANKING VALIDATION")
    print("Rank column missing.")

# 5. Verify saved Top 10
expected_top10 = reconstructed.head(10)[
    ["customer_id", "predicted_probability"]
].reset_index(drop=True)

saved_top10 = top10[
    ["customer_id", "predicted_probability"]
].reset_index(drop=True)

top10_match = (
    saved_top10["customer_id"].tolist()
    == expected_top10["customer_id"].tolist()
    and np.allclose(
        saved_top10["predicted_probability"],
        expected_top10["predicted_probability"]
    )
)

print("\n5. TOP 10 VALIDATION")
print("Saved Top 10 matches reconstructed Top 10:",
      top10_match)

print("\nRECONSTRUCTED TOP 10")
print(expected_top10.to_string(index=False))

# 6. Final audit summary
print("\n6. FINAL AUDIT SUMMARY")

checks = {
    "Exactly 100 official test customers":
        len(official_test) == 100,
    "Exactly 100 saved predictions":
        len(pred) == 100,
    "All customer IDs match":
        official_ids == prediction_ids,
    "No duplicate prediction IDs":
        pred["customer_id"].is_unique,
    "All probabilities valid":
        prob_ok,
    "Saved ranking matches":
        rank_match,
    "Saved Top 10 matches":
        top10_match
}

for name, result in checks.items():
    print(("PASS" if result else "FAIL"), "-", name)

print("\nOVERALL AUDIT:",
      "PASS" if all(checks.values()) else "REQUIRES INVESTIGATION")

print("=" * 75)
print("AUDIT COMPLETE — NO FILES MODIFIED")
print("=" * 75)
```

```python

# ============================================================
# CELL 107 - INDEPENDENT CLASSICAL OPTIMUM VERIFICATION
# Enumerates all 210 possible six-customer combinations
# READ ONLY - DOES NOT MODIFY EXISTING RESULTS
# ============================================================

import os
import itertools
import numpy as np
import pandas as pd

RESULTS_DIR = "/content/drive/MyDrive/ATMIQ_HACKATHON/Results"

# Load saved classical optimum
classical_path = os.path.join(
    RESULTS_DIR,
    "STAGE2_FINAL_OPTIMAL_SELECTION.csv"
)

saved_classical = pd.read_csv(classical_path)

# Use verified Cell 106 input
data = qaoa_input.copy().reset_index(drop=True)

# Optimization constraints
K = 6
BUDGET_LIMIT = 25
CAPACITY_LIMIT = 22

# Enumerate every possible combination
all_results = []

for combo in itertools.combinations(range(len(data)), K):

    selected = data.iloc[list(combo)]

    total_cost = selected["contact_cost"].sum()
    total_capacity = selected["capacity_units"].sum()
    total_value = selected["expected_net_value"].sum()

    feasible = (
        total_cost <= BUDGET_LIMIT
        and total_capacity <= CAPACITY_LIMIT
    )

    all_results.append({
        "customer_ids": tuple(
            selected["customer_id"].tolist()
        ),
        "total_cost": total_cost,
        "total_capacity": total_capacity,
        "expected_net_value": total_value,
        "feasible": feasible
    })

results = pd.DataFrame(all_results)

feasible_results = results[
    results["feasible"]
].copy()

assert len(results) == 210

# Find maximum feasible objective
best_value = feasible_results[
    "expected_net_value"
].max()

optimal_solutions = feasible_results[
    np.isclose(
        feasible_results["expected_net_value"],
        best_value,
        atol=1e-9
    )
]

best = optimal_solutions.iloc[0]

# Compare saved classical IDs
saved_ids = set(
    saved_classical["customer_id"].astype(str)
)

reconstructed_ids = set(best["customer_ids"])

saved_selection_value = data[
    data["customer_id"].astype(str).isin(saved_ids)
]["expected_net_value"].sum()

saved_selection_cost = data[
    data["customer_id"].astype(str).isin(saved_ids)
]["contact_cost"].sum()

saved_selection_capacity = data[
    data["customer_id"].astype(str).isin(saved_ids)
]["capacity_units"].sum()

print("=" * 75)
print("CELL 107 - INDEPENDENT OPTIMUM VERIFICATION")
print("=" * 75)

print("\n1. COMBINATION ENUMERATION")
print("Total combinations:", len(results))
print("Feasible combinations:", len(feasible_results))
print("Infeasible combinations:",
      len(results) - len(feasible_results))

print("\n2. INDEPENDENTLY OPTIMAL SOLUTION")
print("Customer IDs:", list(best["customer_ids"]))
print("Total cost:", best["total_cost"])
print("Total capacity:", best["total_capacity"])
print("Expected net value:",
      round(best["expected_net_value"], 9))

print("\n3. SAVED CLASSICAL SOLUTION")
print("Customer IDs:", sorted(saved_ids))
print("Total cost:", saved_selection_cost)
print("Total capacity:", saved_selection_capacity)
print("Expected net value:",
      round(saved_selection_value, 9))

print("\n4. COMPARISON")
print("Same selected customers:",
      saved_ids == reconstructed_ids)

print("Same optimal objective:",
      np.isclose(
          saved_selection_value,
          best_value,
          atol=1e-9
      ))

print("Saved solution feasible:",
      len(saved_ids) == K
      and saved_selection_cost <= BUDGET_LIMIT
      and saved_selection_capacity <= CAPACITY_LIMIT)

print("\n5. FINAL VERIFICATION")

checks = {
    "210 combinations evaluated": len(results) == 210,
    "Feasible solutions exist": len(feasible_results) > 0,
    "Exactly six customers selected": len(reconstructed_ids) == K,
    "Budget satisfied": best["total_cost"] <= BUDGET_LIMIT,
    "Capacity satisfied": best["total_capacity"] <= CAPACITY_LIMIT,
    "Saved solution matches optimum": saved_ids == reconstructed_ids
}

for name, passed in checks.items():
    print(("PASS" if passed else "FAIL"), "-", name)

print("\nOVERALL:",
      "PASS" if all(checks.values()) else "INVESTIGATION REQUIRED")

print("=" * 75)
print("CELL 107 COMPLETE - NO FILES MODIFIED")
print("=" * 75)
```

```python

# ============================================================
# CELL 111 - QUANTUM OPTIMIZATION DATA PREPARATION
# CORRECTED VERSION
# ORIGINAL DATA PRESERVED
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("CELL 111 - QUANTUM OPTIMIZATION DATA PREPARATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. VERIFY INPUT DATA
# ------------------------------------------------------------

assert "top10" in globals(), "top10 dataframe missing"
assert "stage2" in globals(), "stage2 dataframe missing"

print("\n1. ORIGINAL DATA SCHEMAS")

print("Top10 columns:", top10.columns.tolist())
print("Stage2 columns:", stage2.columns.tolist())

# ------------------------------------------------------------
# 2. VERIFY REQUIRED COLUMNS
# ------------------------------------------------------------

top10_required = [
    "customer_id",
    "predicted_probability"
]

stage2_required = [
    "customer_id",
    "contact_cost",
    "expected_deposit_value",
    "capacity_units",
    "customer_segment"
]

assert all(c in top10.columns for c in top10_required)
assert all(c in stage2.columns for c in stage2_required)

assert top10["customer_id"].is_unique
assert stage2["customer_id"].is_unique

# ------------------------------------------------------------
# 3. MERGE TOP 10 WITH STAGE 2
# ------------------------------------------------------------

quantum_candidates = pd.merge(
    top10[top10_required],
    stage2[stage2_required],
    on="customer_id",
    how="inner",
    validate="one_to_one"
)

assert len(quantum_candidates) == 10, (
    f"Expected 10 matched candidates, "
    f"found {len(quantum_candidates)}"
)

# ------------------------------------------------------------
# 4. CALCULATE EXPECTED NET VALUE
# ------------------------------------------------------------

quantum_candidates["expected_net_value"] = (
    quantum_candidates["predicted_probability"]
    * quantum_candidates["expected_deposit_value"]
    - quantum_candidates["contact_cost"]
)

# ------------------------------------------------------------
# 5. SORT CANDIDATES
# ------------------------------------------------------------

quantum_candidates = (
    quantum_candidates
    .sort_values(
        "predicted_probability",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 6. CREATE QUANTUM ARRAYS
# ------------------------------------------------------------

quantum_values = quantum_candidates[
    "expected_net_value"
].to_numpy(dtype=float)

quantum_costs = quantum_candidates[
    "contact_cost"
].to_numpy(dtype=float)

quantum_capacities = quantum_candidates[
    "capacity_units"
].to_numpy(dtype=float)

N_QUBITS = len(quantum_candidates)

K_SELECTED = 6
BUDGET = 25
CAPACITY_LIMIT = 22

# ------------------------------------------------------------
# 7. DATA VALIDATION
# ------------------------------------------------------------

numeric_columns = [
    "predicted_probability",
    "contact_cost",
    "expected_deposit_value",
    "capacity_units",
    "expected_net_value"
]

assert quantum_candidates[numeric_columns].notna().all().all()
assert np.isfinite(
    quantum_candidates[numeric_columns].to_numpy(dtype=float)
).all()

assert (quantum_candidates["predicted_probability"] >= 0).all()
assert (quantum_candidates["predicted_probability"] <= 1).all()

# ------------------------------------------------------------
# 8. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n2. MATCHED QUANTUM CANDIDATES")

print(quantum_candidates.to_string(index=False))

print("\n3. QUANTUM PARAMETERS")

print("Number of qubits:", N_QUBITS)
print("Required selections:", K_SELECTED)
print("Budget limit:", BUDGET)
print("Capacity limit:", CAPACITY_LIMIT)

print("\n4. QUANTUM ARRAY SHAPES")

print("Values:", quantum_values.shape)
print("Costs:", quantum_costs.shape)
print("Capacities:", quantum_capacities.shape)

print("\n5. OBJECTIVE")

print("Maximize expected_net_value")
print("Subject to:")
print("Exactly 6 customers selected")
print("Total contact cost <= 25")
print("Total capacity <= 22")

print("\n6. DATA INTEGRITY")

print("Candidate count:", len(quantum_candidates))
print("Unique customer IDs:",
      quantum_candidates["customer_id"].nunique())
print("Missing values:",
      quantum_candidates.isna().sum().sum())

print("\nCELL 111 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 112 - QUBO HAMILTONIAN CONSTRUCTION
# EXACT CARDINALITY + BUDGET + CAPACITY CONSTRAINTS
# ============================================================

import numpy as np

print("=" * 70)
print("CELL 112 - QUBO HAMILTONIAN CONSTRUCTION")
print("=" * 70)

# ------------------------------------------------------------
# 1. DEFINE QUBO VARIABLES
# ------------------------------------------------------------

N_CUSTOMERS = 10
N_SLACK_BUDGET = 5
N_SLACK_CAPACITY = 5

N_QUBITS_TOTAL = (
    N_CUSTOMERS
    + N_SLACK_BUDGET
    + N_SLACK_CAPACITY
)

print("\n1. QUBIT ALLOCATION")

print("Customer selection qubits:", N_CUSTOMERS)
print("Budget slack qubits:", N_SLACK_BUDGET)
print("Capacity slack qubits:", N_SLACK_CAPACITY)
print("Total qubits:", N_QUBITS_TOTAL)

# Variable indices:
# 0-9   : Customer selection
# 10-14 : Budget slack
# 15-19 : Capacity slack

# ------------------------------------------------------------
# 2. INITIALIZE QUBO
# ------------------------------------------------------------

qubo = {}
qubo_constant = 0.0

def add_qubo_term(i, j, coefficient):
    key = (min(i, j), max(i, j))
    qubo[key] = qubo.get(key, 0.0) + float(coefficient)

def add_square_constraint(constant, coefficients, penalty):
    """
    Add penalty * (constant + sum(a_i*x_i))^2

    Binary property:
    x_i^2 = x_i
    """

    global qubo_constant

    qubo_constant += penalty * constant**2

    items = list(coefficients.items())

    for i, ai in items:
        linear = penalty * (
            ai**2 + 2 * constant * ai
        )

        add_qubo_term(i, i, linear)

    for p in range(len(items)):
        i, ai = items[p]

        for q in range(p + 1, len(items)):
            j, aj = items[q]

            quadratic = penalty * 2 * ai * aj

            add_qubo_term(i, j, quadratic)

# ------------------------------------------------------------
# 3. PENALTY COEFFICIENT
# ------------------------------------------------------------

PENALTY = 100.0

print("\n2. PENALTY COEFFICIENT:", PENALTY)

# ------------------------------------------------------------
# 4. OBJECTIVE FUNCTION
# ------------------------------------------------------------

# Minimize negative expected net value.
# Equivalent to maximizing expected net value.

for i, value in enumerate(quantum_values):
    add_qubo_term(i, i, -value)

print("\n3. OBJECTIVE FUNCTION ADDED")

# ------------------------------------------------------------
# 5. EXACT CUSTOMER COUNT CONSTRAINT
# ------------------------------------------------------------

cardinality_coefficients = {
    i: 1.0 for i in range(N_CUSTOMERS)
}

add_square_constraint(
    constant=-K_SELECTED,
    coefficients=cardinality_coefficients,
    penalty=PENALTY
)

print("\n4. CARDINALITY CONSTRAINT ADDED")
print("Selected customers = 6")

# ------------------------------------------------------------
# 6. BUDGET CONSTRAINT WITH SLACK VARIABLES
# ------------------------------------------------------------

budget_coefficients = {
    i: float(quantum_costs[i])
    for i in range(N_CUSTOMERS)
}

for bit in range(N_SLACK_BUDGET):
    index = N_CUSTOMERS + bit
    budget_coefficients[index] = float(2**bit)

add_square_constraint(
    constant=-BUDGET,
    coefficients=budget_coefficients,
    penalty=PENALTY
)

print("\n5. BUDGET CONSTRAINT ADDED")
print("Total contact cost <= 25")
print("Budget slack encoding: 1, 2, 4, 8, 16")

# ------------------------------------------------------------
# 7. CAPACITY CONSTRAINT WITH SLACK VARIABLES
# ------------------------------------------------------------

capacity_coefficients = {
    i: float(quantum_capacities[i])
    for i in range(N_CUSTOMERS)
}

for bit in range(N_SLACK_CAPACITY):
    index = (
        N_CUSTOMERS
        + N_SLACK_BUDGET
        + bit
    )

    capacity_coefficients[index] = float(2**bit)

add_square_constraint(
    constant=-CAPACITY_LIMIT,
    coefficients=capacity_coefficients,
    penalty=PENALTY
)

print("\n6. CAPACITY CONSTRAINT ADDED")
print("Total capacity <= 22")
print("Capacity slack encoding: 1, 2, 4, 8, 16")

# ------------------------------------------------------------
# 8. REMOVE NUMERICAL ZERO TERMS
# ------------------------------------------------------------

qubo = {
    key: value
    for key, value in qubo.items()
    if abs(value) > 1e-10
}

# ------------------------------------------------------------
# 9. INDEPENDENTLY VERIFY CLASSICAL OPTIMUM
# ------------------------------------------------------------

classical_ids = [
    "C0081",
    "C0031",
    "C0079",
    "C0327",
    "C0170",
    "C0349"
]

selected_indices = [
    quantum_candidates.index[
        quantum_candidates["customer_id"] == customer_id
    ][0]
    for customer_id in classical_ids
]

x = np.zeros(N_QUBITS_TOTAL, dtype=int)

for i in selected_indices:
    x[i] = 1

# Budget slack = 25 - selected contact cost
selected_cost = int(
    sum(quantum_costs[i] for i in selected_indices)
)

budget_slack = BUDGET - selected_cost

# Capacity slack = 22 - selected capacity
selected_capacity = int(
    sum(quantum_capacities[i] for i in selected_indices)
)

capacity_slack = CAPACITY_LIMIT - selected_capacity

for bit in range(N_SLACK_BUDGET):
    x[N_CUSTOMERS + bit] = (
        budget_slack >> bit
    ) & 1

for bit in range(N_SLACK_CAPACITY):
    index = N_CUSTOMERS + N_SLACK_BUDGET + bit

    x[index] = (
        capacity_slack >> bit
    ) & 1

# ------------------------------------------------------------
# 10. CALCULATE QUBO ENERGY
# ------------------------------------------------------------

qubo_energy = sum(
    coefficient * x[i] * x[j]
    for (i, j), coefficient in qubo.items()
)

full_energy = qubo_energy + qubo_constant

# ------------------------------------------------------------
# 11. VERIFY RESULTS
# ------------------------------------------------------------

print("\n7. QUBO MODEL SUMMARY")

print("Total binary variables:", N_QUBITS_TOTAL)
print("Number of nonzero QUBO terms:", len(qubo))
print("Penalty coefficient:", PENALTY)
print("QUBO constant offset:", qubo_constant)

print("\n8. CLASSICAL OPTIMUM ENCODED")

print("Selected customer IDs:", classical_ids)
print("Selected count:", len(selected_indices))
print("Total contact cost:", selected_cost)
print("Total capacity:", selected_capacity)
print("Budget slack:", budget_slack)
print("Capacity slack:", capacity_slack)

print("\n9. QUBO ENERGY VERIFICATION")

print("QUBO variable energy:", qubo_energy)
print("Full QUBO energy:", full_energy)
print("Expected net value:", quantum_values[selected_indices].sum())

assert len(selected_indices) == 6
assert selected_cost <= BUDGET
assert selected_capacity <= CAPACITY_LIMIT
assert budget_slack >= 0
assert capacity_slack >= 0
assert abs(
    full_energy + quantum_values[selected_indices].sum()
) < 1e-8

print("\nCELL 112 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 113 - QUBO TO ISING HAMILTONIAN
# QISKIT 2.6.2 COMPATIBLE
# ============================================================

import numpy as np
from qiskit.quantum_info import SparsePauliOp, Statevector

print("=" * 70)
print("CELL 113 - QUBO TO ISING HAMILTONIAN")
print("=" * 70)

# ------------------------------------------------------------
# 1. INITIALIZE ISING COEFFICIENTS
# ------------------------------------------------------------

n = N_QUBITS_TOTAL

ising_terms = {}

def add_ising_term(z_indices, coefficient):
    """
    Add coefficient multiplied by product of Z operators.
    Empty indices represent the identity operator.
    """

    key = tuple(sorted(z_indices))

    ising_terms[key] = (
        ising_terms.get(key, 0.0)
        + float(coefficient)
    )

# ------------------------------------------------------------
# 2. ADD QUBO CONSTANT
# ------------------------------------------------------------

add_ising_term((), qubo_constant)

# ------------------------------------------------------------
# 3. CONVERT LINEAR TERMS
# ------------------------------------------------------------

# x_i = (1 - Z_i) / 2

for (i, j), coefficient in qubo.items():

    if i == j:

        add_ising_term(
            (),
            coefficient / 2
        )

        add_ising_term(
            (i,),
            -coefficient / 2
        )

# ------------------------------------------------------------
# 4. CONVERT QUADRATIC TERMS
# ------------------------------------------------------------

# x_i*x_j =
# (1 - Z_i - Z_j + Z_i*Z_j) / 4

    else:

        add_ising_term(
            (),
            coefficient / 4
        )

        add_ising_term(
            (i,),
            -coefficient / 4
        )

        add_ising_term(
            (j,),
            -coefficient / 4
        )

        add_ising_term(
            (i, j),
            coefficient / 4
        )

# ------------------------------------------------------------
# 5. BUILD SPARSE PAULI OPERATOR
# ------------------------------------------------------------

pauli_list = []

for z_indices, coefficient in ising_terms.items():

    if abs(coefficient) < 1e-10:
        continue

    label = ["I"] * n

    for index in z_indices:
        label[n - 1 - index] = "Z"

    pauli_list.append(
        ("".join(label), coefficient)
    )

ising_hamiltonian = SparsePauliOp.from_list(
    pauli_list
)

# ------------------------------------------------------------
# 6. REMOVE NUMERICAL ZERO TERMS
# ------------------------------------------------------------

ising_hamiltonian = ising_hamiltonian.simplify(
    atol=1e-10
)

# ------------------------------------------------------------
# 7. VERIFY QUBIT COUNT
# ------------------------------------------------------------

print("\n1. HAMILTONIAN INFORMATION")

print("Number of qubits:", ising_hamiltonian.num_qubits)
print("Number of Pauli terms:", len(ising_hamiltonian))
print("Operator type:", type(ising_hamiltonian).__name__)

# ------------------------------------------------------------
# 8. VERIFY CLASSICAL OPTIMUM ENERGY
# ------------------------------------------------------------

basis_index = sum(
    int(x[i]) << i
    for i in range(n)
)

optimal_state = Statevector.from_int(
    basis_index,
    dims=2**n
)

quantum_energy = np.real(
    optimal_state.expectation_value(
        ising_hamiltonian
    )
)

print("\n2. CLASSICAL OPTIMUM VERIFICATION")

print("Encoded basis index:", basis_index)
print("Expected QUBO energy:", full_energy)
print("Ising Hamiltonian energy:", quantum_energy)

print(
    "Absolute difference:",
    abs(quantum_energy - full_energy)
)

assert ising_hamiltonian.num_qubits == 20

assert np.isclose(
    quantum_energy,
    full_energy,
    atol=1e-7
)

print("\n3. ISING HAMILTONIAN STATUS")

print("QUBO-to-Ising conversion: SUCCESS")
print("Classical optimum energy verification: SUCCESS")

print("\nCELL 113 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 114 - QAOA QUANTUM CIRCUIT CONSTRUCTION
# QISKIT 2.6.2 + AER 0.18.2
# ============================================================

import numpy as np

from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

print("=" * 70)
print("CELL 114 - QAOA QUANTUM CIRCUIT CONSTRUCTION")
print("=" * 70)

# ------------------------------------------------------------
# 1. VERIFY ISING HAMILTONIAN
# ------------------------------------------------------------

assert "ising_hamiltonian" in globals()
assert ising_hamiltonian.num_qubits == 20

N = ising_hamiltonian.num_qubits

print("\n1. HAMILTONIAN VERIFIED")
print("Number of qubits:", N)
print("Number of Pauli terms:", len(ising_hamiltonian))

# ------------------------------------------------------------
# 2. INITIALIZE QAOA PARAMETERS
# ------------------------------------------------------------

QAOA_DEPTH = 1

gamma = 0.10
beta = 0.10

print("\n2. QAOA PARAMETERS")
print("QAOA depth:", QAOA_DEPTH)
print("Initial gamma:", gamma)
print("Initial beta:", beta)

# ------------------------------------------------------------
# 3. CREATE QAOA CIRCUIT
# ------------------------------------------------------------

qaoa_circuit = QuantumCircuit(N)

# Initial uniform superposition

for qubit in range(N):
    qaoa_circuit.h(qubit)

qaoa_circuit.barrier()

# ------------------------------------------------------------
# 4. COST HAMILTONIAN UNITARY
# ------------------------------------------------------------

# H_cost = constant + sum(h_i Z_i) + sum(J_ij Z_i Z_j)
#
# exp(-i gamma h_i Z_i)  -> RZ(2*gamma*h_i)
# exp(-i gamma J_ij ZiZj) -> RZZ(2*gamma*J_ij)
#
# Constant term is a global phase and can be omitted.

for indices, coefficient in ising_terms.items():

    if len(indices) == 1:

        i = indices[0]

        qaoa_circuit.rz(
            2 * gamma * coefficient,
            i
        )

    elif len(indices) == 2:

        i, j = indices

        qaoa_circuit.rzz(
            2 * gamma * coefficient,
            i,
            j
        )

    elif len(indices) > 2:

        raise ValueError(
            "Unexpected higher-order Ising term."
        )

qaoa_circuit.barrier()

# ------------------------------------------------------------
# 5. MIXER UNITARY
# ------------------------------------------------------------

# H_mixer = sum(X_i)
# exp(-i beta X_i) -> RX(2*beta)

for qubit in range(N):

    qaoa_circuit.rx(
        2 * beta,
        qubit
    )

qaoa_circuit.barrier()

# ------------------------------------------------------------
# 6. ADD MEASUREMENTS
# ------------------------------------------------------------

qaoa_circuit.measure_all()

print("\n3. CIRCUIT CONSTRUCTION")

print("Circuit qubits:", qaoa_circuit.num_qubits)
print("Circuit depth:", qaoa_circuit.depth())
print("Circuit operations:", qaoa_circuit.count_ops())

# ------------------------------------------------------------
# 7. INITIALIZE AER SIMULATOR
# ------------------------------------------------------------

simulator = AerSimulator(
    method="matrix_product_state"
)

SHOTS = 1024

compiled_circuit = transpile(
    qaoa_circuit,
    simulator,
    optimization_level=1
)

print("\n4. SIMULATOR CONFIGURATION")

print("Backend:", simulator.name)
print("Simulation method: Matrix Product State")
print("Shots:", SHOTS)

# ------------------------------------------------------------
# 8. EXECUTE QUANTUM CIRCUIT
# ------------------------------------------------------------

job = simulator.run(
    compiled_circuit,
    shots=SHOTS,
    seed_simulator=42
)

result = job.result()

counts = result.get_counts()

# ------------------------------------------------------------
# 9. VERIFY MEASUREMENTS
# ------------------------------------------------------------

total_shots = sum(counts.values())

print("\n5. QUANTUM EXECUTION RESULTS")

print("Total measurement shots:", total_shots)
print("Unique measured bitstrings:", len(counts))

print("\nTOP 10 MEASUREMENT OUTCOMES")

for bitstring, frequency in sorted(
    counts.items(),
    key=lambda item: item[1],
    reverse=True
)[:10]:

    print(
        bitstring,
        "Frequency:", frequency,
        "Probability:", round(frequency / total_shots, 5)
    )

assert total_shots == SHOTS
assert len(counts) > 0

# ------------------------------------------------------------
# 10. SAVE CIRCUIT RESULTS
# ------------------------------------------------------------

qaoa_initial_counts = counts

print("\n6. EXECUTION STATUS")

print("QAOA circuit construction: SUCCESS")
print("Aer simulation: SUCCESS")
print("Measurement verification: SUCCESS")

print("\nCELL 114 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 115 - QUANTUM MEASUREMENT DECODING
# FEASIBILITY AND OBJECTIVE VALIDATION
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("CELL 115 - QUANTUM MEASUREMENT DECODING")
print("=" * 70)

assert "qaoa_initial_counts" in globals()

decoded_results = []

# Qiskit measurement strings are displayed
# with the highest classical bit on the left.

for bitstring, frequency in qaoa_initial_counts.items():

    bits = np.array(
        [int(b) for b in bitstring.replace(" ", "")][::-1]
    )

    if len(bits) != N_QUBITS_TOTAL:
        continue

    customer_selection = bits[:N_CUSTOMERS]

    selected_indices = np.where(
        customer_selection == 1
    )[0]

    selected_count = len(selected_indices)

    selected_cost = float(
        quantum_costs[selected_indices].sum()
    )

    selected_capacity = float(
        quantum_capacities[selected_indices].sum()
    )

    expected_value = float(
        quantum_values[selected_indices].sum()
    )

    cardinality_ok = selected_count == K_SELECTED
    budget_ok = selected_cost <= BUDGET
    capacity_ok = selected_capacity <= CAPACITY_LIMIT

    feasible = (
        cardinality_ok
        and budget_ok
        and capacity_ok
    )

    customer_ids = quantum_candidates.iloc[
        selected_indices
    ]["customer_id"].tolist()

    decoded_results.append({
        "bitstring": bitstring,
        "frequency": frequency,
        "probability": frequency / sum(
            qaoa_initial_counts.values()
        ),
        "selected_count": selected_count,
        "contact_cost": selected_cost,
        "capacity": selected_capacity,
        "expected_net_value": expected_value,
        "cardinality_ok": cardinality_ok,
        "budget_ok": budget_ok,
        "capacity_ok": capacity_ok,
        "feasible": feasible,
        "customer_ids": customer_ids
    })

quantum_decoded_df = pd.DataFrame(decoded_results)

assert len(quantum_decoded_df) > 0

print("\n1. MEASUREMENT SUMMARY")

print("Unique bitstrings:", len(quantum_decoded_df))
print("Total shots:", quantum_decoded_df["frequency"].sum())

print("\n2. FEASIBILITY SUMMARY")

print(
    "Feasible unique bitstrings:",
    quantum_decoded_df["feasible"].sum()
)

print(
    "Infeasible unique bitstrings:",
    (~quantum_decoded_df["feasible"]).sum()
)

print(
    "Feasible shots:",
    quantum_decoded_df.loc[
        quantum_decoded_df["feasible"],
        "frequency"
    ].sum()
)

print("\n3. TOP MEASURED SOLUTIONS")

print(
    quantum_decoded_df.sort_values(
        "frequency",
        ascending=False
    )[[
        "frequency",
        "selected_count",
        "contact_cost",
        "capacity",
        "expected_net_value",
        "feasible"
    ]].head(10).to_string(index=False)
)

# ------------------------------------------------------------
# 4. BEST FEASIBLE MEASURED SOLUTION
# ------------------------------------------------------------

feasible_results = quantum_decoded_df[
    quantum_decoded_df["feasible"]
].copy()

if len(feasible_results) > 0:

    best_quantum_solution = (
        feasible_results
        .sort_values(
            ["expected_net_value", "frequency"],
            ascending=[False, False]
        )
        .iloc[0]
    )

    print("\n4. BEST FEASIBLE QUANTUM SAMPLE")

    print("Customer IDs:", best_quantum_solution["customer_ids"])
    print("Expected net value:", best_quantum_solution["expected_net_value"])
    print("Contact cost:", best_quantum_solution["contact_cost"])
    print("Capacity:", best_quantum_solution["capacity"])
    print("Frequency:", best_quantum_solution["frequency"])

else:

    best_quantum_solution = None

    print("\n4. NO FEASIBLE QUANTUM SAMPLE FOUND")

print("\nCELL 115 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 116 - QAOA PARAMETER OPTIMIZATION
# EXACT STATEVECTOR EXPECTATION
# ============================================================

import numpy as np
from scipy.optimize import minimize
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

print("=" * 70)
print("CELL 116 - QAOA PARAMETER OPTIMIZATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. VERIFY INPUTS
# ------------------------------------------------------------

assert ising_hamiltonian.num_qubits == 20
assert "ising_terms" in globals()

N = N_QUBITS_TOTAL

# ------------------------------------------------------------
# 2. BUILD PARAMETERIZED QAOA CIRCUIT
# ------------------------------------------------------------

def build_qaoa_state(gamma, beta):

    circuit = QuantumCircuit(N)

    # Initial uniform superposition
    for q in range(N):
        circuit.h(q)

    # Cost Hamiltonian evolution
    for indices, coefficient in ising_terms.items():

        if len(indices) == 1:

            circuit.rz(
                2 * gamma * coefficient,
                indices[0]
            )

        elif len(indices) == 2:

            circuit.rzz(
                2 * gamma * coefficient,
                indices[0],
                indices[1]
            )

    # Mixer evolution
    for q in range(N):
        circuit.rx(2 * beta, q)

    return circuit

# ------------------------------------------------------------
# 3. DEFINE OBJECTIVE FUNCTION
# ------------------------------------------------------------

evaluation_count = 0
optimization_history = []

def qaoa_objective(parameters):

    global evaluation_count

    gamma, beta = parameters

    circuit = build_qaoa_state(gamma, beta)

    state = Statevector.from_instruction(circuit)

    energy = np.real(
        state.expectation_value(ising_hamiltonian)
    )

    evaluation_count += 1

    optimization_history.append({
        "iteration": evaluation_count,
        "gamma": gamma,
        "beta": beta,
        "energy": float(energy)
    })

    if evaluation_count % 5 == 0:
        print(
            f"Evaluation {evaluation_count}: "
            f"Energy = {energy:.6f}, "
            f"Gamma = {gamma:.6f}, "
            f"Beta = {beta:.6f}"
        )

    return float(energy)

# ------------------------------------------------------------
# 4. INITIAL PARAMETERS
# ------------------------------------------------------------

initial_parameters = np.array([
    0.001,
    0.10
])

print("\n1. INITIAL PARAMETERS")
print("Gamma:", initial_parameters[0])
print("Beta:", initial_parameters[1])

# ------------------------------------------------------------
# 5. RUN COBYLA OPTIMIZATION
# ------------------------------------------------------------

constraints = [
    {"type": "ineq", "fun": lambda p: p[0]},
    {"type": "ineq", "fun": lambda p: 0.01 - p[0]},
    {"type": "ineq", "fun": lambda p: p[1]},
    {"type": "ineq", "fun": lambda p: np.pi - p[1]}
]

print("\n2. STARTING QAOA OPTIMIZATION")

qaoa_optimization = minimize(
    qaoa_objective,
    initial_parameters,
    method="COBYLA",
    constraints=constraints,
    options={
        "maxiter": 40,
        "rhobeg": 0.001,
        "tol": 1e-5,
        "disp": True
    }
)

# ------------------------------------------------------------
# 6. SAVE OPTIMIZED PARAMETERS
# ------------------------------------------------------------

optimized_gamma = float(qaoa_optimization.x[0])
optimized_beta = float(qaoa_optimization.x[1])

optimized_energy = float(qaoa_optimization.fun)

print("\n3. OPTIMIZATION RESULTS")

print("Optimization success:", qaoa_optimization.success)
print("Termination message:", qaoa_optimization.message)

print("Optimized gamma:", optimized_gamma)
print("Optimized beta:", optimized_beta)
print("Optimized expected energy:", optimized_energy)

print("Total objective evaluations:", evaluation_count)

# ------------------------------------------------------------
# 7. SAVE OPTIMIZATION HISTORY
# ------------------------------------------------------------

qaoa_history_df = pd.DataFrame(
    optimization_history
)

print("\n4. OPTIMIZATION HISTORY")

print(qaoa_history_df.tail(10).to_string(index=False))

print("\nCELL 116 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 117 - OPTIMIZED QAOA QUANTUM EXECUTION
# ============================================================

import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

print("=" * 70)
print("CELL 117 - OPTIMIZED QAOA EXECUTION")
print("=" * 70)

assert "optimized_gamma" in globals()
assert "optimized_beta" in globals()

N = N_QUBITS_TOTAL

print("\n1. OPTIMIZED PARAMETERS")
print("Gamma:", optimized_gamma)
print("Beta:", optimized_beta)

# ------------------------------------------------------------
# 2. BUILD OPTIMIZED CIRCUIT
# ------------------------------------------------------------

optimized_circuit = QuantumCircuit(N)

for q in range(N):
    optimized_circuit.h(q)

optimized_circuit.barrier()

# Cost Hamiltonian

for indices, coefficient in ising_terms.items():

    if len(indices) == 1:
        optimized_circuit.rz(
            2 * optimized_gamma * coefficient,
            indices[0]
        )

    elif len(indices) == 2:
        optimized_circuit.rzz(
            2 * optimized_gamma * coefficient,
            indices[0],
            indices[1]
        )

optimized_circuit.barrier()

# Mixer Hamiltonian

for q in range(N):
    optimized_circuit.rx(
        2 * optimized_beta,
        q
    )

optimized_circuit.measure_all()

# ------------------------------------------------------------
# 3. SIMULATOR
# ------------------------------------------------------------

SHOTS_OPTIMIZED = 4096

simulator_optimized = AerSimulator(
    method="matrix_product_state"
)

compiled_optimized = transpile(
    optimized_circuit,
    simulator_optimized,
    optimization_level=1
)

print("\n2. CIRCUIT INFORMATION")
print("Qubits:", optimized_circuit.num_qubits)
print("Depth:", optimized_circuit.depth())
print("Shots:", SHOTS_OPTIMIZED)

# ------------------------------------------------------------
# 4. EXECUTE
# ------------------------------------------------------------

optimized_job = simulator_optimized.run(
    compiled_optimized,
    shots=SHOTS_OPTIMIZED,
    seed_simulator=123
)

optimized_result = optimized_job.result()

qaoa_optimized_counts = optimized_result.get_counts()

total_optimized_shots = sum(
    qaoa_optimized_counts.values()
)

assert total_optimized_shots == SHOTS_OPTIMIZED

print("\n3. EXECUTION RESULTS")
print("Total shots:", total_optimized_shots)
print("Unique bitstrings:", len(qaoa_optimized_counts))

print("\n4. TOP 10 MEASUREMENTS")

for bitstring, frequency in sorted(
    qaoa_optimized_counts.items(),
    key=lambda item: item[1],
    reverse=True
)[:10]:

    print(
        bitstring,
        "Frequency:", frequency,
        "Probability:",
        round(frequency / total_optimized_shots, 6)
    )

print("\nCELL 117 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 118 - OPTIMIZED QUANTUM RESULT DECODING
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("CELL 118 - OPTIMIZED QUANTUM RESULT DECODING")
print("=" * 70)

assert "qaoa_optimized_counts" in globals()

optimized_decoded = []

total_shots = sum(qaoa_optimized_counts.values())

for bitstring, frequency in qaoa_optimized_counts.items():

    bits = np.array([
        int(b)
        for b in bitstring.replace(" ", "")
    ][::-1])

    if len(bits) != N_QUBITS_TOTAL:
        continue

    customer_bits = bits[:N_CUSTOMERS]

    selected_indices = np.where(customer_bits == 1)[0]

    selected_count = len(selected_indices)

    selected_cost = float(
        quantum_costs[selected_indices].sum()
    )

    selected_capacity = float(
        quantum_capacities[selected_indices].sum()
    )

    expected_value = float(
        quantum_values[selected_indices].sum()
    )

    feasible = (
        selected_count == K_SELECTED
        and selected_cost <= BUDGET
        and selected_capacity <= CAPACITY_LIMIT
    )

    selected_ids = quantum_candidates.iloc[
        selected_indices
    ]["customer_id"].tolist()

    optimized_decoded.append({
        "bitstring": bitstring,
        "frequency": frequency,
        "probability": frequency / total_shots,
        "selected_count": selected_count,
        "contact_cost": selected_cost,
        "capacity": selected_capacity,
        "expected_net_value": expected_value,
        "feasible": feasible,
        "customer_ids": selected_ids
    })

optimized_quantum_df = pd.DataFrame(
    optimized_decoded
)

feasible_optimized = optimized_quantum_df[
    optimized_quantum_df["feasible"]
].copy()

print("\n1. MEASUREMENT SUMMARY")
print("Total shots:", total_shots)
print("Unique outcomes:", len(optimized_quantum_df))

print("\n2. FEASIBILITY SUMMARY")
print("Feasible unique outcomes:", len(feasible_optimized))
print(
    "Feasible measured shots:",
    int(feasible_optimized["frequency"].sum())
)

print(
    "Feasible shot percentage:",
    round(
        100 * feasible_optimized["frequency"].sum() / total_shots,
        2
    ), "%"
)

print("\n3. BEST FEASIBLE QUANTUM SOLUTIONS")

if len(feasible_optimized) > 0:

    best_optimized_quantum = (
        feasible_optimized
        .sort_values(
            ["expected_net_value", "frequency"],
            ascending=[False, False]
        )
        .reset_index(drop=True)
    )

    print(
        best_optimized_quantum[
            [
                "frequency",
                "probability",
                "selected_count",
                "contact_cost",
                "capacity",
                "expected_net_value",
                "customer_ids"
            ]
        ].head(10).to_string(index=False)
    )

    best_optimized_solution = best_optimized_quantum.iloc[0]

else:

    best_optimized_solution = None

    print("No feasible quantum sample found.")

print("\nCELL 118 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 119 - INDEPENDENT CLASSICAL OPTIMUM VERIFICATION
# ============================================================

from itertools import combinations
import pandas as pd
import numpy as np

print("=" * 70)
print("CELL 119 - INDEPENDENT OPTIMUM VERIFICATION")
print("=" * 70)

all_combinations = list(
    combinations(range(N_CUSTOMERS), K_SELECTED)
)

verified_solutions = []

for indices in all_combinations:

    indices = list(indices)

    cost = float(quantum_costs[indices].sum())

    capacity = float(
        quantum_capacities[indices].sum()
    )

    value = float(quantum_values[indices].sum())

    feasible = (
        cost <= BUDGET
        and capacity <= CAPACITY_LIMIT
    )

    if feasible:

        ids = quantum_candidates.iloc[
            indices
        ]["customer_id"].tolist()

        verified_solutions.append({
            "customer_ids": ids,
            "contact_cost": cost,
            "capacity": capacity,
            "expected_net_value": value
        })

classical_verification_df = pd.DataFrame(
    verified_solutions
)

assert len(all_combinations) == 210
assert len(classical_verification_df) > 0

classical_verification_df = (
    classical_verification_df
    .sort_values(
        "expected_net_value",
        ascending=False
    )
    .reset_index(drop=True)
)

independent_best = classical_verification_df.iloc[0]

print("\n1. ENUMERATION SUMMARY")
print("Total combinations:", len(all_combinations))
print("Feasible combinations:", len(classical_verification_df))
print(
    "Infeasible combinations:",
    len(all_combinations) - len(classical_verification_df)
)

print("\n2. INDEPENDENTLY OPTIMAL SOLUTION")

print("Customer IDs:", independent_best["customer_ids"])
print("Expected net value:", independent_best["expected_net_value"])
print("Contact cost:", independent_best["contact_cost"])
print("Capacity:", independent_best["capacity"])

print("\n3. TOP 5 CLASSICAL SOLUTIONS")

print(
    classical_verification_df.head(5).to_string(index=False)
)

print("\n4. QUANTUM VS CLASSICAL EXACT OPTIMUM")

if best_optimized_solution is not None:

    quantum_value = float(
        best_optimized_solution["expected_net_value"]
    )

    classical_value = float(
        independent_best["expected_net_value"]
    )

    gap = classical_value - quantum_value

    ratio = (
        100 * quantum_value / classical_value
        if classical_value != 0 else np.nan
    )

    print("Classical optimum:", classical_value)
    print("Best quantum sample:", quantum_value)
    print("Absolute gap:", gap)
    print("Quantum/classical percentage:", ratio)

else:

    print("No feasible quantum sample available for comparison.")

print("\nCELL 119 COMPLETE")
print("=" * 70)
```

```python

# ============================================================
# CELL 120 - CLASSICAL VS QUANTUM COMPARISON
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("CELL 120 - CLASSICAL VS QUANTUM COMPARISON")
print("=" * 70)

classical_best = independent_best

classical_value = float(
    classical_best["expected_net_value"]
)

classical_cost = float(
    classical_best["contact_cost"]
)

classical_capacity = float(
    classical_best["capacity"]
)

comparison_rows = [{
    "method": "Exact Classical Enumeration",
    "expected_net_value": classical_value,
    "contact_cost": classical_cost,
    "capacity": classical_capacity,
    "feasible": True,
    "selected_customers": classical_best["customer_ids"]
}]

if best_optimized_solution is not None:

    comparison_rows.append({
        "method": "Optimized QAOA",
        "expected_net_value": float(
            best_optimized_solution["expected_net_value"]
        ),
        "contact_cost": float(
            best_optimized_solution["contact_cost"]
        ),
        "capacity": float(
            best_optimized_solution["capacity"]
        ),
        "feasible": True,
        "selected_customers": best_optimized_solution["customer_ids"]
    })

else:

    comparison_rows.append({
        "method": "Optimized QAOA",
        "expected_net_value": np.nan,
        "contact_cost": np.nan,
        "capacity": np.nan,
        "feasible": False,
        "selected_customers": []
    })

if "best_quantum_solution" in globals() and best_quantum_solution is not None:

    comparison_rows.append({
        "method": "Initial QAOA",
        "expected_net_value": float(
            best_quantum_solution["expected_net_value"]
        ),
        "contact_cost": float(
            best_quantum_solution["contact_cost"]
        ),
        "capacity": float(
            best_quantum_solution["capacity"]
        ),
        "feasible": True,
        "selected_customers": best_quantum_solution["customer_ids"]
    })

comparison_df = pd.DataFrame(comparison_rows)

print("\n1. FINAL COMPARISON TABLE")

print(comparison_df.to_string(index=False))

# ------------------------------------------------------------
# 2. OPTIMIZED QAOA PERFORMANCE
# ------------------------------------------------------------

if best_optimized_solution is not None:

    quantum_value = float(
        best_optimized_solution["expected_net_value"]
    )

    absolute_gap = classical_value - quantum_value

    percentage_gap = (
        100 * absolute_gap / abs(classical_value)
        if classical_value != 0 else np.nan
    )

    approximation_ratio = (
        100 * quantum_value / classical_value
        if classical_value != 0 else np.nan
    )

    print("\n2. OPTIMIZED QAOA PERFORMANCE")

    print("Classical optimum:", classical_value)
    print("Optimized QAOA value:", quantum_value)
    print("Absolute objective gap:", absolute_gap)
    print("Percentage gap:", percentage_gap)
    print("Approximation ratio (%):", approximation_ratio)

    print(
        "Exact customer-set match:",
        set(best_optimized_solution["customer_ids"])
        == set(classical_best["customer_ids"])
    )

else:

    print("\n2. OPTIMIZED QAOA")
    print("No feasible optimized QAOA solution was measured.")

# ------------------------------------------------------------
# 3. FEASIBILITY AND RESOURCE VALIDATION
# ------------------------------------------------------------

print("\n3. CONSTRAINT VALIDATION")

for _, row in comparison_df.iterrows():

    if row["feasible"]:

        print("\nMethod:", row["method"])

        print(
            "Exactly six customers:",
            len(row["selected_customers"]) == K_SELECTED
        )

        print(
            "Budget satisfied:",
            row["contact_cost"] <= BUDGET
        )

        print(
            "Capacity satisfied:",
            row["capacity"] <= CAPACITY_LIMIT
        )

# ------------------------------------------------------------
# 4. FINAL SCIENTIFIC CONCLUSION
# ------------------------------------------------------------

print("\n4. FINAL CONCLUSION")

print("Exact classical optimum independently verified.")

if best_optimized_solution is not None:

    if np.isclose(
        quantum_value,
        classical_value,
        atol=1e-8
    ):

        print("QAOA sampled the exact classical optimal objective.")

    else:

        print("QAOA produced a feasible approximate solution.")

        print(
            "Objective difference:",
            round(absolute_gap, 8)
        )

    print(
        "Quantum solution is feasible:",
        bool(best_optimized_solution["feasible"])
    )

else:

    print("QAOA did not produce a feasible sample.")
    print("Further circuit optimization is required.")

print("\nCELL 120 COMPLETE")
print("=" * 70)
```

```python

# ============================================================================
# CELL 123: PROBLEM STATEMENT & MAIN NOTEBOOK VALIDATION
# ============================================================================
# READ-ONLY: NO FILES WILL BE MODIFIED
# ============================================================================

import json
from pathlib import Path

BASE = Path("/content/drive/MyDrive/ATMIQ_HACKATHON")

PDF_PATH = BASE / "Dataset" / "PROBLEM_STATEMENT_THEATOMIQHACKTHON.pdf"
MAIN_NOTEBOOK = BASE / "Notebooks" / "Classical_ML.ipynb"

print("=" * 75)
print("CELL 123 - OFFICIAL REQUIREMENT & NOTEBOOK VALIDATION")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. OFFICIAL PROBLEM STATEMENT
# ---------------------------------------------------------------------------

print("\n1. OFFICIAL HACKATHON PROBLEM STATEMENT")
print("-" * 75)

if PDF_PATH.exists():

    try:
        from pypdf import PdfReader

        reader = PdfReader(str(PDF_PATH))

        print("PDF pages:", len(reader.pages))

        pdf_text = "\n".join(
            page.extract_text() or ""
            for page in reader.pages
        )

        print("\nEXTRACTED PROBLEM STATEMENT:")
        print(pdf_text[:18000])

    except Exception as error:

        print("PDF extraction failed:", str(error))

else:
    print("Official problem statement not found.")

# ---------------------------------------------------------------------------
# 2. MAIN NOTEBOOK INSPECTION
# ---------------------------------------------------------------------------

print("\n2. MAIN NOTEBOOK INSPECTION")
print("-" * 75)

if MAIN_NOTEBOOK.exists():

    with open(MAIN_NOTEBOOK, "r", encoding="utf-8") as f:
        notebook = json.load(f)

    cells = notebook.get("cells", [])

    print("Notebook:", MAIN_NOTEBOOK.name)
    print("Total cells:", len(cells))

    code_count = 0
    markdown_count = 0
    executed_count = 0

    for cell in cells:

        if cell["cell_type"] == "code":

            code_count += 1

            if cell.get("execution_count") is not None:
                executed_count += 1

        elif cell["cell_type"] == "markdown":
            markdown_count += 1

    print("Code cells:", code_count)
    print("Markdown cells:", markdown_count)
    print("Executed code cells:", executed_count)

    # -----------------------------------------------------------------------
    # 3. SEARCH FOR QUANTUM IMPLEMENTATION
    # -----------------------------------------------------------------------

    print("\n3. QUANTUM IMPLEMENTATION SEARCH")
    print("-" * 75)

    quantum_terms = [
        "qiskit",
        "qaoa",
        "quantum",
        "quantumcircuit",
        "aersimulator",
        "qubo",
        "ising",
        "hamiltonian",
        "vqe",
        "measurement",
        "shots",
        "optimizer",
        "sampler",
        "transpile"
    ]

    matches = {term: [] for term in quantum_terms}

    for index, cell in enumerate(cells):

        source = "".join(cell.get("source", []))
        source_lower = source.lower()

        for term in quantum_terms:

            if term in source_lower:

                matches[term].append(index + 1)

    for term, positions in matches.items():

        print(
            f"{term:20s}: "
            f"{positions[:30] if positions else 'NOT FOUND'}"
        )

    # -----------------------------------------------------------------------
    # 4. NOTEBOOK SECTION OUTLINE
    # -----------------------------------------------------------------------

    print("\n4. NOTEBOOK SECTION OUTLINE")
    print("-" * 75)

    for index, cell in enumerate(cells):

        source = "".join(cell.get("source", [])).strip()

        if not source:
            continue

        first_line = source.split("\n")[0][:120]

        if (
            cell["cell_type"] == "markdown"
            or any(term in source.lower() for term in [
                "qaoa", "quantum", "qubo",
                "stage 1", "stage 2", "stage 3",
                "optimization", "comparison"
            ])
        ):

            print(
                f"Cell {index + 1:04d} | "
                f"{cell['cell_type']:8s} | {first_line}"
            )

    # -----------------------------------------------------------------------
    # 5. FINAL QUANTUM EXECUTION OUTPUT CHECK
    # -----------------------------------------------------------------------

    print("\n5. QUANTUM EXECUTION OUTPUT CHECK")
    print("-" * 75)

    output_count = 0
    error_count = 0

    for index, cell in enumerate(cells):

        if cell["cell_type"] != "code":
            continue

        outputs = cell.get("outputs", [])

        for output in outputs:

            output_count += 1

            if output.get("output_type") == "error":

                error_count += 1

                print(
                    "Error in notebook cell:",
                    index + 1,
                    output.get("ename"),
                    output.get("evalue")
                )

    print("Total stored outputs:", output_count)
    print("Stored execution errors:", error_count)

else:

    print("Main notebook NOT FOUND.")

# ---------------------------------------------------------------------------
# 6. FINAL VALIDATION SUMMARY
# ---------------------------------------------------------------------------

print("\n6. FINAL VALIDATION SUMMARY")
print("-" * 75)

print("Official problem statement available:", PDF_PATH.exists())
print("Main notebook available:", MAIN_NOTEBOOK.exists())

if MAIN_NOTEBOOK.exists():

    print("Main notebook cell count:", len(cells))
    print("Quantum implementation references:",
          sum(len(v) for v in matches.values()))

print("\nIMPORTANT:")
print("Keyword matches alone do not prove a valid quantum implementation.")
print("Compare actual algorithms, data flow and saved outputs.")
print("No project files were changed.")

print("\n" + "=" * 75)
print("CELL 123 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 124: FINAL QAOA IMPLEMENTATION & ERROR AUDIT
# ============================================================================
# Purpose:
# 1. Extract official problem statement using available PDF libraries
# 2. Inspect actual QAOA implementation (cells 81-100)
# 3. Inspect stored outputs and execution counts
# 4. Investigate the stored Cell 42 error
# 5. Verify notebook execution history
#
# READ-ONLY: NO INSTALLATION OR FILE MODIFICATION
# ============================================================================

import json
import importlib
from pathlib import Path

BASE = Path("/content/drive/MyDrive/ATMIQ_HACKATHON")

PDF_PATH = BASE / "Dataset" / "PROBLEM_STATEMENT_THEATOMIQHACKTHON.pdf"
NOTEBOOK_PATH = BASE / "Notebooks" / "Classical_ML.ipynb"

print("=" * 75)
print("CELL 124 - FINAL QAOA IMPLEMENTATION & ERROR AUDIT")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. OFFICIAL PDF EXTRACTION
# ---------------------------------------------------------------------------

print("\n1. OFFICIAL PROBLEM STATEMENT EXTRACTION")
print("-" * 75)

pdf_text = ""

pdf_libraries = ["fitz", "pdfplumber", "PyPDF2", "pypdf"]

for library in pdf_libraries:

    try:

        module = importlib.import_module(library)

        if library == "fitz":

            document = module.open(str(PDF_PATH))

            pdf_text = "\n".join(
                page.get_text()
                for page in document
            )

        elif library == "pdfplumber":

            with module.open(str(PDF_PATH)) as pdf:

                pdf_text = "\n".join(
                    page.extract_text() or ""
                    for page in pdf.pages
                )

        else:

            reader = module.PdfReader(str(PDF_PATH))

            pdf_text = "\n".join(
                page.extract_text() or ""
                for page in reader.pages
            )

        if pdf_text.strip():

            print("Extraction library:", library)
            print("Extracted characters:", len(pdf_text))
            print("\nPROBLEM STATEMENT:")
            print(pdf_text[:14000])
            break

    except Exception as error:

        print(f"{library}: unavailable or extraction failed")
        print("Reason:", str(error)[:200])

if not pdf_text.strip():

    print("PDF text extraction unsuccessful.")
    print("The original PDF remains unchanged.")

# ---------------------------------------------------------------------------
# 2. LOAD MAIN NOTEBOOK
# ---------------------------------------------------------------------------

print("\n2. MAIN NOTEBOOK EXECUTION STATUS")
print("-" * 75)

with open(NOTEBOOK_PATH, "r", encoding="utf-8") as f:

    notebook = json.load(f)

cells = notebook.get("cells", [])

executed = [
    i + 1
    for i, cell in enumerate(cells)
    if cell.get("cell_type") == "code"
    and cell.get("execution_count") is not None
]

print("Total notebook cells:", len(cells))
print("Executed code cells:", len(executed))
print("Unexecuted code cells:", len(cells) - len(executed))
print("Execution counts:", executed)

# ---------------------------------------------------------------------------
# 3. INSPECT QAOA IMPLEMENTATION CELLS
# ---------------------------------------------------------------------------

print("\n3. ACTUAL QAOA IMPLEMENTATION")
print("-" * 75)

for cell_number in range(81, 101):

    if cell_number > len(cells):
        continue

    cell = cells[cell_number - 1]

    source = "".join(cell.get("source", []))

    print("\n" + "=" * 65)
    print("NOTEBOOK CELL:", cell_number)
    print("TYPE:", cell.get("cell_type"))
    print("EXECUTION COUNT:", cell.get("execution_count"))
    print("=" * 65)

    print("\nSOURCE CODE:")
    print(source[:4500] if source else "[EMPTY]")

    outputs = cell.get("outputs", [])

    print("\nOUTPUT SUMMARY:")

    if not outputs:

        print("No stored output.")

    for output in outputs[:4]:

        output_type = output.get("output_type")

        if output_type == "stream":

            print("".join(output.get("text", []))[:1800])

        elif output_type == "execute_result":

            data = output.get("data", {})

            print(
                str(data.get("text/plain", ""))[:1800]
            )

        elif output_type == "display_data":

            data = output.get("data", {})

            print(
                str(data.get("text/plain", "[Rich display output]"))[:1000]
            )

        elif output_type == "error":

            print("ERROR:", output.get("ename"))
            print("MESSAGE:", output.get("evalue"))

# ---------------------------------------------------------------------------
# 4. INVESTIGATE STORED CELL 42 ERROR
# ---------------------------------------------------------------------------

print("\n4. STORED CELL 42 ERROR INVESTIGATION")
print("-" * 75)

error_cell = cells[41]

print("Execution count:", error_cell.get("execution_count"))
print("\nSOURCE CODE:")
print("".join(error_cell.get("source", []))[:5000])

for output in error_cell.get("outputs", []):

    if output.get("output_type") == "error":

        print("\nERROR TYPE:", output.get("ename"))
        print("ERROR MESSAGE:", output.get("evalue"))

        traceback = output.get("traceback", [])

        print("\nTRACEBACK:")

        for line in traceback[-8:]:
            print(line[:500])

# ---------------------------------------------------------------------------
# 5. STORED EXECUTION ERROR INVENTORY
# ---------------------------------------------------------------------------

print("\n5. ALL STORED NOTEBOOK EXECUTION ERRORS")
print("-" * 75)

errors = []

for i, cell in enumerate(cells):

    for output in cell.get("outputs", []):

        if output.get("output_type") == "error":

            errors.append({
                "cell": i + 1,
                "type": output.get("ename"),
                "message": output.get("evalue")
            })

print("Total stored errors:", len(errors))

for error in errors:

    print(
        f"Cell {error['cell']}: "
        f"{error['type']} - {error['message']}"
    )

# ---------------------------------------------------------------------------
# 6. FINAL VALIDATION
# ---------------------------------------------------------------------------

print("\n6. FINAL VALIDATION SUMMARY")
print("-" * 75)

print("Official PDF available:", PDF_PATH.exists())
print("PDF text extracted:", bool(pdf_text.strip()))
print("Main notebook available:", NOTEBOOK_PATH.exists())
print("QAOA cells inspected: 81-100")
print("Executed code cells:", len(executed))
print("Stored execution errors:", len(errors))

print("\nIMPORTANT:")
print("Stored notebook outputs may belong to earlier executions.")
print("Execution counts alone do not establish reproducibility.")
print("Final results must be checked against the actual saved data.")

print("\n" + "=" * 75)
print("CELL 124 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 125 - QAOA EFFECTIVENESS & UNIFORM DISTRIBUTION AUDIT
# READ ONLY - NO FILES MODIFIED
# ============================================================================

import numpy as np
import pandas as pd
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

print("=" * 75)
print("CELL 125 - QAOA EFFECTIVENESS & DISTRIBUTION AUDIT")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. VERIFY REQUIRED VARIABLES
# ---------------------------------------------------------------------------

required_variables = [
    "N_QUBITS_TOTAL",
    "ising_terms",
    "ising_hamiltonian",
    "optimized_gamma",
    "optimized_beta",
    "qaoa_initial_counts",
    "qaoa_optimized_counts"
]

missing = [
    name for name in required_variables
    if name not in globals()
]

if missing:
    raise RuntimeError(
        "Required variables missing: " + ", ".join(missing)
    )

N = N_QUBITS_TOTAL
TOTAL_STATES = 2 ** N

print("\n1. PARAMETERS")
print("Total qubits:", N)
print("Total possible bitstrings:", TOTAL_STATES)
print("Optimized gamma:", optimized_gamma)
print("Optimized beta:", optimized_beta)
print("Absolute optimized gamma:", abs(optimized_gamma))

# ---------------------------------------------------------------------------
# 2. BUILD EXACT QAOA STATE WITHOUT MEASUREMENTS
# ---------------------------------------------------------------------------

def build_audit_state(gamma, beta):

    circuit = QuantumCircuit(N)

    # Initial uniform superposition
    for q in range(N):
        circuit.h(q)

    # Cost Hamiltonian
    for indices, coefficient in ising_terms.items():

        if len(indices) == 1:
            circuit.rz(
                2 * gamma * coefficient,
                indices[0]
            )

        elif len(indices) == 2:
            circuit.rzz(
                2 * gamma * coefficient,
                indices[0],
                indices[1]
            )

    # Mixer Hamiltonian
    for q in range(N):
        circuit.rx(2 * beta, q)

    return circuit


# Initial parameters used in Cell 114
initial_gamma = 0.10
initial_beta = 0.10

initial_state = Statevector.from_instruction(
    build_audit_state(initial_gamma, initial_beta)
)

optimized_state = Statevector.from_instruction(
    build_audit_state(optimized_gamma, optimized_beta)
)

# ---------------------------------------------------------------------------
# 3. COMPARE PROBABILITY DISTRIBUTIONS
# ---------------------------------------------------------------------------

uniform_probability = 1.0 / TOTAL_STATES

def distribution_metrics(state):

    probabilities = np.asarray(
        state.probabilities(),
        dtype=float
    )

    probabilities = probabilities[
        probabilities > 0
    ]

    entropy = -np.sum(
        probabilities * np.log2(probabilities)
    )

    effective_states = 2 ** entropy

    uniform_fidelity = (
        np.sum(np.sqrt(probabilities / TOTAL_STATES))
        ** 2
    )

    return {
        "maximum_probability": float(probabilities.max()),
        "minimum_probability": float(probabilities.min()),
        "mean_probability": float(probabilities.mean()),
        "entropy_bits": float(entropy),
        "effective_states": float(effective_states),
        "uniform_entropy_bits": float(np.log2(TOTAL_STATES)),
        "uniform_fidelity": float(uniform_fidelity),
        "total_variation_from_uniform": float(
            0.5 * np.abs(
                state.probabilities() - uniform_probability
            ).sum()
        )
    }


initial_metrics = distribution_metrics(initial_state)
optimized_metrics = distribution_metrics(optimized_state)

comparison = pd.DataFrame([
    {
        "circuit": "Initial QAOA",
        "gamma": initial_gamma,
        "beta": initial_beta,
        **initial_metrics
    },
    {
        "circuit": "Optimized QAOA",
        "gamma": optimized_gamma,
        "beta": optimized_beta,
        **optimized_metrics
    }
])

print("\n2. EXACT STATEVECTOR DISTRIBUTION COMPARISON")
print(comparison.to_string(index=False))

# ---------------------------------------------------------------------------
# 4. MEASURE HOW MUCH THE COST UNITARY CONTRIBUTES
# ---------------------------------------------------------------------------

print("\n3. COST-HAMILTONIAN CONTRIBUTION")

if abs(optimized_gamma) < 1e-10:
    print("WARNING: Optimized gamma is effectively zero.")
    print("The optimized cost-Hamiltonian evolution is effectively inactive.")
else:
    print("Optimized gamma is nonzero.")

# ---------------------------------------------------------------------------
# 5. COMPARE MEASURED SHOTS WITH UNIFORM EXPECTATION
# ---------------------------------------------------------------------------

initial_shots = sum(qaoa_initial_counts.values())
optimized_shots = sum(qaoa_optimized_counts.values())

initial_unique = len(qaoa_initial_counts)
optimized_unique = len(qaoa_optimized_counts)

print("\n4. MEASUREMENT DIVERSITY")

print("Initial QAOA shots:", initial_shots)
print("Initial unique outcomes:", initial_unique)
print(
    "Initial unique-outcome percentage:",
    round(100 * initial_unique / initial_shots, 4), "%"
)

print("Optimized QAOA shots:", optimized_shots)
print("Optimized unique outcomes:", optimized_unique)
print(
    "Optimized unique-outcome percentage:",
    round(100 * optimized_unique / optimized_shots, 4), "%"
)

print(
    "Expected probability of one bitstring under uniform sampling:",
    uniform_probability
)

# ---------------------------------------------------------------------------
# 6. SCIENTIFIC INTERPRETATION
# ---------------------------------------------------------------------------

print("\n5. SCIENTIFIC INTERPRETATION")

if abs(optimized_gamma) < 1e-10:
    print("RESULT: Optimized cost evolution is effectively inactive.")

if optimized_metrics["uniform_fidelity"] > 0.999999:
    print("RESULT: Optimized state is effectively uniform.")
else:
    print("RESULT: Optimized state differs from the uniform state.")

if (
    optimized_metrics["entropy_bits"]
    >= initial_metrics["entropy_bits"] - 1e-8
):
    print("Entropy did not decrease meaningfully after optimization.")
else:
    print("Optimized state has lower entropy than the initial state.")

print("\n6. FINAL AUDIT STATUS")
print("QAOA circuit construction: PRESENT")
print("QAOA parameter optimization: PRESENT")
print("Optimized cost evolution active:",
      abs(optimized_gamma) >= 1e-10)
print("Files modified: NO")

print("\n" + "=" * 75)
print("CELL 125 COMPLETE")
print("=" * 75)
```

```python

import numpy as np
import math

print("=" * 75)
print("CELL 127 - QAOA MEASUREMENT DISTRIBUTION AUDIT")
print("=" * 75)

TOTAL_STATES = 2 ** N_QUBITS_TOTAL

def audit_counts(counts, label):
    clean_counts = {}

    for bitstring, frequency in counts.items():
        key = str(bitstring).replace(" ", "")
        clean_counts[key] = clean_counts.get(key, 0) + int(frequency)

    frequencies = np.array(list(clean_counts.values()), dtype=float)
    total_shots = int(frequencies.sum())
    probabilities = frequencies / total_shots

    entropy = -np.sum(
        probabilities * np.log2(probabilities + 1e-15)
    )

    effective_states = 2 ** entropy
    collision_probability = np.sum(probabilities ** 2)
    collision_effective_states = 1 / collision_probability

    print(f"\n{label}")
    print("-" * 55)
    print("Total shots:", total_shots)
    print("Unique measured bitstrings:", len(clean_counts))
    print("Possible bitstrings:", TOTAL_STATES)
    print("Observed support percentage:",
          round(100 * len(clean_counts) / TOTAL_STATES, 6), "%")
    print("Measurement entropy:", round(entropy, 6), "bits")
    print("Effective states (entropy):", round(effective_states, 2))
    print("Effective states (collisions):",
          round(collision_effective_states, 2))
    print("Maximum observed frequency:", int(frequencies.max()))
    print("Maximum observed probability:",
          round(float(probabilities.max()), 6))

    expected_unique_uniform = TOTAL_STATES * (
        1 - (1 - 1 / TOTAL_STATES) ** total_shots
    )

    print("Expected unique outcomes under uniform sampling:",
          round(expected_unique_uniform, 2))

    return {
        "shots": total_shots,
        "unique": len(clean_counts),
        "entropy": entropy,
        "effective_states": effective_states,
        "collision_effective_states": collision_effective_states,
        "counts": clean_counts
    }


initial_audit = audit_counts(
    qaoa_initial_counts,
    "1. INITIAL QAOA MEASUREMENTS"
)

optimized_audit = audit_counts(
    qaoa_optimized_counts,
    "2. OPTIMIZED QAOA MEASUREMENTS"
)

print("\n3. OPTIMIZATION PARAMETER CHECK")
print("-" * 55)
print("Optimized gamma:", optimized_gamma)
print("Optimized beta:", optimized_beta)
print("Absolute gamma:", abs(optimized_gamma))

if abs(optimized_gamma) < 1e-8:
    print("WARNING: Cost-unitary angle is effectively zero.")
else:
    print("Cost-unitary angle is nonzero.")

print("\n4. DISTRIBUTION COMPARISON")
print("-" * 55)
print("Initial entropy:", round(initial_audit["entropy"], 6))
print("Optimized entropy:", round(optimized_audit["entropy"], 6))
print("Entropy difference:",
      round(optimized_audit["entropy"] - initial_audit["entropy"], 6))

print("\nCell 127 completed successfully.")
```

```python

# ============================================================================
# CELL 128: QAOA HAMILTONIAN COEFFICIENT AND SCALING AUDIT
# COMPATIBILITY-FIXED VERSION
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 128 - QAOA HAMILTONIAN COEFFICIENT AND SCALING AUDIT")
print("=" * 75)

# --------------------------------------------------------------------------
# 1. VARIABLE AVAILABILITY
# --------------------------------------------------------------------------

required_variables = [
    "N_QUBITS_TOTAL",
    "ising_terms",
    "ising_hamiltonian",
    "optimized_gamma",
    "optimized_beta"
]

print("\n1. REQUIRED VARIABLE CHECK")

for name in required_variables:
    print(f"{name:25s}: {'AVAILABLE' if name in globals() else 'MISSING'}")

missing = [x for x in required_variables if x not in globals()]

if missing:
    raise NameError(f"Missing required variables: {missing}")

# --------------------------------------------------------------------------
# 2. HAMILTONIAN STRUCTURE
# --------------------------------------------------------------------------

print("\n2. HAMILTONIAN STRUCTURE")

coefficients = np.asarray(ising_hamiltonian.coeffs)
pauli_labels = ising_hamiltonian.paulis.to_labels()

print("Number of qubits:", N_QUBITS_TOTAL)
print("Hamiltonian terms:", len(coefficients))
print("Ising dictionary terms:", len(ising_terms))
print("Hamiltonian type:", type(ising_hamiltonian).__name__)

# SparsePauliOp Hermiticity check:
# Pauli operators are Hermitian, so real coefficients imply Hermiticity.

is_hermitian = bool(
    np.all(np.abs(coefficients.imag) < 1e-10)
)

print("Hamiltonian is Hermitian:", is_hermitian)

# --------------------------------------------------------------------------
# 3. COEFFICIENT STATISTICS
# --------------------------------------------------------------------------

print("\n3. COEFFICIENT STATISTICS")

abs_coefficients = np.abs(coefficients)

print("Maximum absolute coefficient:", np.max(abs_coefficients))
print("Minimum absolute coefficient:", np.min(abs_coefficients))
print("Mean absolute coefficient:", np.mean(abs_coefficients))
print("Median absolute coefficient:", np.median(abs_coefficients))
print("Coefficient L1 norm:", np.sum(abs_coefficients))
print("Coefficient L2 norm:", np.linalg.norm(abs_coefficients))

# --------------------------------------------------------------------------
# 4. IDENTITY TERM CHECK
# --------------------------------------------------------------------------

print("\n4. IDENTITY TERM CHECK")

identity_mask = np.array([
    all(char == "I" for char in label)
    for label in pauli_labels
])

identity_coefficients = coefficients[identity_mask]
non_identity_coefficients = coefficients[~identity_mask]

identity_energy = np.sum(identity_coefficients)

print("Identity terms:", len(identity_coefficients))
print("Non-identity terms:", len(non_identity_coefficients))
print("Identity contribution:", identity_energy)
print("Non-identity coefficient count:", len(non_identity_coefficients))

# --------------------------------------------------------------------------
# 5. OPTIMIZED ANGLE SCALING
# --------------------------------------------------------------------------

print("\n5. OPTIMIZED ANGLE SCALING")

gamma = float(optimized_gamma)
beta = float(optimized_beta)

max_cost_angle = abs(gamma) * np.max(abs_coefficients)

print("Optimized gamma:", gamma)
print("Optimized beta:", beta)
print("Maximum cost-unitary angle:", max_cost_angle)
print("Absolute gamma:", abs(gamma))

if max_cost_angle < 1e-8:
    print("WARNING: Cost-unitary evolution is effectively inactive.")
else:
    print("Cost-unitary angle is numerically non-negligible.")

# --------------------------------------------------------------------------
# 6. HAMILTONIAN COEFFICIENT DISTRIBUTION
# --------------------------------------------------------------------------

print("\n6. COEFFICIENT DISTRIBUTION")

coefficient_df = pd.DataFrame({
    "pauli_string": pauli_labels,
    "coefficient": coefficients.real,
    "imaginary_part": coefficients.imag,
    "absolute_coefficient": abs_coefficients
})

coefficient_df = coefficient_df.sort_values(
    "absolute_coefficient",
    ascending=False
).reset_index(drop=True)

print("\nTOP 15 HAMILTONIAN TERMS")

display(coefficient_df.head(15))

# --------------------------------------------------------------------------
# 7. NUMERICAL VALIDITY
# --------------------------------------------------------------------------

print("\n7. NUMERICAL VALIDITY")

print("All coefficients finite:",
      bool(np.all(np.isfinite(coefficients))))

print("All coefficients real within tolerance:",
      is_hermitian)

print("Non-identity Hamiltonian norm:",
      np.linalg.norm(non_identity_coefficients))

# --------------------------------------------------------------------------
# 8. FINAL DIAGNOSTIC
# --------------------------------------------------------------------------

print("\n8. FINAL DIAGNOSTIC")

if not np.all(np.isfinite(coefficients)):
    print("CRITICAL: Hamiltonian contains non-finite coefficients.")

elif not is_hermitian:
    print("CRITICAL: Hamiltonian is not Hermitian.")

elif len(non_identity_coefficients) == 0:
    print("CRITICAL: Hamiltonian contains no non-identity terms.")

elif max_cost_angle < 1e-8:
    print("CRITICAL: Optimized gamma suppresses cost evolution.")

else:
    print("Hamiltonian structure requires no immediate numerical warning.")

print("\n" + "=" * 75)
print("CELL 128 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 130: QAOA OBJECTIVE SENSITIVITY AND PARAMETER LANDSCAPE AUDIT
# ============================================================================

import numpy as np
import pandas as pd
from qiskit.quantum_info import Statevector

print("=" * 75)
print("CELL 130 - QAOA OBJECTIVE SENSITIVITY AUDIT")
print("=" * 75)

# --------------------------------------------------------------------------
# 1. BASELINE
# --------------------------------------------------------------------------

print("\n1. BASELINE PARAMETERS")

test_beta = float(optimized_beta)

print("Current optimized gamma:", optimized_gamma)
print("Current optimized beta:", test_beta)

# --------------------------------------------------------------------------
# 2. INDEPENDENT ENERGY EVALUATION
# --------------------------------------------------------------------------

def independent_qaoa_energy(gamma, beta):

    circuit = build_qaoa_state(
        float(gamma),
        float(beta)
    )

    state = Statevector.from_instruction(circuit)

    energy = np.real(
        state.expectation_value(ising_hamiltonian)
    )

    return float(energy)

# --------------------------------------------------------------------------
# 3. GAMMA SENSITIVITY
# --------------------------------------------------------------------------

print("\n2. GAMMA SENSITIVITY TEST")

gamma_values = [
    -0.00010,
    -0.00005,
    -0.000025,
    -0.00001,
    0.0,
    0.00001,
    0.000025,
    0.00005,
    0.00010,
    0.00050,
    0.00100
]

gamma_results = []

for gamma in gamma_values:

    energy = independent_qaoa_energy(
        gamma,
        test_beta
    )

    gamma_results.append({
        "gamma": gamma,
        "beta": test_beta,
        "energy": energy
    })

gamma_df = pd.DataFrame(gamma_results)

display(gamma_df)

# --------------------------------------------------------------------------
# 4. BETA SENSITIVITY
# --------------------------------------------------------------------------

print("\n3. BETA SENSITIVITY TEST")

test_gamma = -0.00005

beta_values = [
    0.0,
    0.025,
    0.05,
    0.075,
    0.09,
    0.099,
    0.10,
    0.11,
    0.15,
    0.25,
    0.50
]

beta_results = []

for beta in beta_values:

    energy = independent_qaoa_energy(
        test_gamma,
        beta
    )

    beta_results.append({
        "gamma": test_gamma,
        "beta": beta,
        "energy": energy
    })

beta_df = pd.DataFrame(beta_results)

display(beta_df)

# --------------------------------------------------------------------------
# 5. BEST TESTED PARAMETERS
# --------------------------------------------------------------------------

print("\n4. BEST TESTED PARAMETERS")

combined_df = pd.concat(
    [gamma_df, beta_df],
    ignore_index=True
)

best_index = combined_df["energy"].idxmin()
best_row = combined_df.loc[best_index]

print("Best tested gamma:", best_row["gamma"])
print("Best tested beta:", best_row["beta"])
print("Best tested energy:", best_row["energy"])

print("Original optimized energy:",
      globals().get("optimized_energy", np.nan))

# --------------------------------------------------------------------------
# 6. GAMMA ZERO BASELINE
# --------------------------------------------------------------------------

print("\n5. GAMMA ZERO BASELINE")

zero_energy = independent_qaoa_energy(
    0.0,
    test_beta
)

negative_energy = independent_qaoa_energy(
    -0.00005,
    test_beta
)

positive_energy = independent_qaoa_energy(
    0.00005,
    test_beta
)

print("Gamma = 0:", zero_energy)
print("Gamma = -0.00005:", negative_energy)
print("Gamma = +0.00005:", positive_energy)

print("Negative gamma improvement:",
      zero_energy - negative_energy)

print("Positive gamma improvement:",
      zero_energy - positive_energy)

# --------------------------------------------------------------------------
# 7. OBJECTIVE LANDSCAPE DIAGNOSTIC
# --------------------------------------------------------------------------

print("\n6. OBJECTIVE LANDSCAPE DIAGNOSTIC")

gamma_energy_range = (
    gamma_df["energy"].max() -
    gamma_df["energy"].min()
)

beta_energy_range = (
    beta_df["energy"].max() -
    beta_df["energy"].min()
)

print("Gamma energy range:", gamma_energy_range)
print("Beta energy range:", beta_energy_range)

if gamma_energy_range < 1e-8:
    print("WARNING: Objective appears insensitive to gamma.")
else:
    print("Gamma has measurable objective sensitivity.")

if beta_energy_range < 1e-8:
    print("WARNING: Objective appears insensitive to beta.")
else:
    print("Beta has measurable objective sensitivity.")

# --------------------------------------------------------------------------
# 8. ORIGINAL VARIABLE PRESERVATION
# --------------------------------------------------------------------------

print("\n7. ORIGINAL OPTIMIZATION VARIABLE PRESERVATION")

print("optimized_gamma:", optimized_gamma)
print("optimized_beta:", optimized_beta)

print("\n" + "=" * 75)
print("CELL 130 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 133: QAOA OPTIMIZATION LANDSCAPE DIAGNOSTIC
# ============================================================================

import numpy as np
import pandas as pd
from qiskit.quantum_info import Statevector

print("=" * 75)
print("CELL 133 - QAOA OPTIMIZATION LANDSCAPE DIAGNOSTIC")
print("=" * 75)

# ----------------------------------------------------------------
# 1. VERIFY REQUIRED VARIABLES
# ----------------------------------------------------------------

required = [
    "ising_hamiltonian",
    "ising_terms",
    "N_QUBITS_TOTAL",
    "optimized_gamma",
    "optimized_beta",
    "optimized_energy",
    "build_qaoa_state"
]

missing = [x for x in required if x not in globals()]

assert not missing, f"Missing variables: {missing}"

assert N_QUBITS_TOTAL == 20
assert ising_hamiltonian.num_qubits == 20

# ----------------------------------------------------------------
# 2. INDEPENDENT ENERGY EVALUATION
# ----------------------------------------------------------------

def diagnostic_energy(gamma, beta):

    circuit = build_qaoa_state(
        float(gamma),
        float(beta)
    )

    state = Statevector.from_instruction(circuit)

    energy = np.real(
        state.expectation_value(ising_hamiltonian)
    )

    return float(energy)

# ----------------------------------------------------------------
# 3. GAMMA SENSITIVITY
# ----------------------------------------------------------------

gamma_values = [
    0.0,
    0.00001,
    0.0001,
    0.001,
    0.005,
    0.01
]

gamma_results = []

print("\n1. GAMMA SENSITIVITY TEST")

for gamma in gamma_values:

    energy = diagnostic_energy(
        gamma,
        optimized_beta
    )

    gamma_results.append({
        "gamma": gamma,
        "beta": optimized_beta,
        "energy": energy
    })

    print(
        f"Gamma: {gamma:.8f} | "
        f"Energy: {energy:.8f}"
    )

gamma_df = pd.DataFrame(gamma_results)

# ----------------------------------------------------------------
# 4. BETA SENSITIVITY
# ----------------------------------------------------------------

beta_values = [
    0.05,
    0.075,
    optimized_beta,
    0.125,
    0.15
]

beta_results = []

print("\n2. BETA SENSITIVITY TEST")

for beta in beta_values:

    energy = diagnostic_energy(
        0.0,
        beta
    )

    beta_results.append({
        "gamma": 0.0,
        "beta": beta,
        "energy": energy
    })

    print(
        f"Beta: {beta:.8f} | "
        f"Energy: {energy:.8f}"
    )

beta_df = pd.DataFrame(beta_results)

# ----------------------------------------------------------------
# 5. ORIGINAL VS DIAGNOSTIC RESULTS
# ----------------------------------------------------------------

print("\n3. ORIGINAL OPTIMIZATION RESULT")

print("Original gamma:", optimized_gamma)
print("Original beta:", optimized_beta)
print("Original energy:", optimized_energy)

# ----------------------------------------------------------------
# 6. BEST TESTED GAMMA
# ----------------------------------------------------------------

best_gamma_row = gamma_df.loc[
    gamma_df["energy"].idxmin()
]

print("\n4. BEST TESTED GAMMA")

print("Gamma:", best_gamma_row["gamma"])
print("Beta:", best_gamma_row["beta"])
print("Energy:", best_gamma_row["energy"])

# ----------------------------------------------------------------
# 7. BEST TESTED BETA
# ----------------------------------------------------------------

best_beta_row = beta_df.loc[
    beta_df["energy"].idxmin()
]

print("\n5. BEST TESTED BETA AT ZERO GAMMA")

print("Gamma:", best_beta_row["gamma"])
print("Beta:", best_beta_row["beta"])
print("Energy:", best_beta_row["energy"])

# ----------------------------------------------------------------
# 8. DIAGNOSTIC CONCLUSION
# ----------------------------------------------------------------

print("\n6. DIAGNOSTIC CONCLUSION")

if np.allclose(
    gamma_df["energy"].values,
    gamma_df["energy"].iloc[0],
    rtol=1e-8,
    atol=1e-6
):

    print("WARNING: Energy is nearly insensitive to gamma.")

else:

    print("Gamma changes the expected energy.")

if abs(optimized_gamma) < 1e-8:

    print("WARNING: Optimizer converged effectively to zero gamma.")

print("\nCELL 133 COMPLETED")
print("=" * 75)
```

```python

# ============================================================
# CELL 135 - QAOA CIRCUIT AUDIT AND PARAMETER SCAN
# ERROR-CORRECTED 20-QUBIT VERSION
# ============================================================

import os
import json
import numpy as np
import pandas as pd

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, SparsePauliOp

print("=" * 75)
print("CELL 135 - QAOA CIRCUIT AUDIT AND PARAMETER SCAN")
print("=" * 75)

# ------------------------------------------------------------
# 1. VERIFY INPUTS
# ------------------------------------------------------------

assert "ising_terms" in globals(), "ising_terms is missing."
assert "ising_hamiltonian" in globals(), "ising_hamiltonian is missing."

N = ising_hamiltonian.num_qubits

assert N == 20, f"Expected 20 qubits, found {N}."
assert len(ising_terms) > 0, "ising_terms is empty."

print("\n1. HAMILTONIAN STRUCTURE")
print("Qubits:", N)
print("Ising dictionary terms:", len(ising_terms))
print("Pauli terms:", len(ising_hamiltonian))

term_degrees = [len(indices) for indices in ising_terms]

print("Maximum interaction degree:", max(term_degrees))
print("Identity terms:", sum(d == 0 for d in term_degrees))
print("Single-qubit terms:", sum(d == 1 for d in term_degrees))
print("Two-qubit terms:", sum(d == 2 for d in term_degrees))
print("Higher-order terms:", sum(d > 2 for d in term_degrees))

assert max(term_degrees) <= 2, (
    "Unsupported higher-order Ising terms detected."
)

# ------------------------------------------------------------
# 2. RECONSTRUCT AND COMPARE HAMILTONIANS
# ------------------------------------------------------------

print("\n2. ISING DICTIONARY VS QISKIT HAMILTONIAN")

reconstructed_terms = []

for indices, coefficient in ising_terms.items():

    label = ["I"] * N

    for qubit in indices:

        assert 0 <= qubit < N, (
            f"Invalid qubit index: {qubit}"
        )

        label[N - 1 - qubit] = "Z"

    reconstructed_terms.append(
        ("".join(label), complex(coefficient))
    )

reconstructed_hamiltonian = SparsePauliOp.from_list(
    reconstructed_terms
).simplify(atol=1e-12)

original_hamiltonian = ising_hamiltonian.simplify(
    atol=1e-12
)

# Compare coefficient dictionaries. This avoids relying on
# the order in which Pauli terms are stored.

def pauli_coefficient_map(operator):

    result = {}

    for label, coefficient in operator.to_list():

        result[label] = (
            result.get(label, 0.0j) + complex(coefficient)
        )

    return result


original_map = pauli_coefficient_map(
    original_hamiltonian
)

reconstructed_map = pauli_coefficient_map(
    reconstructed_hamiltonian
)

all_labels = set(original_map) | set(reconstructed_map)

coefficient_differences = {
    label: (
        reconstructed_map.get(label, 0.0j)
        - original_map.get(label, 0.0j)
    )
    for label in all_labels
}

coefficient_differences = {
    label: value
    for label, value in coefficient_differences.items()
    if abs(value) > 1e-9
}

max_difference = max(
    (abs(value) for value in coefficient_differences.values()),
    default=0.0
)

hamiltonian_consistent = len(coefficient_differences) == 0

print("Original Pauli terms:", len(original_map))
print("Reconstructed Pauli terms:", len(reconstructed_map))
print("Difference terms:", len(coefficient_differences))
print("Maximum coefficient difference:", max_difference)
print("Hamiltonian consistency:", hamiltonian_consistent)

if not hamiltonian_consistent:

    print("\nWARNING: HAMILTONIAN MISMATCH DETECTED")

    for label, difference_value in list(
        coefficient_differences.items()
    )[:20]:

        print(
            "Pauli:", label,
            "| Difference:", difference_value
        )

    print(
        "\nThe original Hamiltonian will NOT be overwritten."
    )

else:
    print("PASS - Hamiltonian representations match.")

# ------------------------------------------------------------
# 3. VERIFY REAL-VALUED COEFFICIENTS
# ------------------------------------------------------------

print("\n3. COEFFICIENT VALIDATION")

dictionary_coefficients = np.array(
    [complex(c) for c in ising_terms.values()],
    dtype=complex
)

hamiltonian_coefficients = np.asarray(
    ising_hamiltonian.coeffs,
    dtype=complex
)

coefficients_finite = (
    np.all(np.isfinite(dictionary_coefficients.real))
    and np.all(np.isfinite(dictionary_coefficients.imag))
    and np.all(np.isfinite(hamiltonian_coefficients.real))
    and np.all(np.isfinite(hamiltonian_coefficients.imag))
)

coefficients_real = (
    np.allclose(dictionary_coefficients.imag, 0.0, atol=1e-10)
    and np.allclose(hamiltonian_coefficients.imag, 0.0, atol=1e-10)
)

print("All coefficients finite:", coefficients_finite)
print("All coefficients real:", coefficients_real)

assert coefficients_finite, "Non-finite Hamiltonian coefficients."
assert coefficients_real, (
    "Complex coefficients detected. Inspect the Hamiltonian."
)

# ------------------------------------------------------------
# 4. REBUILD ORIGINAL QAOA CIRCUIT
# ------------------------------------------------------------

def build_qaoa_state(gamma, beta):

    circuit = QuantumCircuit(N)

    # Uniform superposition
    for q in range(N):
        circuit.h(q)

    # Cost Hamiltonian evolution
    for indices, coefficient in ising_terms.items():

        coefficient = float(np.real(coefficient))

        if len(indices) == 0:
            # Identity contributes only a global phase.
            continue

        elif len(indices) == 1:

            circuit.rz(
                2.0 * gamma * coefficient,
                indices[0]
            )

        elif len(indices) == 2:

            circuit.rzz(
                2.0 * gamma * coefficient,
                indices[0],
                indices[1]
            )

    # Mixer Hamiltonian evolution
    for q in range(N):
        circuit.rx(2.0 * beta, q)

    return circuit


print("\n4. CIRCUIT STRUCTURE CHECK")

test_circuit = build_qaoa_state(
    gamma=0.001,
    beta=0.1
)

print("Circuit qubits:", test_circuit.num_qubits)
print("Circuit operations:", test_circuit.count_ops())
print("Circuit construction: SUCCESS")

assert test_circuit.num_qubits == N

# ------------------------------------------------------------
# 5. ZERO-GAMMA ENERGY VERIFICATION
# ------------------------------------------------------------

identity_coefficient = float(
    np.real(
        sum(
            coefficient
            for indices, coefficient in ising_terms.items()
            if len(indices) == 0
        )
    )
)

beta_test_values = [
    0.0,
    0.1,
    0.5,
    1.0,
    np.pi / 2
]

zero_gamma_energies = []

print("\n5. ZERO-GAMMA VERIFICATION")
print("Identity coefficient:", identity_coefficient)

for beta in beta_test_values:

    circuit = build_qaoa_state(
        gamma=0.0,
        beta=beta
    )

    state = Statevector.from_instruction(circuit)

    energy = float(
        np.real(
            state.expectation_value(ising_hamiltonian)
        )
    )

    zero_gamma_energies.append(energy)

    print(
        f"Beta={beta:.6f} | Energy={energy:.10f}"
    )

zero_gamma_deviation = max(
    abs(energy - identity_coefficient)
    for energy in zero_gamma_energies
)

zero_gamma_verified = np.allclose(
    zero_gamma_energies,
    identity_coefficient,
    atol=1e-7
)

print(
    "Maximum zero-gamma deviation:",
    zero_gamma_deviation
)

print("Zero-gamma verified:", zero_gamma_verified)

# Do not stop the entire cell if this audit fails.
# Preserve the failure in the final report.

# ------------------------------------------------------------
# 6. EXACT PARAMETER GRID SCAN
# ------------------------------------------------------------

gamma_values = [
    0.0,
    0.0001,
    0.00025,
    0.0005,
    0.001,
    0.0025,
    0.005,
    0.0075,
    0.01
]

beta_values = [
    0.0,
    np.pi / 8,
    np.pi / 4,
    3 * np.pi / 8,
    np.pi / 2
]

scan_results = []

print("\n6. RUNNING EXACT STATEVECTOR SCAN")
print("Gamma candidates:", len(gamma_values))
print("Beta candidates:", len(beta_values))
print("Total combinations:", len(gamma_values) * len(beta_values))

for gamma in gamma_values:

    for beta in beta_values:

        circuit = build_qaoa_state(
            gamma=gamma,
            beta=beta
        )

        state = Statevector.from_instruction(circuit)

        energy = float(
            np.real(
                state.expectation_value(ising_hamiltonian)
            )
        )

        scan_results.append({
            "gamma": gamma,
            "beta": beta,
            "energy": energy
        })

scan_df = pd.DataFrame(scan_results)

assert len(scan_df) == 45

assert np.isfinite(
    scan_df["energy"].to_numpy()
).all()

# ------------------------------------------------------------
# 7. PARAMETER RANKING
# ------------------------------------------------------------

# This preserves the original ascending-energy convention.
# Confirm that the Ising objective is formulated for minimization.

scan_df = scan_df.sort_values(
    "energy",
    ascending=True
).reset_index(drop=True)

print("\n7. BEST PARAMETER COMBINATIONS")
print(scan_df.head(10).to_string(index=False))

best_scan = scan_df.iloc[0]

print("\n8. BEST GRID RESULT")
print("Gamma:", best_scan["gamma"])
print("Beta:", best_scan["beta"])
print("Expected energy:", best_scan["energy"])

# ------------------------------------------------------------
# 8. FINAL AUDIT
# ------------------------------------------------------------

checks = {
    "Hamiltonian has 20 qubits": N == 20,

    "No unsupported higher-order terms":
        max(term_degrees) <= 2,

    "Hamiltonian dictionary matches Qiskit":
        hamiltonian_consistent,

    "Hamiltonian coefficients finite":
        coefficients_finite,

    "Hamiltonian coefficients real":
        coefficients_real,

    "Circuit has 20 qubits":
        test_circuit.num_qubits == 20,

    "Zero-gamma energy verified":
        zero_gamma_verified,

    "Parameter scan completed":
        len(scan_df) == 45,

    "All scanned energies finite":
        np.isfinite(scan_df["energy"]).all()
}

print("\n9. FINAL AUDIT")

for name, passed in checks.items():

    print(
        ("PASS" if passed else "FAIL"),
        "-",
        name
    )

overall_pass = all(checks.values())

print(
    "\nOVERALL:",
    "PASS" if overall_pass else "INVESTIGATION REQUIRED"
)

# ------------------------------------------------------------
# 9. EXPORT RESULTS
# ------------------------------------------------------------

OUTPUT_DIR = "/content/ATMIQ_FINAL_RESULTS"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

scan_path = os.path.join(
    OUTPUT_DIR,
    "cell_135_qaoa_parameter_scan.csv"
)

scan_df.to_csv(
    scan_path,
    index=False
)

audit_report = {
    "cell": 135,
    "qubits": int(N),
    "ising_dictionary_terms": int(len(ising_terms)),
    "pauli_terms": int(len(ising_hamiltonian)),
    "maximum_interaction_degree": int(max(term_degrees)),
    "identity_terms": int(sum(d == 0 for d in term_degrees)),
    "single_qubit_terms": int(sum(d == 1 for d in term_degrees)),
    "two_qubit_terms": int(sum(d == 2 for d in term_degrees)),
    "higher_order_terms": int(sum(d > 2 for d in term_degrees)),
    "hamiltonian_consistent": bool(hamiltonian_consistent),
    "hamiltonian_max_coefficient_difference": float(max_difference),
    "hamiltonian_difference_terms": int(len(coefficient_differences)),
    "coefficients_finite": bool(coefficients_finite),
    "coefficients_real": bool(coefficients_real),
    "identity_coefficient": identity_coefficient,
    "zero_gamma_max_deviation": float(zero_gamma_deviation),
    "zero_gamma_verified": bool(zero_gamma_verified),
    "parameter_combinations": int(len(scan_df)),
    "best_gamma": float(best_scan["gamma"]),
    "best_beta": float(best_scan["beta"]),
    "best_energy": float(best_scan["energy"]),
    "optimization_direction": "minimize",
    "all_checks_passed": bool(overall_pass),
    "checks": {
        key: bool(value)
        for key, value in checks.items()
    }
}

report_path = os.path.join(
    OUTPUT_DIR,
    "cell_135_qaoa_circuit_audit.json"
)

with open(report_path, "w") as f:
    json.dump(
        audit_report,
        f,
        indent=4
    )

print("\n10. EXPORTED FILES")
print(scan_path)
print(report_path)

print("\nCELL 135 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 136B - INDEPENDENT QAOA STATEVECTOR VERIFICATION
# ============================================================================

import numpy as np
import pandas as pd
from qiskit.quantum_info import Statevector, SparsePauliOp

print("=" * 75)
print("CELL 136B - INDEPENDENT QAOA VERIFICATION")
print("=" * 75)

# 1. PARAMETER VERIFICATION

assert N == 20
assert len(ising_terms) == 186

gamma_value = float(optimized_gamma)
beta_value = float(optimized_beta)

print("\n1. OPTIMIZED PARAMETERS")
print("Gamma:", gamma_value)
print("Beta:", beta_value)
print("Grid energy:", optimized_energy)

# 2. RECONSTRUCT ORIGINAL CIRCUIT

verification_circuit = build_qaoa_state(
    gamma_value,
    beta_value
)

assert verification_circuit.num_qubits == N

print("\n2. CIRCUIT RECONSTRUCTION")
print("Qubits:", verification_circuit.num_qubits)
print("Operations:", verification_circuit.count_ops())

# 3. STATEVECTOR SIMULATION

statevector = Statevector.from_instruction(
    verification_circuit
)

probabilities = np.abs(statevector.data) ** 2

print("\n3. STATEVECTOR VERIFICATION")
print("Dimension:", len(statevector.data))
print("Total probability:", probabilities.sum())
print("Maximum probability:", probabilities.max())

assert len(probabilities) == 2 ** N
assert np.isclose(probabilities.sum(), 1.0)

# 4. RECONSTRUCT HAMILTONIAN

pauli_labels = []
pauli_coefficients = []

for indices, coefficient in ising_terms.items():

    label = ["I"] * N

    for qubit in indices:
        label[N - 1 - qubit] = "Z"

    pauli_labels.append("".join(label))
    pauli_coefficients.append(float(coefficient))

independent_hamiltonian = SparsePauliOp(
    pauli_labels,
    coeffs=np.asarray(pauli_coefficients, dtype=complex)
)

# 5. INDEPENDENT EXPECTED ENERGY

independent_energy = float(
    np.real(
        statevector.expectation_value(
            independent_hamiltonian
        )
    )
)

print("\n4. INDEPENDENT ENERGY VERIFICATION")
print("Cell 135 energy:", optimized_energy)
print("Reconstructed energy:", independent_energy)
print("Absolute difference:",
      abs(independent_energy - optimized_energy))

# 6. TOP 10 PROBABILITY STATES

top_indices = np.argsort(probabilities)[::-1][:10]

top_probability_df = pd.DataFrame({
    "state_index": top_indices,
    "bitstring": [
        format(i, f"0{N}b")
        for i in top_indices
    ],
    "probability": probabilities[top_indices]
})

print("\n5. TOP 10 QUANTUM STATES")
print(top_probability_df.to_string(index=False))

# 7. FINAL AUDIT

print("\n6. FINAL AUDIT")

print("PASS - Original circuit reconstructed")
print("PASS - 20-qubit statevector generated")
print("PASS - Probability normalization verified")
print("PASS - Ising Hamiltonian independently reconstructed")

if np.isclose(
    independent_energy,
    optimized_energy,
    rtol=1e-8,
    atol=1e-6
):
    print("PASS - Independent energy matches Cell 135")
else:
    print("WARNING - Energy mismatch detected")
    print("Check Hamiltonian convention and circuit construction.")

print("\nCELL 136B COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 138 - EXACT QAOA STATEVECTOR FEASIBILITY DECODING
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 138 - EXACT QAOA FEASIBILITY DECODING")
print("=" * 75)

# 1. VERIFY ORIGINAL ENCODING

assert N == 20
assert N_CUSTOMERS == 10
assert K_SELECTED == 6
assert len(quantum_candidates) == 10

customer_ids = quantum_candidates["customer_id"].tolist()

costs = np.asarray(quantum_costs, dtype=int)
capacities = np.asarray(quantum_capacities, dtype=int)

assert len(costs) == 10
assert len(capacities) == 10

budget_limit = 25
capacity_limit = 22

slack_weights = np.array([1, 2, 4, 8, 16])

print("\n1. ORIGINAL QUBIT MAPPING")
print("Customer qubits: 0-9")
print("Budget slack: 10-14")
print("Capacity slack: 15-19")
print("Customer IDs:", customer_ids)

# 2. DECODE ALL STATE INDICES

indices = np.arange(2**N, dtype=np.uint32)

selected_count_all = np.zeros(len(indices), dtype=np.int16)
selected_cost_all = np.zeros(len(indices), dtype=np.int32)
selected_capacity_all = np.zeros(len(indices), dtype=np.int32)

for q in range(10):
    bit_values = ((indices >> q) & 1).astype(np.int32)

    selected_count_all += bit_values
    selected_cost_all += bit_values * costs[q]
    selected_capacity_all += bit_values * capacities[q]

# 3. DECODE SLACK VARIABLES

budget_slack_all = np.zeros(len(indices), dtype=np.int32)
capacity_slack_all = np.zeros(len(indices), dtype=np.int32)

for j, weight in enumerate(slack_weights):
    budget_slack_all += (
        ((indices >> (10 + j)) & 1).astype(np.int32) * weight
    )

    capacity_slack_all += (
        ((indices >> (15 + j)) & 1).astype(np.int32) * weight
    )

print("\n2. DECODED STATE SPACE")
print("Total states:", len(indices))
print("Customer decision variables:", N_CUSTOMERS)
print("Budget slack maximum:", slack_weights.sum())
print("Capacity slack maximum:", slack_weights.sum())

# 4. APPLY ORIGINAL FEASIBILITY CONSTRAINTS

count_valid = selected_count_all == K_SELECTED

budget_valid = (
    selected_cost_all + budget_slack_all == budget_limit
)

capacity_valid = (
    selected_capacity_all + capacity_slack_all == capacity_limit
)

feasible_mask = count_valid & budget_valid & capacity_valid

feasible_indices = indices[feasible_mask]
feasible_probabilities = np.asarray(
    np.abs(statevector.data[feasible_indices])**2,
    dtype=float
)

print("\n3. FEASIBILITY RESULTS")
print("Count-valid states:", int(count_valid.sum()))
print("Budget-valid states:", int(budget_valid.sum()))
print("Capacity-valid states:", int(capacity_valid.sum()))
print("Fully feasible states:", len(feasible_indices))
print("Total feasible probability:",
      feasible_probabilities.sum())

# 5. RANK FEASIBLE QUANTUM STATES

ranking = np.argsort(feasible_probabilities)[::-1][:10]

decoded_records = []

for rank, position in enumerate(ranking, start=1):

    state_index = int(feasible_indices[position])

    selected = [
        customer_ids[q]
        for q in range(10)
        if (state_index >> q) & 1
    ]

    decoded_records.append({
        "rank": rank,
        "state_index": state_index,
        "bitstring": format(state_index, "020b"),
        "selected_customers": selected,
        "selected_count": int(selected_count_all[state_index]),
        "contact_cost": int(selected_cost_all[state_index]),
        "capacity": int(selected_capacity_all[state_index]),
        "budget_slack": int(budget_slack_all[state_index]),
        "capacity_slack": int(capacity_slack_all[state_index]),
        "probability": float(
            np.abs(statevector.data[state_index])**2
        )
    })

feasible_quantum_df = pd.DataFrame(decoded_records)

print("\n4. TOP 10 FEASIBLE QUANTUM SOLUTIONS")
print(feasible_quantum_df.to_string(index=False))

# 6. FINAL VALIDATION

print("\n5. FINAL AUDIT")

print("PASS - Original customer mapping preserved")
print("PASS - Original slack encoding preserved")
print("PASS - Exactly-six constraint checked")
print("PASS - Budget constraint checked")
print("PASS - Capacity constraint checked")
print("PASS - Exact statevector probabilities used")

print("\nTOTAL FEASIBLE STATES:", len(feasible_indices))
print("FEASIBLE PROBABILITY MASS:",
      feasible_probabilities.sum())

print("\nCELL 138 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 141 - EXHAUSTIVE QUANTUM VS CLASSICAL COMPARISON
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 141 - QUANTUM CLASSICAL OBJECTIVE COMPARISON")
print("=" * 75)

# --------------------------------------------------------------------------
# 1. VERIFY ORIGINAL OBJECTIVE
# --------------------------------------------------------------------------

assert len(quantum_values) == 10
assert len(quantum_costs) == 10
assert len(quantum_candidates) == 10

customer_ids = quantum_candidates["customer_id"].tolist()

print("\n1. ORIGINAL OBJECTIVE")

print("Objective: Maximize expected net value")
print("Expected net value = predicted probability * expected deposit - cost")

# Independently reconstruct quantum values

reconstructed_values = (
    quantum_candidates["predicted_probability"].to_numpy()
    * quantum_candidates["expected_deposit_value"].to_numpy()
    - quantum_candidates["contact_cost"].to_numpy()
)

print("Maximum coefficient difference:",
      np.max(np.abs(reconstructed_values - quantum_values)))

assert np.allclose(
    reconstructed_values,
    quantum_values,
    atol=1e-5
)

# --------------------------------------------------------------------------
# 2. EXHAUSTIVE EVALUATION OF FEASIBLE COMBINATIONS
# --------------------------------------------------------------------------

records = []

for state_index in feasible_indices:

    state_index = int(state_index)

    selected_indices_local = [
        q for q in range(N_CUSTOMERS)
        if (state_index >> q) & 1
    ]

    selected_ids = [
        customer_ids[q] for q in selected_indices_local
    ]

    expected_net_value = float(
        sum(quantum_values[q] for q in selected_indices_local)
    )

    contact_cost = int(
        sum(quantum_costs[q] for q in selected_indices_local)
    )

    capacity = int(
        sum(capacities[q] for q in selected_indices_local)
    )

    probability = float(
        np.abs(statevector.data[state_index])**2
    )

    records.append({
        "state_index": state_index,
        "selected_customers": selected_ids,
        "expected_net_value": expected_net_value,
        "contact_cost": contact_cost,
        "capacity": capacity,
        "probability": probability
    })

comparison_df = pd.DataFrame(records)

comparison_df = comparison_df.sort_values(
    "expected_net_value",
    ascending=False
).reset_index(drop=True)

comparison_df.insert(
    0, "rank",
    np.arange(1, len(comparison_df) + 1)
)

print("\n2. EXHAUSTIVE FEASIBLE SOLUTIONS")
print("Number evaluated:", len(comparison_df))

print("\nTOP 10 BY EXPECTED NET VALUE")

print(comparison_df.head(10).to_string(index=False))

# --------------------------------------------------------------------------
# 3. IDENTIFY BEST QUANTUM-SUPPORTED SOLUTION
# --------------------------------------------------------------------------

best_quantum = comparison_df.iloc[0]

print("\n3. BEST QUANTUM-SUPPORTED FEASIBLE SOLUTION")

print("Customers:", best_quantum["selected_customers"])
print("Expected net value:", best_quantum["expected_net_value"])
print("Contact cost:", best_quantum["contact_cost"])
print("Capacity:", best_quantum["capacity"])
print("State probability:", best_quantum["probability"])

# --------------------------------------------------------------------------
# 4. INDEPENDENT CLASSICAL BENCHMARK
# --------------------------------------------------------------------------

classical_ids = [
    "C0081", "C0031", "C0079",
    "C0327", "C0170", "C0349"
]

classical_indices = [
    customer_ids.index(cid) for cid in classical_ids
]

classical_objective = float(
    sum(quantum_values[q] for q in classical_indices)
)

classical_cost = int(
    sum(quantum_costs[q] for q in classical_indices)
)

classical_capacity = int(
    sum(capacities[q] for q in classical_indices)
)

print("\n4. CLASSICAL BENCHMARK")

print("Customers:", classical_ids)
print("Expected net value:", classical_objective)
print("Contact cost:", classical_cost)
print("Capacity:", classical_capacity)

# --------------------------------------------------------------------------
# 5. OPTIMALITY GAP
# --------------------------------------------------------------------------

quantum_objective = float(
    best_quantum["expected_net_value"]
)

if abs(classical_objective) > 1e-12:
    optimality_gap = (
        (classical_objective - quantum_objective)
        / abs(classical_objective)
    ) * 100
else:
    optimality_gap = np.nan

print("\n5. QUANTUM-CLASSICAL COMPARISON")

print("Classical objective:", classical_objective)
print("Best feasible quantum objective:", quantum_objective)
print("Absolute objective difference:",
      abs(classical_objective - quantum_objective))

print("Relative optimality gap (%):", optimality_gap)

# --------------------------------------------------------------------------
# 6. PROBABILITY OF MEASURING CLASSICAL OPTIMUM
# --------------------------------------------------------------------------

classical_state_indices = comparison_df[
    comparison_df["selected_customers"].apply(
        lambda x: set(x) == set(classical_ids)
    )
]

classical_success_probability = float(
    classical_state_indices["probability"].sum()
)

print("\n6. CLASSICAL OPTIMUM QUANTUM PROBABILITY")

print("Matching feasible states:", len(classical_state_indices))
print("Success probability:", classical_success_probability)
print("Success probability (%):",
      classical_success_probability * 100)

# --------------------------------------------------------------------------
# 7. FEASIBLE PROBABILITY AND CONDITIONAL SUCCESS
# --------------------------------------------------------------------------

total_feasible_probability = float(
    comparison_df["probability"].sum()
)

if total_feasible_probability > 0:
    conditional_success = (
        classical_success_probability
        / total_feasible_probability
    )
else:
    conditional_success = 0.0

print("\n7. PROBABILITY ANALYSIS")

print("Total feasible probability:",
      total_feasible_probability)

print("Conditional probability of classical optimum:",
      conditional_success)

print("Conditional probability (%):",
      conditional_success * 100)

# --------------------------------------------------------------------------
# 8. FINAL AUDIT
# --------------------------------------------------------------------------

print("\n8. FINAL AUDIT")

print("PASS - Original expected net value reconstructed")
print("PASS - All feasible combinations evaluated")
print("PASS - Classical benchmark independently rescored")
print("PASS - Quantum objective calculated")
print("PASS - Optimality gap calculated")
print("PASS - Quantum success probability calculated")

print("\nCELL 141 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 142 - QAOA SAMPLING QUALITY AND FEASIBLE-STATE PROBABILITY AUDIT
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 142 - QAOA SAMPLING QUALITY EVALUATION")
print("=" * 75)

# 1. Validate the exact feasible-state distribution
assert len(comparison_df) == 210
assert "probability" in comparison_df.columns
assert "expected_net_value" in comparison_df.columns

total_probability = float(np.sum(np.abs(statevector.data) ** 2))
feasible_probability = float(comparison_df["probability"].sum())
infeasible_probability = total_probability - feasible_probability

print("\n1. STATEVECTOR PROBABILITY")
print("Total statevector probability:", total_probability)
print("Feasible-state probability:", feasible_probability)
print("Infeasible-state probability:", infeasible_probability)
print("Feasible probability (%):", feasible_probability * 100)

# 2. Rank feasible states by measurement probability
probability_ranked = comparison_df.sort_values(
    "probability", ascending=False
).reset_index(drop=True)

probability_ranked.insert(
    0, "probability_rank", np.arange(1, len(probability_ranked) + 1)
)

print("\n2. TOP 10 MOST PROBABLE FEASIBLE STATES")
display_columns = [
    "probability_rank",
    "selected_customers",
    "probability",
    "expected_net_value",
    "contact_cost",
    "capacity"
]
print(probability_ranked[display_columns].head(10).to_string(index=False))

# 3. Evaluate the most probable feasible state
most_probable_feasible = probability_ranked.iloc[0]

most_probable_objective = float(
    most_probable_feasible["expected_net_value"]
)

print("\n3. MOST PROBABLE FEASIBLE SOLUTION")
print("Customers:", most_probable_feasible["selected_customers"])
print("Probability:", float(most_probable_feasible["probability"]))
print("Expected net value:", most_probable_objective)
print("Contact cost:", int(most_probable_feasible["contact_cost"]))
print("Capacity:", int(most_probable_feasible["capacity"]))

# 4. Compare most probable feasible state with the classical optimum
classical_objective = float(classical_objective)

sampling_objective_gap = (
    (classical_objective - most_probable_objective)
    / abs(classical_objective)
    * 100
    if abs(classical_objective) > 1e-12
    else np.nan
)

classical_probability = float(classical_success_probability)

classical_probability_rank = int(
    probability_ranked.index[
        probability_ranked["selected_customers"].apply(
            lambda x: set(x) == set(classical_ids)
        )
    ][0] + 1
)

print("\n4. CLASSICAL OPTIMUM SAMPLING RANK")
print("Classical optimum probability rank:", classical_probability_rank)
print("Classical optimum probability:", classical_probability)
print("Most probable feasible objective:", most_probable_objective)
print("Classical optimum objective:", classical_objective)
print("Sampling objective gap (%):", sampling_objective_gap)

# 5. Conditional expected objective among feasible measurements
conditional_expected_objective = (
    float(
        np.sum(
            comparison_df["expected_net_value"]
            * comparison_df["probability"]
        )
    ) / feasible_probability
    if feasible_probability > 0
    else np.nan
)

print("\n5. CONDITIONAL EXPECTED OBJECTIVE")
print(
    "Expected net value given a feasible measurement:",
    conditional_expected_objective
)

# 6. Cumulative probability captured by the top-ranked feasible states
print("\n6. CUMULATIVE FEASIBLE PROBABILITY")

for k in [1, 5, 10, 25, 50, 100, 210]:
    cumulative = float(
        probability_ranked["probability"].head(k).sum()
    )
    conditional_cumulative = (
        cumulative / feasible_probability
        if feasible_probability > 0
        else 0.0
    )

    print(
        f"Top {k:3d} feasible states: "
        f"probability={cumulative:.12e} | "
        f"conditional={conditional_cumulative * 100:.6f}%"
    )

# 7. Final audit
assert np.isclose(total_probability, 1.0, atol=1e-8)
assert np.isclose(
    feasible_probability + infeasible_probability,
    total_probability,
    atol=1e-10
)
assert 1 <= classical_probability_rank <= len(probability_ranked)

print("\n7. FINAL AUDIT")
print("PASS - Statevector probabilities normalized")
print("PASS - Feasible and infeasible probability mass reconciled")
print("PASS - Feasible states ranked by probability")
print("PASS - Classical optimum sampling rank identified")
print("PASS - Conditional expected objective calculated")

print("\nCELL 142 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 144 - QAOA PENALTY SENSITIVITY ANALYSIS
# CORRECTED NET-OBJECTIVE VERSION
# ============================================================================

import numpy as np
import pandas as pd
import os
import json
from itertools import combinations

print("=" * 75)
print("CELL 144 - QAOA PENALTY SENSITIVITY ANALYSIS")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. LOAD ORIGINAL QUBO INPUTS WITHOUT MODIFYING THEM
# ---------------------------------------------------------------------------

if all(name in globals() for name in [
    "quantum_values",
    "quantum_costs",
    "quantum_capacities"
]):

    sensitivity_values = np.asarray(
        quantum_values, dtype=float
    ).copy()

    sensitivity_costs = np.asarray(
        quantum_costs, dtype=float
    ).copy()

    sensitivity_capacities = np.asarray(
        quantum_capacities, dtype=float
    ).copy()

elif all(name in globals() for name in [
    "values", "costs", "capacities"
]):

    sensitivity_values = np.asarray(
        values, dtype=float
    ).copy()

    sensitivity_costs = np.asarray(
        costs, dtype=float
    ).copy()

    sensitivity_capacities = np.asarray(
        capacities, dtype=float
    ).copy()

else:

    raise NameError(
        "Original objective and constraint arrays are unavailable."
    )

N = len(sensitivity_values)

assert N == 10, f"Expected 10 customers, found {N}"

assert len(sensitivity_costs) == N
assert len(sensitivity_capacities) == N

assert np.all(np.isfinite(sensitivity_values))
assert np.all(np.isfinite(sensitivity_costs))
assert np.all(np.isfinite(sensitivity_capacities))

assert np.all(sensitivity_costs >= 0)
assert np.all(sensitivity_capacities >= 0)

# ---------------------------------------------------------------------------
# 2. ORIGINAL OPTIMIZATION CONFIGURATION
# ---------------------------------------------------------------------------

N_SLACK_BUDGET = 5
N_SLACK_CAPACITY = 5

TOTAL_QUBITS = (
    N + N_SLACK_BUDGET + N_SLACK_CAPACITY
)

K = 6

BUDGET_LIMIT = 25
CAPACITY_LIMIT = 22

slack_weights = 2 ** np.arange(5)

penalty_values = [
    1, 5, 10, 25, 50, 100, 200, 500
]

print("\n1. ORIGINAL OPTIMIZATION CONFIGURATION")

print("Customer qubits:", N)
print("Budget slack qubits:", N_SLACK_BUDGET)
print("Capacity slack qubits:", N_SLACK_CAPACITY)
print("Total qubits:", TOTAL_QUBITS)
print("Required customers:", K)
print("Budget limit:", BUDGET_LIMIT)
print("Capacity limit:", CAPACITY_LIMIT)
print("Penalty values:", penalty_values)

print("\nObjective definition:")
print("Net value = Expected value - Contact cost")

# ---------------------------------------------------------------------------
# 3. INDEPENDENT CLASSICAL BENCHMARK
# ---------------------------------------------------------------------------

print("\n2. VERIFYING CLASSICAL OPTIMUM")

classical_results = []

for selected in combinations(range(N), K):

    selected = np.array(selected, dtype=int)

    selected_value = float(
        sensitivity_values[selected].sum()
    )

    selected_cost = float(
        sensitivity_costs[selected].sum()
    )

    selected_capacity = float(
        sensitivity_capacities[selected].sum()
    )

    selected_net_value = (
        selected_value - selected_cost
    )

    feasible = (
        selected_cost <= BUDGET_LIMIT
        and selected_capacity <= CAPACITY_LIMIT
    )

    if feasible:

        classical_results.append({
            "Selected Indices": selected.tolist(),
            "Expected Value": selected_value,
            "Contact Cost": selected_cost,
            "Capacity": selected_capacity,
            "Net Objective": selected_net_value
        })

classical_df = pd.DataFrame(classical_results)

assert len(classical_df) > 0, (
    "No feasible classical solutions found."
)

classical_df = classical_df.sort_values(
    "Net Objective",
    ascending=False
).reset_index(drop=True)

CLASSICAL_OPTIMUM = float(
    classical_df.iloc[0]["Net Objective"]
)

CLASSICAL_OPTIMUM_INDICES = (
    classical_df.iloc[0]["Selected Indices"]
)

print("Feasible combinations:", len(classical_df))
print("Classical net optimum:", CLASSICAL_OPTIMUM)
print("Optimal customer indices:", CLASSICAL_OPTIMUM_INDICES)

print("\nClassical optimum details:")

print(
    classical_df.head(5).to_string(index=False)
)

# ---------------------------------------------------------------------------
# 4. VERIFY ENCODING DIMENSIONS
# ---------------------------------------------------------------------------

print("\n3. ENUMERATING ENCODED STATE SPACE")

total_states = 2 ** TOTAL_QUBITS

assert total_states == 1048576

all_encoded_states = np.arange(
    total_states,
    dtype=np.uint32
)

encoded_bits = np.column_stack([
    ((all_encoded_states >> i) & 1)
    for i in range(TOTAL_QUBITS)
]).astype(np.int8)

customer_bits = encoded_bits[:, :N]

budget_slack_bits = encoded_bits[
    :, N:N + N_SLACK_BUDGET
]

capacity_slack_bits = encoded_bits[
    :, N + N_SLACK_BUDGET:
]

selected_counts = customer_bits.sum(axis=1)

total_costs_sensitivity = (
    customer_bits @ sensitivity_costs
)

total_capacities_sensitivity = (
    customer_bits @ sensitivity_capacities
)

total_values_sensitivity = (
    customer_bits @ sensitivity_values
)

# Correct net objective:
total_net_values_sensitivity = (
    total_values_sensitivity
    - total_costs_sensitivity
)

budget_slack_values = (
    budget_slack_bits @ slack_weights
)

capacity_slack_values = (
    capacity_slack_bits @ slack_weights
)

budget_residuals = (
    total_costs_sensitivity
    + budget_slack_values
    - BUDGET_LIMIT
)

capacity_residuals = (
    total_capacities_sensitivity
    + capacity_slack_values
    - CAPACITY_LIMIT
)

cardinality_residuals = (
    selected_counts - K
)

print("Total encoded states:", total_states)
print("Customer combinations:", 2 ** N)
print("Encoded state enumeration completed.")

# ---------------------------------------------------------------------------
# 5. PENALTY SENSITIVITY ANALYSIS
# ---------------------------------------------------------------------------

print("\n4. TESTING PENALTY STRENGTHS")

sensitivity_results = []

for penalty_strength in penalty_values:

    encoded_energy = (
        -total_net_values_sensitivity
        + penalty_strength * cardinality_residuals ** 2
        + penalty_strength * budget_residuals ** 2
        + penalty_strength * capacity_residuals ** 2
    )

    best_energy = float(np.min(encoded_energy))

    # Identify all ground states, allowing numerical tolerance.
    ground_state_indices = np.flatnonzero(
        np.isclose(
            encoded_energy,
            best_energy,
            rtol=1e-10,
            atol=1e-9
        )
    )

    best_index = int(ground_state_indices[0])

    best_customer_bits = customer_bits[best_index]

    selected_indices = np.flatnonzero(
        best_customer_bits
    ).tolist()

    selected_count = int(
        selected_counts[best_index]
    )

    selected_cost = float(
        total_costs_sensitivity[best_index]
    )

    selected_capacity = float(
        total_capacities_sensitivity[best_index]
    )

    selected_value = float(
        total_values_sensitivity[best_index]
    )

    selected_net_value = float(
        total_net_values_sensitivity[best_index]
    )

    budget_residual = float(
        budget_residuals[best_index]
    )

    capacity_residual = float(
        capacity_residuals[best_index]
    )

    cardinality_residual = int(
        cardinality_residuals[best_index]
    )

    feasible = (
        selected_count == K
        and abs(budget_residual) < 1e-9
        and abs(capacity_residual) < 1e-9
    )

    ground_state_feasibility = []

    for index in ground_state_indices:

        is_feasible = (
            selected_counts[index] == K
            and abs(budget_residuals[index]) < 1e-9
            and abs(capacity_residuals[index]) < 1e-9
        )

        ground_state_feasibility.append(is_feasible)

    feasible_ground_states = int(
        sum(ground_state_feasibility)
    )

    if (
        "quantum_candidates" in globals()
        and "customer_id" in quantum_candidates.columns
        and len(quantum_candidates) == N
    ):

        candidate_ids = (
            quantum_candidates.iloc[selected_indices][
                "customer_id"
            ].tolist()
        )

    else:

        candidate_ids = selected_indices

    optimal_selection_found = (
        selected_indices == CLASSICAL_OPTIMUM_INDICES
    )

    sensitivity_results.append({

        "Penalty": penalty_strength,

        "Minimum Encoded Energy": best_energy,

        "Expected Value": selected_value,

        "Contact Cost": selected_cost,

        "Net Objective": selected_net_value,

        "Selected Count": selected_count,

        "Capacity": selected_capacity,

        "Cardinality Residual": cardinality_residual,

        "Budget Residual": budget_residual,

        "Capacity Residual": capacity_residual,

        "Feasible": feasible,

        "Ground State Degeneracy": len(
            ground_state_indices
        ),

        "Feasible Ground States": feasible_ground_states,

        "Classical Optimum Selected": optimal_selection_found,

        "Selected Indices": str(selected_indices),

        "Selected Customers": str(candidate_ids)
    })

    print(f"\nPenalty = {penalty_strength}")

    print("Minimum encoded energy:", round(best_energy, 8))

    print("Selected customers:", candidate_ids)

    print("Expected value:", round(selected_value, 8))

    print("Contact cost:", round(selected_cost, 8))

    print("Net objective:", round(selected_net_value, 8))

    print("Capacity:", selected_capacity)

    print("Selected count:", selected_count)

    print("Feasible:", feasible)

    print(
        "Ground state degeneracy:",
        len(ground_state_indices)
    )

    print(
        "Feasible ground states:",
        feasible_ground_states
    )

    print(
        "Classical optimum selected:",
        optimal_selection_found
    )

# ---------------------------------------------------------------------------
# 6. SENSITIVITY COMPARISON
# ---------------------------------------------------------------------------

sensitivity_df = pd.DataFrame(
    sensitivity_results
)

print("\n5. PENALTY SENSITIVITY SUMMARY")

summary_columns = [
    "Penalty",
    "Minimum Encoded Energy",
    "Expected Value",
    "Contact Cost",
    "Net Objective",
    "Selected Count",
    "Capacity",
    "Feasible"
]

print(
    sensitivity_df[
        summary_columns
    ].to_string(index=False)
)

# ---------------------------------------------------------------------------
# 7. EFFECTIVE PENALTY ANALYSIS
# ---------------------------------------------------------------------------

print("\n6. EFFECTIVE PENALTY ANALYSIS")

feasible_penalties = sensitivity_df.loc[
    sensitivity_df["Feasible"],
    "Penalty"
].tolist()

if feasible_penalties:

    print(
        "Penalties producing feasible ground states:",
        feasible_penalties
    )

    print(
        "Minimum tested feasible penalty:",
        min(feasible_penalties)
    )

else:

    print(
        "WARNING: No tested penalty produced a feasible ground state."
    )

# ---------------------------------------------------------------------------
# 8. CLASSICAL OPTIMUM COMPARISON
# ---------------------------------------------------------------------------

sensitivity_df["Gap From Classical Optimum (%)"] = (
    (
        CLASSICAL_OPTIMUM
        - sensitivity_df["Net Objective"]
    ) / max(abs(CLASSICAL_OPTIMUM), 1e-12)
) * 100

print("\n7. CLASSICAL OPTIMUM COMPARISON")

comparison_columns = [
    "Penalty",
    "Net Objective",
    "Gap From Classical Optimum (%)",
    "Feasible",
    "Classical Optimum Selected"
]

print(
    sensitivity_df[
        comparison_columns
    ].to_string(index=False)
)

# ---------------------------------------------------------------------------
# 9. EXPORT RESULTS
# ---------------------------------------------------------------------------

if "results_dir" not in globals():

    results_dir = (
        "/content/drive/MyDrive/ATMIQ_HACKATHON/Results"
    )

os.makedirs(
    results_dir,
    exist_ok=True
)

sensitivity_path = os.path.join(
    results_dir,
    "CELL144_QAOA_PENALTY_SENSITIVITY.csv"
)

classical_path = os.path.join(
    results_dir,
    "CELL144_CLASSICAL_BENCHMARK.csv"
)

report_path = os.path.join(
    results_dir,
    "CELL144_PENALTY_SENSITIVITY_REPORT.json"
)

sensitivity_df.to_csv(
    sensitivity_path,
    index=False
)

classical_df.to_csv(
    classical_path,
    index=False
)

report = {

    "cell": 144,

    "qubits": int(TOTAL_QUBITS),

    "customer_variables": int(N),

    "slack_qubits": int(
        N_SLACK_BUDGET + N_SLACK_CAPACITY
    ),

    "total_encoded_states": int(total_states),

    "required_customers": int(K),

    "budget_limit": float(BUDGET_LIMIT),

    "capacity_limit": float(CAPACITY_LIMIT),

    "objective": "expected_value - contact_cost",

    "classical_net_optimum": float(CLASSICAL_OPTIMUM),

    "classical_optimum_indices": CLASSICAL_OPTIMUM_INDICES,

    "feasible_classical_combinations": int(
        len(classical_df)
    ),

    "penalties_tested": penalty_values,

    "feasible_penalties": feasible_penalties,

    "all_penalty_tests_completed": (
        len(sensitivity_df) == len(penalty_values)
    ),

    "minimum_penalty": int(min(penalty_values)),

    "maximum_penalty": int(max(penalty_values)),

    "results_file": sensitivity_path
}

with open(report_path, "w") as f:

    json.dump(
        report,
        f,
        indent=4
    )

print("\n8. SAVED RESULTS")

print(sensitivity_path)
print(classical_path)
print(report_path)

# ---------------------------------------------------------------------------
# 10. FINAL AUDIT
# ---------------------------------------------------------------------------

checks = {

    "Original arrays preserved":
        len(sensitivity_values) == N
        and len(sensitivity_costs) == N
        and len(sensitivity_capacities) == N,

    "All 1,048,576 encoded states enumerated":
        total_states == 1048576,

    "Eight penalty strengths evaluated":
        len(sensitivity_df) == 8,

    "All energy values finite":
        np.isfinite(
            sensitivity_df[
                "Minimum Encoded Energy"
            ]
        ).all(),

    "All selected counts valid":
        sensitivity_df[
            "Selected Count"
        ].between(0, N).all(),

    "Classical benchmark independently calculated":
        len(classical_df) > 0,

    "Classical optimum is finite":
        np.isfinite(CLASSICAL_OPTIMUM),

    "Sensitivity CSV saved":
        os.path.exists(sensitivity_path),

    "Classical benchmark CSV saved":
        os.path.exists(classical_path),

    "JSON report saved":
        os.path.exists(report_path)
}

print("\n9. FINAL AUDIT")

for name, passed in checks.items():

    print(
        ("PASS" if passed else "FAIL"),
        "-",
        name
    )

overall_pass = all(checks.values())

print(
    "\nOVERALL:",
    "PASS" if overall_pass
    else "INVESTIGATION REQUIRED"
)

print("\nCELL 144 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 145 - EXACT QUBO PENALTY THRESHOLD ANALYSIS
# ============================================================================

import numpy as np
import pandas as pd
import os

print("=" * 75)
print("CELL 145 - EXACT QUBO PENALTY THRESHOLD ANALYSIS")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. VERIFY CELL 144 DEPENDENCIES
# ---------------------------------------------------------------------------

required_variables = [
    "sensitivity_values",
    "sensitivity_costs",
    "sensitivity_capacities",
    "cardinality_residuals",
    "budget_residuals",
    "capacity_residuals",
    "total_values_sensitivity",
    "customer_bits"
]

missing = [
    variable for variable in required_variables
    if variable not in globals()
]

if missing:
    raise NameError(
        "Missing Cell 144 variables. Run Cell 144 first: " + str(missing)
    )

CLASSICAL_OPTIMUM = 37.30689110902086

# ---------------------------------------------------------------------------
# 2. CALCULATE TOTAL CONSTRAINT VIOLATION
# ---------------------------------------------------------------------------

print("\n1. CONSTRAINT VIOLATION ANALYSIS")

violation_scores = (
    cardinality_residuals.astype(float) ** 2
    + budget_residuals.astype(float) ** 2
    + capacity_residuals.astype(float) ** 2
)

assert len(violation_scores) == 2 ** 20

feasible_mask = np.isclose(
    violation_scores,
    0.0,
    atol=1e-12
)

infeasible_mask = ~feasible_mask

print("Total encoded states:", len(violation_scores))
print("Feasible encoded states:", int(feasible_mask.sum()))
print("Infeasible encoded states:", int(infeasible_mask.sum()))

# ---------------------------------------------------------------------------
# 3. DERIVE THE EXACT PENALTY THRESHOLD
# ---------------------------------------------------------------------------

print("\n2. THEORETICAL PENALTY THRESHOLD")

# Energy = -objective + penalty * violation
#
# For an infeasible state not to beat the classical optimum:
#
# P >= (objective - classical optimum) / violation

candidate_ratios = (
    total_values_sensitivity[infeasible_mask]
    - CLASSICAL_OPTIMUM
) / violation_scores[infeasible_mask]

positive_ratios = np.maximum(candidate_ratios, 0.0)

exact_threshold = float(np.max(positive_ratios))

infeasible_indices = np.flatnonzero(infeasible_mask)

critical_position = int(np.argmax(candidate_ratios))

critical_state_index = int(
    infeasible_indices[critical_position]
)

critical_objective = float(
    total_values_sensitivity[critical_state_index]
)

critical_violation = float(
    violation_scores[critical_state_index]
)

critical_ratio = float(
    candidate_ratios[critical_position]
)

print("Theoretical minimum penalty threshold:",
      exact_threshold)

print("Critical encoded state index:", critical_state_index)
print("Critical state objective:", critical_objective)
print("Critical state violation score:", critical_violation)
print("Critical state penalty requirement:", critical_ratio)

# ---------------------------------------------------------------------------
# 4. INSPECT THE CRITICAL INFEASIBLE STATE
# ---------------------------------------------------------------------------

print("\n3. CRITICAL INFEASIBLE STATE")

critical_bits = customer_bits[critical_state_index]

critical_selected_indices = np.flatnonzero(critical_bits)

if (
    "quantum_candidates" in globals()
    and "customer_id" in quantum_candidates.columns
):
    critical_customer_ids = quantum_candidates.iloc[
        critical_selected_indices
    ]["customer_id"].tolist()
else:
    critical_customer_ids = critical_selected_indices.tolist()

print("Selected customers:", critical_customer_ids)
print("Selected count:", int(critical_bits.sum()))

print(
    "Contact cost:",
    float(total_costs_sensitivity[critical_state_index])
)

print(
    "Capacity:",
    float(total_capacities_sensitivity[critical_state_index])
)

print(
    "Cardinality residual:",
    int(cardinality_residuals[critical_state_index])
)

print(
    "Budget residual:",
    float(budget_residuals[critical_state_index])
)

print(
    "Capacity residual:",
    float(capacity_residuals[critical_state_index])
)

# ---------------------------------------------------------------------------
# 5. TEST PENALTIES AROUND THE EXACT THRESHOLD
# ---------------------------------------------------------------------------

print("\n4. FINE-GRAINED PENALTY TESTING")

test_penalties = sorted(set([
    1.0,
    4.0,
    5.0,
    10.0,
    100.0,
    exact_threshold * 0.90,
    exact_threshold * 0.99,
    exact_threshold,
    exact_threshold * 1.001,
    exact_threshold * 1.01,
    exact_threshold * 1.05,
    exact_threshold * 1.10
]))

fine_results = []

for penalty in test_penalties:

    energies = (
        -total_values_sensitivity
        + penalty * violation_scores
    )

    minimum_energy = float(np.min(energies))

    minimum_indices = np.flatnonzero(
        np.isclose(
            energies,
            minimum_energy,
            rtol=1e-10,
            atol=1e-9
        )
    )

    ground_state_feasible = bool(
        np.all(feasible_mask[minimum_indices])
    )

    best_index = int(minimum_indices[0])

    fine_results.append({
        "Penalty": penalty,
        "Minimum Energy": minimum_energy,
        "Number of Ground States": len(minimum_indices),
        "Ground State Feasible": ground_state_feasible,
        "Objective Value": float(
            total_values_sensitivity[best_index]
        ),
        "Violation Score": float(
            violation_scores[best_index]
        )
    })

fine_df = pd.DataFrame(fine_results)

print(
    fine_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.10f}"
    )
)

# ---------------------------------------------------------------------------
# 6. COMPARE ORIGINAL PENALTY WITH THRESHOLD
# ---------------------------------------------------------------------------

print("\n5. ORIGINAL PENALTY EFFICIENCY")

ORIGINAL_PENALTY = 100.0

penalty_multiple = (
    ORIGINAL_PENALTY / exact_threshold
    if exact_threshold > 0 else np.inf
)

print("Exact threshold:", exact_threshold)
print("Original penalty:", ORIGINAL_PENALTY)
print("Original penalty / threshold:", penalty_multiple)

if ORIGINAL_PENALTY > exact_threshold:
    print("PASS - Original penalty exceeds the threshold.")
else:
    print("WARNING - Original penalty does not exceed the threshold.")

# ---------------------------------------------------------------------------
# 7. SAVE RESULTS
# ---------------------------------------------------------------------------

if "results_dir" not in globals():
    results_dir = "/content/drive/MyDrive/ATMIQ_HACKATHON/Results"

os.makedirs(results_dir, exist_ok=True)

threshold_path = os.path.join(
    results_dir,
    "CELL145_EXACT_PENALTY_THRESHOLD.csv"
)

fine_path = os.path.join(
    results_dir,
    "CELL145_FINE_GRAINED_PENALTY_TEST.csv"
)

threshold_df = pd.DataFrame([{
    "Theoretical Penalty Threshold": exact_threshold,
    "Critical State Index": critical_state_index,
    "Critical Objective": critical_objective,
    "Critical Violation": critical_violation,
    "Original Penalty": ORIGINAL_PENALTY,
    "Original Penalty Multiple": penalty_multiple,
    "Classical Optimum": CLASSICAL_OPTIMUM
}])

threshold_df.to_csv(threshold_path, index=False)
fine_df.to_csv(fine_path, index=False)

print("\n6. SAVED RESULTS")
print(threshold_path)
print(fine_path)

# ---------------------------------------------------------------------------
# 8. FINAL AUDIT
# ---------------------------------------------------------------------------

assert exact_threshold >= 0

assert np.isfinite(exact_threshold)

assert critical_violation > 0

assert np.isclose(
    critical_ratio,
    exact_threshold,
    atol=1e-10
)

assert len(fine_df) == len(test_penalties)

assert os.path.exists(threshold_path)
assert os.path.exists(fine_path)

print("\n7. FINAL AUDIT")
print("PASS - All encoded states evaluated")
print("PASS - Feasible and infeasible states separated")
print("PASS - Exact penalty threshold derived")
print("PASS - Critical infeasible state identified")
print("PASS - Fine-grained penalty sweep completed")
print("PASS - Original penalty compared")
print("PASS - CSV reports saved")

print("\nCELL 145 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 146 - QAOA GROUND-STATE STABILITY AND ENERGY-GAP ANALYSIS
# ============================================================================

import numpy as np
import pandas as pd
import os

print("=" * 75)
print("CELL 146 - GROUND-STATE STABILITY ANALYSIS")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. VERIFY DEPENDENCIES
# ---------------------------------------------------------------------------

required = [
    "total_values_sensitivity",
    "violation_scores",
    "feasible_mask",
    "customer_bits",
    "exact_threshold"
]

missing = [x for x in required if x not in globals()]

if missing:
    raise NameError(
        f"Run Cells 144 and 145 first. Missing: {missing}"
    )

CLASSICAL_OPTIMUM = 37.30689110902086

assert len(total_values_sensitivity) == 1048576
assert len(violation_scores) == 1048576
assert len(feasible_mask) == 1048576

# ---------------------------------------------------------------------------
# 2. DEFINE PENALTY VALUES
# ---------------------------------------------------------------------------

print("\n1. PENALTY CONFIGURATION")

penalty_test_values = [
    1.80,
    2.00,
    3.00,
    5.00,
    10.00,
    25.00,
    100.00
]

print("Theoretical threshold:", exact_threshold)
print("Test penalties:", penalty_test_values)

# ---------------------------------------------------------------------------
# 3. EVALUATE ENERGY LANDSCAPE
# ---------------------------------------------------------------------------

print("\n2. EXACT ENERGY LANDSCAPE ANALYSIS")

stability_results = []

for penalty in penalty_test_values:

    energies = (
        -total_values_sensitivity
        + penalty * violation_scores
    )

    # Global ground state
    ground_energy = float(np.min(energies))

    ground_indices = np.flatnonzero(
        np.isclose(
            energies,
            ground_energy,
            rtol=1e-10,
            atol=1e-9
        )
    )

    # Distinct energy levels
    sorted_energies = np.sort(energies)

    distinct_energies = np.unique(
        np.round(sorted_energies, 9)
    )

    second_energy = (
        float(distinct_energies[1])
        if len(distinct_energies) > 1
        else ground_energy
    )

    energy_gap = second_energy - ground_energy

    # Feasible and infeasible energy minima
    feasible_minimum = float(
        np.min(energies[feasible_mask])
    )

    infeasible_minimum = float(
        np.min(energies[~feasible_mask])
    )

    # Feasible ground-state verification
    ground_is_feasible = bool(
        np.all(feasible_mask[ground_indices])
    )

    # Ground-state customer selections
    best_index = int(ground_indices[0])

    selected_indices = np.flatnonzero(
        customer_bits[best_index]
    )

    if (
        "quantum_candidates" in globals()
        and "customer_id" in quantum_candidates.columns
    ):
        selected_customers = quantum_candidates.iloc[
            selected_indices
        ]["customer_id"].tolist()
    else:
        selected_customers = selected_indices.tolist()

    stability_results.append({
        "Penalty": penalty,
        "Ground Energy": ground_energy,
        "Second Distinct Energy": second_energy,
        "Energy Gap": energy_gap,
        "Ground State Degeneracy": len(ground_indices),
        "Feasible Ground State": ground_is_feasible,
        "Minimum Feasible Energy": feasible_minimum,
        "Minimum Infeasible Energy": infeasible_minimum,
        "Infeasible Energy Separation": (
            infeasible_minimum - feasible_minimum
        ),
        "Selected Customers": str(selected_customers)
    })

    print(f"\nPenalty: {penalty}")
    print("Ground energy:", round(ground_energy, 9))
    print("Second distinct energy:", round(second_energy, 9))
    print("Energy gap:", round(energy_gap, 9))
    print("Ground-state degeneracy:", len(ground_indices))
    print("Ground state feasible:", ground_is_feasible)
    print("Minimum feasible energy:", round(feasible_minimum, 9))
    print("Minimum infeasible energy:", round(infeasible_minimum, 9))
    print(
        "Infeasible energy separation:",
        round(infeasible_minimum - feasible_minimum, 9)
    )
    print("Selected customers:", selected_customers)

# ---------------------------------------------------------------------------
# 4. CREATE COMPARISON TABLE
# ---------------------------------------------------------------------------

print("\n3. GROUND-STATE STABILITY SUMMARY")

stability_df = pd.DataFrame(stability_results)

print(
    stability_df[
        [
            "Penalty",
            "Ground Energy",
            "Energy Gap",
            "Ground State Degeneracy",
            "Feasible Ground State",
            "Infeasible Energy Separation"
        ]
    ].to_string(index=False)
)

# ---------------------------------------------------------------------------
# 5. EVALUATE PENALTY ROBUSTNESS
# ---------------------------------------------------------------------------

print("\n4. PENALTY ROBUSTNESS ANALYSIS")

for _, row in stability_df.iterrows():

    if row["Feasible Ground State"]:
        status = "FEASIBLE"
    else:
        status = "INFEASIBLE"

    if row["Ground State Degeneracy"] > 1:
        status += " - DEGENERATE"

    print(
        f"Penalty {row['Penalty']:>7.2f}: "
        f"{status}"
    )

robust_penalties = stability_df[
    stability_df["Feasible Ground State"]
    & (stability_df["Ground State Degeneracy"] == 1)
]["Penalty"].tolist()

print("\nPenalties with unique feasible ground states:")
print(robust_penalties)

# ---------------------------------------------------------------------------
# 6. QUANTIFY ENERGY-GAP IMPROVEMENT
# ---------------------------------------------------------------------------

print("\n5. ENERGY-GAP COMPARISON")

baseline_gap = float(
    stability_df.loc[
        stability_df["Penalty"] == 100.0,
        "Energy Gap"
    ].iloc[0]
)

for _, row in stability_df.iterrows():

    gap = float(row["Energy Gap"])

    print(
        f"Penalty {row['Penalty']:>7.2f} | "
        f"Gap = {gap:.9f} | "
        f"Relative to penalty 100 = "
        f"{(gap / baseline_gap * 100) if baseline_gap > 0 else 0:.4f}%"
    )

# ---------------------------------------------------------------------------
# 7. SAVE RESULTS
# ---------------------------------------------------------------------------

if "results_dir" not in globals():
    results_dir = "/content/drive/MyDrive/ATMIQ_HACKATHON/Results"

os.makedirs(results_dir, exist_ok=True)

stability_path = os.path.join(
    results_dir,
    "CELL146_GROUND_STATE_STABILITY.csv"
)

stability_df.to_csv(stability_path, index=False)

print("\n6. SAVED RESULTS")
print(stability_path)

# ---------------------------------------------------------------------------
# 8. FINAL AUDIT
# ---------------------------------------------------------------------------

assert len(stability_df) == len(penalty_test_values)

assert stability_df["Feasible Ground State"].all()

assert stability_df["Ground State Degeneracy"].eq(1).all()

assert np.all(
    stability_df["Energy Gap"].to_numpy() > 0
)

assert np.all(
    stability_df["Infeasible Energy Separation"].to_numpy() > 0
)

assert os.path.exists(stability_path)

print("\n7. FINAL AUDIT")
print("PASS - All penalty landscapes evaluated")
print("PASS - Ground states identified")
print("PASS - Energy gaps calculated")
print("PASS - Ground-state degeneracy evaluated")
print("PASS - Feasible and infeasible energy separation calculated")
print("PASS - All tested penalties produce unique feasible ground states")
print("PASS - Stability report saved")

print("\nCELL 146 COMPLETE")
print("=" * 75)
```

```python

# ============================================================================
# CELL 148 - BINARY-TO-ISING MAPPING AND OBJECTIVE AUDIT
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 148 - BINARY-TO-ISING MAPPING AND OBJECTIVE AUDIT")
print("=" * 75)

# 1. IMPORTANT VARIABLE INVENTORY

print("\n1. AVAILABLE MODEL VARIABLES")

candidate_names = [
    "values",
    "costs",
    "capacities",
    "quantum_values",
    "quantum_costs",
    "quantum_capacities",
    "sensitivity_values",
    "sensitivity_costs",
    "sensitivity_capacities",
    "customer_bits",
    "budget_slack_bits",
    "capacity_slack_bits",
    "N",
    "N_QUBITS_TOTAL",
    "N_SLACK_BUDGET",
    "N_SLACK_CAPACITY",
    "K",
    "BUDGET_LIMIT",
    "CAPACITY_LIMIT",
    "PENALTY",
    "exact_threshold"
]

for name in candidate_names:

    if name in globals():

        obj = globals()[name]

        if isinstance(obj, np.ndarray):
            print(
                f"{name}: ndarray, shape={obj.shape}, "
                f"dtype={obj.dtype}"
            )

        elif isinstance(obj, (list, tuple)):
            print(
                f"{name}: {type(obj).__name__}, "
                f"length={len(obj)}, sample={obj[:5]}"
            )

        else:
            print(f"{name}: {obj}")

    else:
        print(f"{name}: NOT AVAILABLE")


# 2. IDENTIFY CUSTOMER AND SLACK VARIABLE ARRAYS

print("\n2. ARRAY CONTENT CHECK")

for name in [
    "values",
    "costs",
    "capacities",
    "quantum_values",
    "quantum_costs",
    "quantum_capacities",
    "sensitivity_values",
    "sensitivity_costs",
    "sensitivity_capacities"
]:

    if name in globals():

        arr = np.asarray(globals()[name])

        if arr.ndim == 1:
            print(f"\n{name}")
            print("Length:", len(arr))
            print("First values:", arr[:10])


# 3. ISING TERM STRUCTURE

print("\n3. ISING TERM STRUCTURE")

term_sizes = pd.Series(
    [len(term) for term in ising_terms.keys()]
).value_counts().sort_index()

print("Term-size distribution:")
print(term_sizes)

print("Identity coefficient:",
      float(ising_terms.get((), 0.0)))

print("Single-qubit terms:",
      sum(len(term) == 1 for term in ising_terms))

print("Two-qubit terms:",
      sum(len(term) == 2 for term in ising_terms))

# 4. QUBIT INDEX MAPPING

print("\n4. QUBIT INDEX MAPPING")

mapping_rows = []

for q in range(N_QUBITS_TOTAL):

    if q < 10:
        role = "Customer decision variable"
    elif q < 15:
        role = "Budget slack variable"
    else:
        role = "Capacity slack variable"

    mapping_rows.append({
        "qubit_index": q,
        "expected_role": role
    })

mapping_df = pd.DataFrame(mapping_rows)

print(mapping_df.to_string(index=False))


# 5. RECONSTRUCT ISING ENERGY FOR COMPUTATIONAL BASIS STATES

print("\n5. ISING ENERGY CONVENTION")

# For a computational basis state:
# x_q = 0 -> Z_q = +1
# x_q = 1 -> Z_q = -1

def calculate_ising_energy_from_bits(bit_values):

    bit_values = np.asarray(bit_values, dtype=int)

    if len(bit_values) != N_QUBITS_TOTAL:
        raise ValueError("Incorrect bitstring length")

    z_values = 1 - 2 * bit_values

    energy = 0.0

    for indices, coefficient in ising_terms.items():

        if len(indices) == 0:
            energy += coefficient

        elif len(indices) == 1:
            energy += coefficient * z_values[indices[0]]

        elif len(indices) == 2:
            energy += (
                coefficient
                * z_values[indices[0]]
                * z_values[indices[1]]
            )

    return float(energy)


# All-zero and all-one diagnostic states
zero_bits = np.zeros(N_QUBITS_TOTAL, dtype=int)
one_bits = np.ones(N_QUBITS_TOTAL, dtype=int)

print("All-zero bitstring energy:",
      calculate_ising_energy_from_bits(zero_bits))

print("All-one bitstring energy:",
      calculate_ising_energy_from_bits(one_bits))


# 6. CHECK ISING ENERGY AGAINST SPARSEPAULIOP

print("\n6. SPARSEPAULIOP CONSISTENCY")

from qiskit.quantum_info import Statevector

test_indices = [0, 1, 12345, 543210, 1048575]

mapping_errors = []

for state_index in test_indices:

    bits = np.array([
        (state_index >> q) & 1
        for q in range(N_QUBITS_TOTAL)
    ], dtype=int)

    energy_dict = calculate_ising_energy_from_bits(bits)

    basis_state = Statevector.from_int(
        state_index,
        2**N_QUBITS_TOTAL
    )

    energy_operator = float(np.real(
        basis_state.expectation_value(ising_hamiltonian)
    ))

    error = abs(energy_dict - energy_operator)
    mapping_errors.append(error)

    print(
        f"State {state_index}: "
        f"dictionary={energy_dict:.8f}, "
        f"operator={energy_operator:.8f}, "
        f"error={error:.3e}"
    )

max_mapping_error = max(mapping_errors)

print("\nMaximum mapping error:", max_mapping_error)


# 7. FINAL AUDIT

print("\n7. FINAL AUDIT")

audit_148 = {
    "20_qubits_confirmed": N_QUBITS_TOTAL == 20,
    "186_ising_terms_confirmed": len(ising_terms) == 186,
    "identity_term_present": () in ising_terms,
    "all_indices_in_range": max(
        q for term in ising_terms for q in term
    ) < N_QUBITS_TOTAL,
    "ising_operator_mapping": max_mapping_error < 1e-7
}

for name, passed in audit_148.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(audit_148.values())

print("\nCELL 148 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 149 - RECONSTRUCT PENALTY-DEPENDENT ISING HAMILTONIAN
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 149 - PENALTY-DEPENDENT ISING HAMILTONIAN RECONSTRUCTION")
print("=" * 75)

# 1. ORIGINAL MODEL INPUTS

v = np.asarray(quantum_values, dtype=float)
c = np.asarray(quantum_costs, dtype=float)
cap = np.asarray(quantum_capacities, dtype=float)

assert len(v) == N
assert len(c) == N
assert len(cap) == N

assert N == 10
assert N_QUBITS_TOTAL == 20

# Variable layout
customer_indices = list(range(0, 10))
budget_slack_indices = list(range(10, 15))
capacity_slack_indices = list(range(15, 20))

budget_weights = 2 ** np.arange(N_SLACK_BUDGET)
capacity_weights = 2 ** np.arange(N_SLACK_CAPACITY)

assert len(budget_weights) == 5
assert len(capacity_weights) == 5

# 2. GENERIC QUBO-TO-ISING CONVERTER

def build_penalty_ising(penalty):

    penalty = float(penalty)

    if penalty < 0:
        raise ValueError("Penalty must be non-negative.")

    n = N_QUBITS_TOTAL

    # QUBO:
    # E(x) = constant + sum(linear[i]*x[i])
    #        + sum(pair[i,j]*x[i]*x[j])

    constant = 0.0
    linear = np.zeros(n, dtype=float)
    pair = {}

    def add_squared_constraint(indices, weights, target):

        nonlocal constant, linear, pair

        target = float(target)
        weights = np.asarray(weights, dtype=float)

        # (sum(a_i*x_i) - target)^2
        constant += penalty * target**2

        for idx, weight in zip(indices, weights):

            linear[idx] += penalty * (
                weight**2 - 2.0 * target * weight
            )

        for i in range(len(indices)):

            for j in range(i + 1, len(indices)):

                qi = indices[i]
                qj = indices[j]

                key = (min(qi, qj), max(qi, qj))

                pair[key] = pair.get(key, 0.0) + (
                    2.0 * penalty
                    * weights[i]
                    * weights[j]
                )

    # Cardinality constraint
    cardinality_indices = customer_indices
    cardinality_weights = np.ones(N)

    add_squared_constraint(
        cardinality_indices,
        cardinality_weights,
        K
    )

    # Budget constraint
    budget_indices = (
        customer_indices + budget_slack_indices
    )

    budget_all_weights = np.concatenate([
        c,
        budget_weights.astype(float)
    ])

    add_squared_constraint(
        budget_indices,
        budget_all_weights,
        BUDGET_LIMIT
    )

    # Capacity constraint
    capacity_indices = (
        customer_indices + capacity_slack_indices
    )

    capacity_all_weights = np.concatenate([
        cap,
        capacity_weights.astype(float)
    ])

    add_squared_constraint(
        capacity_indices,
        capacity_all_weights,
        CAPACITY_LIMIT
    )

    # Original objective: maximize expected value.
    # Convert maximization into minimization.
    for i in customer_indices:
        linear[i] -= v[i]

    # QUBO -> Ising:
    # x_i = (1 - Z_i)/2

    ising = {}

    # Identity coefficient
    identity = constant

    identity += np.sum(linear) / 2.0

    for (i, j), coefficient in pair.items():
        identity += coefficient / 4.0

    ising[()] = float(identity)

    # Single-qubit Z coefficients
    for i in range(n):

        z_coefficient = -linear[i] / 2.0

        for (a, b), coefficient in pair.items():

            if a == i or b == i:
                z_coefficient -= coefficient / 4.0

        if abs(z_coefficient) > 1e-12:
            ising[(i,)] = float(z_coefficient)

    # Two-qubit ZZ coefficients
    for (i, j), coefficient in pair.items():

        zz_coefficient = coefficient / 4.0

        if abs(zz_coefficient) > 1e-12:
            ising[(i, j)] = float(zz_coefficient)

    return ising


# 3. RECONSTRUCT ORIGINAL PENALTY-100 HAMILTONIAN

print("\n1. RECONSTRUCTING ORIGINAL PENALTY")

reconstructed_ising_100 = build_penalty_ising(PENALTY)

print("Original penalty:", PENALTY)
print("Original terms:", len(ising_terms))
print("Reconstructed terms:", len(reconstructed_ising_100))

# 4. TERM-BY-TERM COMPARISON

print("\n2. ORIGINAL HAMILTONIAN COMPARISON")

all_terms = sorted(
    set(ising_terms.keys()) |
    set(reconstructed_ising_100.keys()),
    key=lambda x: (len(x), x)
)

comparison_rows = []

for term in all_terms:

    original = float(ising_terms.get(term, 0.0))
    rebuilt = float(reconstructed_ising_100.get(term, 0.0))

    comparison_rows.append({
        "term": str(term),
        "original": original,
        "reconstructed": rebuilt,
        "absolute_error": abs(original - rebuilt)
    })

hamiltonian_comparison_df = pd.DataFrame(comparison_rows)

max_hamiltonian_error = float(
    hamiltonian_comparison_df["absolute_error"].max()
)

print(
    hamiltonian_comparison_df.head(15).to_string(index=False)
)

print("\nMaximum coefficient error:", max_hamiltonian_error)

print(
    "Matching term count:",
    sum(
        np.isclose(
            ising_terms.get(term, 0.0),
            reconstructed_ising_100.get(term, 0.0),
            atol=1e-8
        )
        for term in all_terms
    )
)

# 5. RECONSTRUCTION SUMMARY

print("\n3. HAMILTONIAN SUMMARY")

for label, terms in [
    ("Original", ising_terms),
    ("Reconstructed", reconstructed_ising_100)
]:

    print(f"\n{label}")
    print("Identity:", terms.get((), 0.0))
    print("Single-qubit terms:",
          sum(len(t) == 1 for t in terms))
    print("Two-qubit terms:",
          sum(len(t) == 2 for t in terms))

# 6. TEST DIFFERENT PENALTIES

print("\n4. ALTERNATIVE PENALTY HAMILTONIANS")

penalties_to_test = [
    1.8,
    2.0,
    3.0,
    5.0,
    10.0,
    25.0,
    50.0,
    100.0
]

penalty_hamiltonian_summary = []

for p in penalties_to_test:

    terms = build_penalty_ising(p)

    penalty_hamiltonian_summary.append({
        "penalty": p,
        "term_count": len(terms),
        "identity_coefficient": terms.get((), 0.0),
        "single_qubit_terms": sum(
            len(t) == 1 for t in terms
        ),
        "two_qubit_terms": sum(
            len(t) == 2 for t in terms
        )
    })

penalty_hamiltonian_df = pd.DataFrame(
    penalty_hamiltonian_summary
)

print(penalty_hamiltonian_df.to_string(index=False))

# 7. FINAL AUDIT

print("\n5. FINAL AUDIT")

audit_149 = {
    "reconstructed_term_count": (
        len(reconstructed_ising_100) == len(ising_terms)
    ),
    "maximum_coefficient_error": (
        max_hamiltonian_error < 1e-7
    ),
    "original_penalty_preserved": PENALTY == 100.0,
    "20_qubit_reconstruction": (
        max(
            q for term in reconstructed_ising_100
            for q in term
        ) < N_QUBITS_TOTAL
    ),
    "all_penalty_values_reconstructed": (
        len(penalty_hamiltonian_df) == len(penalties_to_test)
    )
}

for name, passed in audit_149.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(audit_149.values())

print("\nCELL 149 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 152 - QAOA CIRCUIT DIMENSION CORRECTION
# FULLY CORRECTED 20-QUBIT VERSION
# ============================================================================

import numpy as np
from qiskit import QuantumCircuit

print("=" * 75)
print("CELL 152 - QAOA CIRCUIT DIMENSION CORRECTION")
print("=" * 75)

# ---------------------------------------------------------------------------
# 1. VERIFY ORIGINAL DIMENSIONS
# ---------------------------------------------------------------------------

print("\n1. ORIGINAL DIMENSION")

assert "ising_hamiltonian" in globals()
assert "ising_terms" in globals()

HAMILTONIAN_QUBITS = ising_hamiltonian.num_qubits

assert HAMILTONIAN_QUBITS == 20, (
    f"Expected 20 Hamiltonian qubits, found {HAMILTONIAN_QUBITS}"
)

N_QUBITS_TOTAL = HAMILTONIAN_QUBITS

print("Customer variables (N):", N)
print("Total QAOA qubits:", N_QUBITS_TOTAL)
print("Hamiltonian qubits:", HAMILTONIAN_QUBITS)
print("Ising dictionary terms:", len(ising_terms))

# ---------------------------------------------------------------------------
# 2. VERIFY ISING INDICES AND TERM DEGREES
# ---------------------------------------------------------------------------

print("\n2. ISING INDEX VALIDATION")

all_indices = [
    index
    for indices in ising_terms.keys()
    for index in indices
]

assert len(ising_terms) > 0

if all_indices:

    print("Minimum qubit index:", min(all_indices))
    print("Maximum qubit index:", max(all_indices))

    assert min(all_indices) >= 0
    assert max(all_indices) < N_QUBITS_TOTAL

else:
    raise ValueError("No non-identity Ising indices found.")

term_degrees = [
    len(indices)
    for indices in ising_terms.keys()
]

identity_count = sum(
    degree == 0 for degree in term_degrees
)

single_count = sum(
    degree == 1 for degree in term_degrees
)

two_count = sum(
    degree == 2 for degree in term_degrees
)

higher_count = sum(
    degree > 2 for degree in term_degrees
)

print("Identity terms:", identity_count)
print("Single-qubit terms:", single_count)
print("Two-qubit terms:", two_count)
print("Higher-order terms:", higher_count)

assert higher_count == 0, (
    "Unsupported higher-order Ising terms detected."
)

assert identity_count <= 1, (
    "Unexpected multiple identity entries in Ising dictionary."
)

# ---------------------------------------------------------------------------
# 3. VERIFY OPTIMIZATION PARAMETERS
# ---------------------------------------------------------------------------

print("\n3. OPTIMIZATION PARAMETER VALIDATION")

assert "optimized_gamma" in globals()
assert "optimized_beta" in globals()

gamma_value = float(optimized_gamma)
beta_value = float(optimized_beta)

assert np.isfinite(gamma_value)
assert np.isfinite(beta_value)

print("Gamma:", gamma_value)
print("Beta:", beta_value)

if "optimized_energy" in globals():

    energy_value = float(optimized_energy)

    print("Optimization energy:", energy_value)

else:

    energy_value = None

    print("Optimization energy: NOT AVAILABLE")

# ---------------------------------------------------------------------------
# 4. REBUILD QAOA CIRCUIT
# ---------------------------------------------------------------------------

print("\n4. REBUILDING QAOA FUNCTION")

def build_qaoa_state(gamma, beta):

    gamma = float(gamma)
    beta = float(beta)

    if not np.isfinite(gamma) or not np.isfinite(beta):
        raise ValueError("Non-finite QAOA parameters.")

    circuit = QuantumCircuit(N_QUBITS_TOTAL)

    # Uniform superposition
    circuit.h(range(N_QUBITS_TOTAL))

    # Cost Hamiltonian evolution
    for indices, coefficient in ising_terms.items():

        coefficient = float(np.real(coefficient))

        if not np.isfinite(coefficient):
            raise ValueError(
                f"Non-finite coefficient for term {indices}"
            )

        if len(indices) == 0:

            # Identity term contributes only a global phase.
            continue

        elif len(indices) == 1:

            circuit.rz(
                2.0 * gamma * coefficient,
                indices[0]
            )

        elif len(indices) == 2:

            circuit.rzz(
                2.0 * gamma * coefficient,
                indices[0],
                indices[1]
            )

        else:

            raise ValueError(
                f"Unsupported Ising term: {indices}"
            )

    # Mixer Hamiltonian evolution
    circuit.rx(
        2.0 * beta,
        range(N_QUBITS_TOTAL)
    )

    return circuit


print("QAOA function rebuilt successfully.")

# ---------------------------------------------------------------------------
# 5. BUILD TEST CIRCUIT
# ---------------------------------------------------------------------------

print("\n5. BUILDING TEST CIRCUIT")

test_circuit = build_qaoa_state(
    gamma_value,
    beta_value
)

print("Circuit qubits:", test_circuit.num_qubits)
print("Circuit classical bits:", test_circuit.num_clbits)
print("Circuit depth:", test_circuit.depth())
print("Circuit operations:", test_circuit.count_ops())

# ---------------------------------------------------------------------------
# 6. DIMENSION AND GATE VALIDATION
# ---------------------------------------------------------------------------

print("\n6. CIRCUIT VALIDATION")

gate_counts = test_circuit.count_ops()

expected_cost_gates = single_count + two_count

actual_cost_gates = (
    gate_counts.get("rz", 0)
    + gate_counts.get("rzz", 0)
)

checks = {

    "Hamiltonian has 20 qubits":
        HAMILTONIAN_QUBITS == 20,

    "Circuit has 20 qubits":
        test_circuit.num_qubits == 20,

    "No classical bits":
        test_circuit.num_clbits == 0,

    "Exactly 20 Hadamard gates":
        gate_counts.get("h", 0) == 20,

    "Exactly 20 mixer gates":
        gate_counts.get("rx", 0) == 20,

    "Cost gates match non-identity terms":
        actual_cost_gates == expected_cost_gates,

    "No unsupported higher-order terms":
        higher_count == 0,

    "All Ising indices valid":
        min(all_indices) >= 0
        and max(all_indices) < 20,

    "Optimization parameters finite":
        np.isfinite(gamma_value)
        and np.isfinite(beta_value)
}

for name, passed in checks.items():

    print(
        f"{name}: {'PASS' if passed else 'FAIL'}"
    )

print("\nGate count details:")
print("Hadamard gates:", gate_counts.get("h", 0))
print("RZ gates:", gate_counts.get("rz", 0))
print("RZZ gates:", gate_counts.get("rzz", 0))
print("Mixer RX gates:", gate_counts.get("rx", 0))
print("Expected cost gates:", expected_cost_gates)
print("Actual cost gates:", actual_cost_gates)

# ---------------------------------------------------------------------------
# 7. PARAMETER PRESERVATION
# ---------------------------------------------------------------------------

print("\n7. OPTIMIZATION PARAMETER PRESERVATION")

print("Gamma:", optimized_gamma)
print("Beta:", optimized_beta)

if energy_value is not None:
    print("Energy:", energy_value)

parameters_preserved = (
    np.isclose(gamma_value, float(optimized_gamma))
    and np.isclose(beta_value, float(optimized_beta))
)

print(
    "Parameters preserved:",
    "PASS" if parameters_preserved else "FAIL"
)

checks["Optimization parameters preserved"] = (
    parameters_preserved
)

# ---------------------------------------------------------------------------
# 8. FINAL AUDIT
# ---------------------------------------------------------------------------

print("\n8. FINAL AUDIT")

overall_pass = all(checks.values())

for name, passed in checks.items():

    print(
        f"{'PASS' if passed else 'FAIL'} - {name}"
    )

print("\nOVERALL:")
print(
    "PASS" if overall_pass
    else "INVESTIGATION REQUIRED"
)

# ---------------------------------------------------------------------------
# 9. EXPORT AUDIT REPORT
# ---------------------------------------------------------------------------

import os
import json

OUTPUT_DIR = "/content/ATMIQ_FINAL_RESULTS"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

audit_report = {

    "cell": 152,

    "hamiltonian_qubits": int(HAMILTONIAN_QUBITS),

    "circuit_qubits": int(test_circuit.num_qubits),

    "customer_variables": int(N),

    "ising_dictionary_terms": int(len(ising_terms)),

    "identity_terms": int(identity_count),

    "single_qubit_terms": int(single_count),

    "two_qubit_terms": int(two_count),

    "higher_order_terms": int(higher_count),

    "expected_cost_gates": int(expected_cost_gates),

    "actual_cost_gates": int(actual_cost_gates),

    "hadamard_gates": int(gate_counts.get("h", 0)),

    "mixer_gates": int(gate_counts.get("rx", 0)),

    "gamma": gamma_value,

    "beta": beta_value,

    "energy": energy_value,

    "all_checks_passed": bool(overall_pass),

    "checks": {
        key: bool(value)
        for key, value in checks.items()
    }
}

report_path = os.path.join(
    OUTPUT_DIR,
    "cell_152_qaoa_circuit_dimension_audit.json"
)

with open(report_path, "w") as f:

    json.dump(
        audit_report,
        f,
        indent=4
    )

print("\n9. AUDIT REPORT SAVED")
print(report_path)

print("\nCELL 152 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 152A - IDENTITY TERM HANDLING AND QAOA CIRCUIT REPAIR
# ============================================================================

from qiskit import QuantumCircuit
import numpy as np

print("=" * 75)
print("CELL 152A - IDENTITY TERM HANDLING")
print("=" * 75)

# 1. IDENTIFY ISING TERMS
identity_coefficient = float(ising_terms.get((), 0.0))

single_terms = {
    k: v for k, v in ising_terms.items()
    if len(k) == 1
}

two_terms = {
    k: v for k, v in ising_terms.items()
    if len(k) == 2
}

print("\n1. ISING TERM CLASSIFICATION")
print("Identity coefficient:", identity_coefficient)
print("Identity terms:", int(() in ising_terms))
print("Single-qubit terms:", len(single_terms))
print("Two-qubit terms:", len(two_terms))
print("Total terms:", len(ising_terms))

assert len(ising_terms) == 186
assert len(single_terms) == 20
assert len(two_terms) == 165

# 2. REBUILD QAOA FUNCTION
print("\n2. REBUILDING QAOA FUNCTION")

def build_qaoa_state(gamma, beta):

    n_qubits = N_QUBITS_TOTAL
    circuit = QuantumCircuit(n_qubits)

    # Uniform superposition
    circuit.h(range(n_qubits))

    # Cost Hamiltonian evolution
    for indices, coefficient in ising_terms.items():

        coefficient = float(coefficient)

        # Identity term: global phase only.
        # No gate is needed for measurement probabilities.
        if len(indices) == 0:
            continue

        elif len(indices) == 1:
            circuit.rz(
                2 * gamma * coefficient,
                indices[0]
            )

        elif len(indices) == 2:
            circuit.rzz(
                2 * gamma * coefficient,
                indices[0],
                indices[1]
            )

        else:
            raise ValueError(
                f"Unsupported Ising term: {indices}"
            )

    # Mixer Hamiltonian evolution
    for q in range(n_qubits):
        circuit.rx(2 * beta, q)

    return circuit


# 3. BUILD TEST CIRCUIT
print("\n3. BUILDING TEST CIRCUIT")

test_circuit = build_qaoa_state(
    optimized_gamma,
    optimized_beta
)

print("Circuit qubits:", test_circuit.num_qubits)
print("Circuit depth:", test_circuit.depth())
print("Circuit operations:", test_circuit.count_ops())

# 4. VERIFY GATE COUNTS
print("\n4. GATE COUNT VALIDATION")

gate_counts = test_circuit.count_ops()

print("Hadamard gates:", gate_counts.get("h", 0))
print("RZ gates:", gate_counts.get("rz", 0))
print("RZZ gates:", gate_counts.get("rzz", 0))
print("RX gates:", gate_counts.get("rx", 0))

assert test_circuit.num_qubits == 20
assert gate_counts.get("h", 0) == 20
assert gate_counts.get("rz", 0) == 20
assert gate_counts.get("rzz", 0) == 165
assert gate_counts.get("rx", 0) == 20

# 5. IDENTITY TERM VALIDATION
print("\n5. IDENTITY TERM VALIDATION")

print("Identity coefficient preserved:", identity_coefficient)
print("Identity term omitted from physical gates: YES")
print("Reason: Global phase only")

# 6. PARAMETER PRESERVATION
print("\n6. OPTIMIZATION PARAMETERS")

print("Gamma:", optimized_gamma)
print("Beta:", optimized_beta)
print("Energy:", optimized_energy)

# 7. FINAL AUDIT
print("\n7. FINAL AUDIT")

audit = {
    "20_qubit_circuit": test_circuit.num_qubits == 20,
    "identity_term_preserved": () in ising_terms,
    "20_single_qubit_terms": len(single_terms) == 20,
    "165_two_qubit_terms": len(two_terms) == 165,
    "20_hadamard_gates": gate_counts.get("h", 0) == 20,
    "20_rz_gates": gate_counts.get("rz", 0) == 20,
    "165_rzz_gates": gate_counts.get("rzz", 0) == 165,
    "20_rx_gates": gate_counts.get("rx", 0) == 20
}

for name, passed in audit.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(audit.values())

print("\nCELL 152A COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 155 - QAOA OBJECTIVE FUNCTION RECONSTRUCTION
# ============================================================================

import numpy as np
from qiskit.quantum_info import Statevector

print("=" * 75)
print("CELL 155 - QAOA OBJECTIVE FUNCTION RECONSTRUCTION")
print("=" * 75)

# 1. DEFINE EXACT EXPECTATION VALUE
print("\n1. DEFINING QAOA OBJECTIVE")

def qaoa_objective_20(parameters):

    gamma = float(parameters[0])
    beta = float(parameters[1])

    circuit = build_qaoa_state(gamma, beta)

    state = Statevector.from_instruction(circuit)

    energy = state.expectation_value(
        ising_hamiltonian
    )

    return float(np.real(energy))


print("Objective function created.")
print("Hamiltonian qubits:", ising_hamiltonian.num_qubits)

# 2. VERIFY ORIGINAL PARAMETERS
print("\n2. ORIGINAL PARAMETER VERIFICATION")

original_parameters = np.array([
    optimized_gamma,
    optimized_beta
], dtype=float)

reconstructed_energy = qaoa_objective_20(
    original_parameters
)

print("Gamma:", original_parameters[0])
print("Beta:", original_parameters[1])
print("Reconstructed energy:", reconstructed_energy)
print("Previously verified energy:", corrected_qaoa_energy)

energy_error = abs(
    reconstructed_energy - corrected_qaoa_energy
)

print("Absolute error:", energy_error)

assert energy_error < 1e-6

# 3. LOCAL PARAMETER SENSITIVITY
print("\n3. LOCAL PARAMETER SENSITIVITY")

gamma_test = optimized_gamma
beta_test = optimized_beta

test_points = [
    ("Original", gamma_test, beta_test),
    ("Gamma lower", gamma_test * 0.9, beta_test),
    ("Gamma higher", gamma_test * 1.1, beta_test),
    ("Beta lower", gamma_test, beta_test - 0.01),
    ("Beta higher", gamma_test, beta_test + 0.01)
]

sensitivity_results = []

for label, gamma, beta in test_points:

    energy = qaoa_objective_20(
        [gamma, beta]
    )

    sensitivity_results.append(
        (label, gamma, beta, energy)
    )

for label, gamma, beta, energy in sensitivity_results:

    print(
        f"{label:15s} | "
        f"gamma={gamma:.10f} | "
        f"beta={beta:.10f} | "
        f"energy={energy:.8f}"
    )

# 4. SAVE VERIFIED OBJECTIVE
print("\n4. SAVING OBJECTIVE")

qaoa_objective_verified = qaoa_objective_20

print("qaoa_objective_verified: AVAILABLE")

# 5. FINAL AUDIT
print("\n5. FINAL AUDIT")

audit = {
    "objective_function_created": callable(
        qaoa_objective_verified
    ),
    "20_qubit_hamiltonian": (
        ising_hamiltonian.num_qubits == 20
    ),
    "original_energy_reproduced": (
        energy_error < 1e-6
    ),
    "finite_energy": np.isfinite(
        reconstructed_energy
    ),
    "five_sensitivity_evaluations": (
        len(sensitivity_results) == 5
    )
}

for name, passed in audit.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(audit.values())

print("\nCELL 155 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 156 - FRESH QAOA NUMERICAL OPTIMIZATION
# ============================================================================

import numpy as np
from scipy.optimize import minimize

print("=" * 75)
print("CELL 156 - FRESH QAOA NUMERICAL OPTIMIZATION")
print("=" * 75)

# 1. INITIALIZE OPTIMIZATION
print("\n1. OPTIMIZATION INITIALIZATION")

initial_gamma = float(optimized_gamma)
initial_beta = float(optimized_beta)

# Scale gamma for better numerical optimization.
initial_parameters = np.array([
    initial_gamma * 1000.0,
    initial_beta
])

print("Initial gamma:", initial_gamma)
print("Initial beta:", initial_beta)
print("Initial energy:", corrected_qaoa_energy)

# 2. DEFINE SCALED OBJECTIVE
print("\n2. DEFINING SCALED OBJECTIVE")

def scaled_qaoa_objective(parameters):

    gamma = float(parameters[0]) / 1000.0
    beta = float(parameters[1])

    return qaoa_objective_verified([gamma, beta])


# 3. RUN FRESH OPTIMIZATION
print("\n3. RUNNING POWELL OPTIMIZATION")
print("This may take some time for a 20-qubit statevector.")

qaoa_optimization_156 = minimize(
    scaled_qaoa_objective,
    initial_parameters,
    method="Powell",
    bounds=[
        (0.0, 1.0),       # gamma: 0 to 0.001
        (0.0, np.pi)      # beta: 0 to pi
    ],
    options={
        "maxiter": 30,
        "maxfev": 150,
        "xtol": 1e-4,
        "ftol": 1e-8,
        "disp": True
    }
)

# 4. EXTRACT OPTIMIZED PARAMETERS
print("\n4. NEW OPTIMIZATION RESULTS")

new_gamma = float(
    qaoa_optimization_156.x[0] / 1000.0
)

new_beta = float(
    qaoa_optimization_156.x[1]
)

new_energy = float(
    qaoa_optimization_156.fun
)

print("Optimization success:",
      qaoa_optimization_156.success)

print("Termination message:",
      qaoa_optimization_156.message)

print("Iterations:",
      qaoa_optimization_156.nit)

print("Function evaluations:",
      qaoa_optimization_156.nfev)

print("New gamma:", new_gamma)
print("New beta:", new_beta)
print("New minimum energy:", new_energy)

# 5. COMPARE WITH PREVIOUS RESULT
print("\n5. OLD VS NEW COMPARISON")

energy_improvement = (
    corrected_qaoa_energy - new_energy
)

print("Previous energy:", corrected_qaoa_energy)
print("New energy:", new_energy)
print("Energy improvement:", energy_improvement)

if energy_improvement > 1e-8:
    print("Result: ENERGY IMPROVED")
elif abs(energy_improvement) <= 1e-8:
    print("Result: NO SIGNIFICANT CHANGE")
else:
    print("Result: NEW RESULT IS WORSE")

# 6. INDEPENDENTLY VERIFY NEW PARAMETERS
print("\n6. INDEPENDENT ENERGY VERIFICATION")

new_test_circuit = build_qaoa_state(
    new_gamma,
    new_beta
)

new_state = Statevector.from_instruction(
    new_test_circuit
)

verified_new_energy = float(
    np.real(
        new_state.expectation_value(
            ising_hamiltonian
        )
    )
)

verification_error = abs(
    verified_new_energy - new_energy
)

print("Optimizer energy:", new_energy)
print("Independent energy:", verified_new_energy)
print("Verification error:", verification_error)

assert verification_error < 1e-6

# 7. SAVE NEW RESULTS
optimized_gamma_156 = new_gamma
optimized_beta_156 = new_beta
optimized_energy_156 = verified_new_energy

# 8. FINAL AUDIT
print("\n7. FINAL AUDIT")

audit = {
    "optimization_result_available": (
        np.isfinite(new_energy)
    ),
    "gamma_within_bounds": (
        0 <= new_gamma <= 0.001
    ),
    "beta_within_bounds": (
        0 <= new_beta <= np.pi
    ),
    "independent_energy_verified": (
        verification_error < 1e-6
    ),
    "20_qubit_circuit": (
        new_test_circuit.num_qubits == 20
    )
}

for name, passed in audit.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(audit.values())

print("\nCELL 156 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 166 - EXACT QAOA STATEVECTOR AND MEASUREMENT VALIDATION
# ============================================================================

import numpy as np
import pandas as pd
from qiskit.quantum_info import Statevector

print("=" * 75)
print("CELL 166 - EXACT QAOA STATEVECTOR AND MEASUREMENT VALIDATION")
print("=" * 75)

# 1. LOAD BEST PARAMETERS
gamma_final = float(optimized_gamma)
beta_final = float(optimized_beta)
energy_reference = float(optimized_energy)

print("\n1. FINAL QAOA PARAMETERS")
print("Gamma:", gamma_final)
print("Beta:", beta_final)
print("Reference energy:", energy_reference)
print("Total qubits:", N_QUBITS_TOTAL)

# 2. RECONSTRUCT OPTIMIZED CIRCUIT
final_qaoa_circuit = build_qaoa_state(
    gamma_final,
    beta_final
)

print("\n2. CIRCUIT VALIDATION")
print("Circuit qubits:", final_qaoa_circuit.num_qubits)
print("Circuit depth:", final_qaoa_circuit.depth())
print("Circuit operations:", final_qaoa_circuit.count_ops())

assert final_qaoa_circuit.num_qubits == N_QUBITS_TOTAL

# 3. EXACT STATEVECTOR
final_statevector = Statevector.from_instruction(
    final_qaoa_circuit
)

probabilities = np.asarray(
    final_statevector.probabilities(),
    dtype=float
)

print("\n3. STATEVECTOR VALIDATION")
print("Statevector dimension:", len(final_statevector.data))
print("Probability array size:", len(probabilities))
print("Probability sum:", probabilities.sum())
print("Minimum probability:", probabilities.min())
print("Maximum probability:", probabilities.max())

# 4. EXACT EXPECTATION VALUE
exact_energy = float(np.real(
    final_statevector.expectation_value(
        ising_hamiltonian
    )
))

energy_error = abs(exact_energy - energy_reference)

print("\n4. EXACT ENERGY VERIFICATION")
print("Reference energy:", energy_reference)
print("Statevector energy:", exact_energy)
print("Absolute energy error:", energy_error)

# 5. PROBABILITY DISTRIBUTION ANALYSIS
nonzero_probabilities = probabilities[
    probabilities > 0
]

shannon_entropy = -np.sum(
    nonzero_probabilities *
    np.log2(nonzero_probabilities)
)

effective_states = 2 ** shannon_entropy

print("\n5. PROBABILITY DISTRIBUTION")
print("Nonzero states:", len(nonzero_probabilities))
print("Shannon entropy:", shannon_entropy)
print("Effective states:", effective_states)
print("Maximum possible entropy:", N_QUBITS_TOTAL)

# 6. TOP 10 MOST PROBABLE STATES
top_indices = np.argsort(probabilities)[-10:][::-1]

top_states_166 = []

for index in top_indices:

    bitstring = format(
        int(index),
        f"0{N_QUBITS_TOTAL}b"
    )

    top_states_166.append({
        "bitstring": bitstring,
        "probability": float(probabilities[index]),
        "index": int(index)
    })

top_states_166_df = pd.DataFrame(top_states_166)

print("\n6. TOP 10 MOST PROBABLE STATES")
print(top_states_166_df.to_string(index=False))

# 7. GENERATE 4096 SIMULATED SHOTS
NUM_SHOTS_166 = 4096

rng_166 = np.random.default_rng(42)

sampled_indices = rng_166.choice(
    len(probabilities),
    size=NUM_SHOTS_166,
    p=probabilities
)

sampled_bitstrings = [
    format(int(index), f"0{N_QUBITS_TOTAL}b")
    for index in sampled_indices
]

unique_strings, frequencies = np.unique(
    sampled_bitstrings,
    return_counts=True
)

qaoa_final_counts_166 = {
    str(bitstring): int(frequency)
    for bitstring, frequency in zip(
        unique_strings,
        frequencies
    )
}

print("\n7. MEASUREMENT SUMMARY")
print("Total shots:", NUM_SHOTS_166)
print("Unique measured states:", len(qaoa_final_counts_166))
print("Most frequent state:", max(
    qaoa_final_counts_166,
    key=qaoa_final_counts_166.get
))
print("Maximum frequency:", max(
    qaoa_final_counts_166.values()
))

# 8. FINAL AUDIT
checks_166 = {
    "correct_qubit_count":
        final_qaoa_circuit.num_qubits == N_QUBITS_TOTAL,

    "correct_statevector_dimension":
        len(probabilities) == 2 ** N_QUBITS_TOTAL,

    "normalized_probabilities":
        np.isclose(probabilities.sum(), 1.0, atol=1e-10),

    "finite_probabilities":
        np.isfinite(probabilities).all(),

    "exact_energy_verified":
        energy_error < 1e-7,

    "measurement_shots_correct":
        sum(qaoa_final_counts_166.values()) == NUM_SHOTS_166,

    "valid_probability_distribution":
        np.all(probabilities >= 0)
}

print("\n8. FINAL AUDIT")

for name, passed in checks_166.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(checks_166.values()), "CELL 166 AUDIT FAILED"

print("\nFINAL QAOA PARAMETERS")
print("Gamma:", gamma_final)
print("Beta:", beta_final)
print("Energy:", exact_energy)

print("\nCELL 166 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 167 - QUANTUM MEASUREMENT DECODING AND FEASIBILITY ANALYSIS
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 167 - QUANTUM MEASUREMENT DECODING AND FEASIBILITY ANALYSIS")
print("=" * 75)

# 1. MODEL CONFIGURATION
N_CUSTOMERS = 10
K_SELECTED = 6
BUDGET = 25
CAPACITY_LIMIT = 22

customer_ids = quantum_candidates["customer_id"].tolist()

cost_array = np.asarray(quantum_costs).flatten()
capacity_array = np.asarray(quantum_capacities).flatten()
value_array = np.asarray(quantum_values).flatten()

assert len(customer_ids) == N_CUSTOMERS
assert len(cost_array) == N_CUSTOMERS
assert len(capacity_array) == N_CUSTOMERS
assert len(value_array) == N_CUSTOMERS

print("\n1. MODEL CONFIGURATION")
print("Customers:", N_CUSTOMERS)
print("Required selections:", K_SELECTED)
print("Budget:", BUDGET)
print("Capacity limit:", CAPACITY_LIMIT)
print("Customer IDs:", customer_ids)

# 2. VALIDATE MEASUREMENT DATA
assert len(sampled_indices) == NUM_SHOTS_166
assert len(qaoa_final_counts_166) > 0

print("\n2. MEASUREMENT DATA")
print("Total shots:", sum(qaoa_final_counts_166.values()))
print("Unique measured states:", len(qaoa_final_counts_166))

# 3. DECODE EACH UNIQUE MEASURED STATE
decoded_results_167 = []

for bitstring, frequency in qaoa_final_counts_166.items():

    # Qiskit displayed bitstrings are reversed relative
    # to the qubit-index convention used by the model.
    bits = bitstring[::-1]

    # Customer selection: qubits 0-9
    customer_bits = np.array(
        [int(bits[q]) for q in range(N_CUSTOMERS)]
    )

    # Budget slack: qubits 10-14
    budget_slack_bits = np.array(
        [int(bits[q]) for q in range(10, 15)]
    )

    # Capacity slack: qubits 15-19
    capacity_slack_bits = np.array(
        [int(bits[q]) for q in range(15, 20)]
    )

    budget_weights = np.array([1, 2, 4, 8, 16])
    capacity_weights = np.array([1, 2, 4, 8, 16])

    budget_slack = int(
        np.dot(budget_slack_bits, budget_weights)
    )

    capacity_slack = int(
        np.dot(capacity_slack_bits, capacity_weights)
    )

    selected_indices = np.where(customer_bits == 1)[0]

    selected_count = len(selected_indices)

    selected_cost = int(
        np.sum(cost_array[selected_indices])
    )

    selected_capacity = int(
        np.sum(capacity_array[selected_indices])
    )

    selected_value = float(
        np.sum(value_array[selected_indices])
    )

    # Exact slack-encoded constraint validation
    count_valid = selected_count == K_SELECTED

    budget_valid = (
        selected_cost + budget_slack == BUDGET
    )

    capacity_valid = (
        selected_capacity + capacity_slack == CAPACITY_LIMIT
    )

    feasible = (
        count_valid
        and budget_valid
        and capacity_valid
    )

    selected_customers = [
        customer_ids[i] for i in selected_indices
    ]

    decoded_results_167.append({

        "bitstring": bitstring,

        "frequency": int(frequency),

        "selected_count": selected_count,

        "selected_customers": selected_customers,

        "contact_cost": selected_cost,

        "budget_slack": budget_slack,

        "total_budget_encoded":
            selected_cost + budget_slack,

        "capacity": selected_capacity,

        "capacity_slack": capacity_slack,

        "total_capacity_encoded":
            selected_capacity + capacity_slack,

        "expected_net_value": selected_value,

        "count_valid": count_valid,

        "budget_valid": budget_valid,

        "capacity_valid": capacity_valid,

        "feasible": feasible
    })

# 4. CREATE DECODING DATAFRAME
quantum_decoded_167 = pd.DataFrame(
    decoded_results_167
)

print("\n3. DECODING SUMMARY")
print("Decoded unique states:", len(quantum_decoded_167))
print("Total decoded shots:",
      quantum_decoded_167["frequency"].sum())

# 5. FEASIBILITY ANALYSIS
feasible_results_167 = quantum_decoded_167[
    quantum_decoded_167["feasible"]
].copy()

infeasible_results_167 = quantum_decoded_167[
    ~quantum_decoded_167["feasible"]
].copy()

feasible_shots_167 = int(
    feasible_results_167["frequency"].sum()
)

infeasible_shots_167 = int(
    infeasible_results_167["frequency"].sum()
)

print("\n4. FEASIBILITY SUMMARY")
print("Feasible unique states:", len(feasible_results_167))
print("Infeasible unique states:", len(infeasible_results_167))
print("Feasible shots:", feasible_shots_167)
print("Infeasible shots:", infeasible_shots_167)

print(
    "Feasible shot percentage:",
    round(
        100 * feasible_shots_167 / NUM_SHOTS_166,
        4
    )
)

# 6. TOP FEASIBLE SOLUTIONS
if len(feasible_results_167) > 0:

    feasible_results_167 = feasible_results_167.sort_values(
        by=["expected_net_value", "frequency"],
        ascending=[False, False]
    ).reset_index(drop=True)

    print("\n5. TOP 10 FEASIBLE QUANTUM SOLUTIONS")

    columns_to_display = [
        "frequency",
        "selected_count",
        "selected_customers",
        "contact_cost",
        "capacity",
        "expected_net_value"
    ]

    print(
        feasible_results_167[
            columns_to_display
        ].head(10).to_string(index=False)
    )

    best_quantum_solution_167 = (
        feasible_results_167.iloc[0].to_dict()
    )

    print("\n6. BEST FEASIBLE QUANTUM SOLUTION")
    print("Customers:",
          best_quantum_solution_167["selected_customers"])
    print("Contact cost:",
          best_quantum_solution_167["contact_cost"])
    print("Capacity:",
          best_quantum_solution_167["capacity"])
    print("Expected net value:",
          best_quantum_solution_167["expected_net_value"])
    print("Measurement frequency:",
          best_quantum_solution_167["frequency"])

else:

    best_quantum_solution_167 = None

    print("\n5. NO FEASIBLE SOLUTION FOUND IN THE SAMPLED SHOTS.")
    print("Do not interpret this as proof that no feasible")
    print("solution exists in the complete state space.")

# 7. CLASSICAL BENCHMARK COMPARISON
classical_best_value_167 = 37.306891109

print("\n7. CLASSICAL BENCHMARK")

print("Classical expected net value:",
      classical_best_value_167)

if best_quantum_solution_167 is not None:

    quantum_best_value_167 = float(
        best_quantum_solution_167["expected_net_value"]
    )

    print("Quantum expected net value:",
          quantum_best_value_167)

    print("Difference:",
          quantum_best_value_167 - classical_best_value_167)

    print(
        "Quantum matches classical:",
        np.isclose(
            quantum_best_value_167,
            classical_best_value_167,
            atol=1e-8
        )
    )

# 8. FINAL AUDIT
checks_167 = {

    "all_unique_states_decoded":
        len(quantum_decoded_167) ==
        len(qaoa_final_counts_166),

    "all_measurement_shots_accounted":
        quantum_decoded_167["frequency"].sum() ==
        NUM_SHOTS_166,

    "no_missing_feasibility_labels":
        quantum_decoded_167["feasible"].notna().all(),

    "feasible_and_infeasible_partition_correct":
        len(feasible_results_167) +
        len(infeasible_results_167) ==
        len(quantum_decoded_167),

    "feasible_constraints_valid":
        (
            feasible_results_167["count_valid"].all()
            and feasible_results_167["budget_valid"].all()
            and feasible_results_167["capacity_valid"].all()
        ) if len(feasible_results_167) > 0 else True,

    "finite_objective_values":
        np.isfinite(
            quantum_decoded_167["expected_net_value"]
        ).all()
}

print("\n8. FINAL AUDIT")

for name, passed in checks_167.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(checks_167.values()), "CELL 167 AUDIT FAILED"

# 9. FINAL SUMMARY
print("\nFINAL QUANTUM DECODING SUMMARY")
print("Total shots:", NUM_SHOTS_166)
print("Unique states:", len(quantum_decoded_167))
print("Feasible unique states:", len(feasible_results_167))
print("Feasible shots:", feasible_shots_167)

print("\nCELL 167 COMPLETED")
print("=" * 75)
```

```python

# ============================================================================
# CELL 170 - QAOA FEASIBLE STATE-SPACE PROBABILITY AUDIT
# ============================================================================

import numpy as np
import pandas as pd
from itertools import combinations

print("=" * 75)
print("CELL 170 - QAOA FEASIBLE STATE-SPACE PROBABILITY AUDIT")
print("=" * 75)

# --------------------------------------------------------------------------
# 1. CONFIGURATION
# --------------------------------------------------------------------------

N_CUSTOMERS_170 = len(quantum_candidates)
N_SELECT_170 = 6
N_QUBITS_170 = 20
STATE_SPACE_170 = 2 ** N_QUBITS_170

BUDGET_170 = 25
CAPACITY_LIMIT_170 = 22

BUDGET_SLACK_WEIGHTS_170 = [1, 2, 4, 8, 16]
CAPACITY_SLACK_WEIGHTS_170 = [1, 2, 4, 8, 16]

df_170 = quantum_candidates.reset_index(drop=True).copy()

print("\n1. CONFIGURATION")
print("Customers:", N_CUSTOMERS_170)
print("Required selections:", N_SELECT_170)
print("Qubits:", N_QUBITS_170)
print("Total computational states:", STATE_SPACE_170)

# --------------------------------------------------------------------------
# 2. ENUMERATE ALL SIX-CUSTOMER SELECTIONS
# --------------------------------------------------------------------------

valid_selections_170 = []
valid_encoded_states_170 = []

for combo in combinations(range(N_CUSTOMERS_170), N_SELECT_170):

    selected = df_170.iloc[list(combo)]

    selected_cost = int(selected["contact_cost"].sum())
    selected_capacity = int(selected["capacity_units"].sum())
    selected_value = float(selected["expected_net_value"].sum())

    budget_slack = BUDGET_170 - selected_cost
    capacity_slack = CAPACITY_LIMIT_170 - selected_capacity

    feasible = (
        selected_cost <= BUDGET_170
        and selected_capacity <= CAPACITY_LIMIT_170
    )

    if not feasible:
        continue

    # Five-bit binary slack representation.
    if not (
        0 <= budget_slack <= 31
        and 0 <= capacity_slack <= 31
    ):
        raise ValueError("Slack value cannot be represented with five bits.")

    budget_slack_bits = [
        (budget_slack >> i) & 1 for i in range(5)
    ]

    capacity_slack_bits = [
        (capacity_slack >> i) & 1 for i in range(5)
    ]

    # Qiskit convention: qubit 0 is the least significant bit.
    state_index = 0

    for customer_index in combo:
        state_index |= (1 << customer_index)

    for i, bit in enumerate(budget_slack_bits):
        if bit:
            state_index |= (1 << (10 + i))

    for i, bit in enumerate(capacity_slack_bits):
        if bit:
            state_index |= (1 << (15 + i))

    valid_selections_170.append({
        "selected_customers": selected["customer_id"].tolist(),
        "selected_count": len(combo),
        "expected_net_value": selected_value,
        "contact_cost": selected_cost,
        "capacity": selected_capacity,
        "budget_slack": budget_slack,
        "capacity_slack": capacity_slack,
        "encoded_state_index": state_index
    })

    valid_encoded_states_170.append(state_index)

feasible_selection_count_170 = len(valid_selections_170)

valid_selection_df_170 = pd.DataFrame(valid_selections_170)

print("\n2. CLASSICAL FEASIBILITY ENUMERATION")
print("Total six-customer combinations:",
      len(list(combinations(range(N_CUSTOMERS_170), N_SELECT_170))))
print("Feasible selections:", feasible_selection_count_170)
print("Infeasible selections:",
      210 - feasible_selection_count_170)

# --------------------------------------------------------------------------
# 3. VALID QUANTUM STATE-SPACE SIZE
# --------------------------------------------------------------------------

unique_encoded_states_170 = len(set(valid_encoded_states_170))

valid_state_fraction_170 = (
    unique_encoded_states_170 / STATE_SPACE_170
)

valid_state_percentage_170 = valid_state_fraction_170 * 100

print("\n3. VALID QUANTUM STATE-SPACE")
print("Unique valid 20-qubit states:", unique_encoded_states_170)
print("Total 20-qubit states:", STATE_SPACE_170)
print("Valid state fraction:", valid_state_fraction_170)
print("Valid state percentage:", valid_state_percentage_170)

# --------------------------------------------------------------------------
# 4. UNIFORM SAMPLING BASELINE
# --------------------------------------------------------------------------

NUM_SHOTS_170 = 4096

expected_uniform_feasible_shots_170 = (
    NUM_SHOTS_170 * valid_state_fraction_170
)

observed_feasible_shots_170 = int(feasible_shots_167)

observed_feasible_rate_170 = (
    observed_feasible_shots_170 / NUM_SHOTS_170
)

print("\n4. SAMPLING BASELINE")
print("Shots:", NUM_SHOTS_170)
print("Expected feasible shots under uniform sampling:",
      expected_uniform_feasible_shots_170)
print("Observed feasible shots:", observed_feasible_shots_170)
print("Observed feasible rate (%):",
      observed_feasible_rate_170 * 100)

# --------------------------------------------------------------------------
# 5. EXACT STATEVECTOR PROBABILITY MASS ON VALID STATES
# --------------------------------------------------------------------------

exact_feasible_probability_170 = float(
    np.sum(probabilities[np.array(valid_encoded_states_170, dtype=int)])
)

print("\n5. EXACT STATEVECTOR FEASIBLE PROBABILITY")
print("Total probability:", float(np.sum(probabilities)))
print("Probability mass on valid states:",
      exact_feasible_probability_170)
print("Probability mass on invalid states:",
      1.0 - exact_feasible_probability_170)
print("Expected feasible shots from this statevector:",
      exact_feasible_probability_170 * NUM_SHOTS_170)

# --------------------------------------------------------------------------
# 6. BEST FEASIBLE QUANTUM AND CLASSICAL SOLUTIONS
# --------------------------------------------------------------------------

classical_best_170 = valid_selection_df_170.sort_values(
    "expected_net_value",
    ascending=False
).iloc[0]

quantum_best_170 = best_quantum_solution_167

classical_value_170 = float(
    classical_best_170["expected_net_value"]
)

quantum_value_170 = float(
    quantum_best_170["expected_net_value"]
)

print("\n6. SOLUTION COMPARISON")
print("Classical optimum:", classical_value_170)
print("Best sampled quantum value:", quantum_value_170)
print("Objective difference:",
      classical_value_170 - quantum_value_170)

print("\nClassical optimal customers:")
print(classical_best_170["selected_customers"])

print("\nBest sampled quantum customers:")
print(quantum_best_170["selected_customers"])

# --------------------------------------------------------------------------
# 7. AUDIT
# --------------------------------------------------------------------------

checks_170 = {
    "exactly_210_selections_enumerated":
        len(list(combinations(range(N_CUSTOMERS_170), N_SELECT_170))) == 210,

    "all_210_selections_are_feasible":
        feasible_selection_count_170 == 210,

    "all_encoded_states_are_unique":
        unique_encoded_states_170 == feasible_selection_count_170,

    "all_encoded_states_within_20_qubits":
        all(0 <= x < STATE_SPACE_170 for x in valid_encoded_states_170),

    "statevector_probability_normalized":
        np.isclose(np.sum(probabilities), 1.0, atol=1e-8),

    "valid_probability_mass_in_range":
        -1e-10 <= exact_feasible_probability_170 <= 1.0 + 1e-10,

    "classical_optimum_matches_cell_169":
        np.isclose(
            classical_value_170,
            classical_optimum_169,
            atol=1e-8
        )
}

print("\n7. FINAL AUDIT")

for name, passed in checks_170.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(checks_170.values()), "Cell 170 audit failed."

print("\n" + "=" * 75)
print("CELL 170 COMPLETED SUCCESSFULLY")
print("=" * 75)
```

```python

# ============================================================================
# CELL 171 - DETAILED QAOA CONSTRAINT PROBABILITY DIAGNOSIS
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 75)
print("CELL 171 - DETAILED QAOA CONSTRAINT PROBABILITY DIAGNOSIS")
print("=" * 75)

# --------------------------------------------------------------------------
# 1. CONFIGURATION
# --------------------------------------------------------------------------

N_QUBITS_171 = 20
N_STATES_171 = 2 ** N_QUBITS_171

N_CUSTOMERS_171 = 10
N_SELECT_171 = 6

BUDGET_171 = 25
CAPACITY_LIMIT_171 = 22

df_171 = quantum_candidates.reset_index(drop=True).copy()

indices_171 = np.arange(N_STATES_171, dtype=np.uint32)

probabilities_171 = np.asarray(probabilities, dtype=np.float64)

assert len(probabilities_171) == N_STATES_171

print("\n1. CONFIGURATION")
print("Qubits:", N_QUBITS_171)
print("States:", N_STATES_171)
print("Required selections:", N_SELECT_171)
print("Budget:", BUDGET_171)
print("Capacity limit:", CAPACITY_LIMIT_171)

# --------------------------------------------------------------------------
# 2. DECODE CUSTOMER SELECTION BITS
# --------------------------------------------------------------------------

customer_bits_171 = np.column_stack([
    ((indices_171 >> i) & 1)
    for i in range(N_CUSTOMERS_171)
]).astype(np.int8)

selected_count_171 = customer_bits_171.sum(axis=1)

selected_cost_171 = (
    customer_bits_171 @ quantum_costs.astype(np.int64)
)

selected_capacity_171 = (
    customer_bits_171 @ quantum_capacities.astype(np.int64)
)

selected_value_171 = (
    customer_bits_171 @ quantum_values.astype(np.float64)
)

# --------------------------------------------------------------------------
# 3. DECODE SLACK BITS
# --------------------------------------------------------------------------

budget_slack_171 = np.zeros(N_STATES_171, dtype=np.int64)
capacity_slack_171 = np.zeros(N_STATES_171, dtype=np.int64)

for i, weight in enumerate([1, 2, 4, 8, 16]):
    budget_slack_171 += (
        ((indices_171 >> (10 + i)) & 1) * weight
    )

    capacity_slack_171 += (
        ((indices_171 >> (15 + i)) & 1) * weight
    )

# --------------------------------------------------------------------------
# 4. INDIVIDUAL CONSTRAINT CHECKS
# --------------------------------------------------------------------------

correct_selection_count_171 = (
    selected_count_171 == N_SELECT_171
)

correct_budget_slack_171 = (
    selected_cost_171 + budget_slack_171 == BUDGET_171
)

correct_capacity_slack_171 = (
    selected_capacity_171 + capacity_slack_171
    == CAPACITY_LIMIT_171
)

budget_within_limit_171 = selected_cost_171 <= BUDGET_171

capacity_within_limit_171 = (
    selected_capacity_171 <= CAPACITY_LIMIT_171
)

fully_feasible_171 = (
    correct_selection_count_171
    & correct_budget_slack_171
    & correct_capacity_slack_171
)

# --------------------------------------------------------------------------
# 5. PROBABILITY MASS BY SELECTION COUNT
# --------------------------------------------------------------------------

selection_count_rows_171 = []

for count in range(N_CUSTOMERS_171 + 1):

    mask = selected_count_171 == count

    mass = float(probabilities_171[mask].sum())

    selection_count_rows_171.append({
        "selected_count": count,
        "number_of_states": int(mask.sum()),
        "probability_mass": mass,
        "probability_percent": mass * 100
    })

selection_count_df_171 = pd.DataFrame(
    selection_count_rows_171
)

print("\n2. PROBABILITY BY NUMBER OF SELECTED CUSTOMERS")
print(selection_count_df_171.to_string(index=False))

# --------------------------------------------------------------------------
# 6. INDIVIDUAL CONSTRAINT PROBABILITIES
# --------------------------------------------------------------------------

mass_exact_selection_171 = float(
    probabilities_171[correct_selection_count_171].sum()
)

mass_correct_budget_171 = float(
    probabilities_171[correct_budget_slack_171].sum()
)

mass_correct_capacity_171 = float(
    probabilities_171[correct_capacity_slack_171].sum()
)

mass_budget_limit_171 = float(
    probabilities_171[budget_within_limit_171].sum()
)

mass_capacity_limit_171 = float(
    probabilities_171[capacity_within_limit_171].sum()
)

mass_all_constraints_171 = float(
    probabilities_171[fully_feasible_171].sum()
)

print("\n3. INDIVIDUAL CONSTRAINT PROBABILITY MASS")

print("Exactly six selected:",
      mass_exact_selection_171)

print("Correct budget slack:",
      mass_correct_budget_171)

print("Correct capacity slack:",
      mass_correct_capacity_171)

print("Budget inequality satisfied:",
      mass_budget_limit_171)

print("Capacity inequality satisfied:",
      mass_capacity_limit_171)

print("All three constraints satisfied:",
      mass_all_constraints_171)

# --------------------------------------------------------------------------
# 7. CONDITIONAL CONSTRAINT DIAGNOSIS
# --------------------------------------------------------------------------

selection_and_budget_171 = (
    correct_selection_count_171
    & correct_budget_slack_171
)

selection_and_capacity_171 = (
    correct_selection_count_171
    & correct_capacity_slack_171
)

selection_budget_capacity_171 = (
    correct_selection_count_171
    & correct_budget_slack_171
    & correct_capacity_slack_171
)

mass_selection_budget_171 = float(
    probabilities_171[selection_and_budget_171].sum()
)

mass_selection_capacity_171 = float(
    probabilities_171[selection_and_capacity_171].sum()
)

print("\n4. COMBINED CONSTRAINT PROBABILITIES")

print("Selection + budget:",
      mass_selection_budget_171)

print("Selection + capacity:",
      mass_selection_capacity_171)

print("Selection + budget + capacity:",
      mass_all_constraints_171)

# --------------------------------------------------------------------------
# 8. INVALID STATE CLASSIFICATION
# --------------------------------------------------------------------------

invalid_selection_171 = ~correct_selection_count_171

invalid_budget_171 = (
    correct_selection_count_171
    & ~correct_budget_slack_171
)

invalid_capacity_171 = (
    correct_selection_count_171
    & correct_budget_slack_171
    & ~correct_capacity_slack_171
)

invalid_categories_171 = {
    "Wrong selection count": invalid_selection_171,
    "Wrong budget slack after correct selection": invalid_budget_171,
    "Wrong capacity slack after correct selection and budget": invalid_capacity_171
}

print("\n5. DISJOINT INVALID-STATE PROBABILITY BREAKDOWN")

for name, mask in invalid_categories_171.items():

    mass = float(probabilities_171[mask].sum())

    print(name + ":")
    print("  Number of states:", int(mask.sum()))
    print("  Probability mass:", mass)
    print("  Percentage:", mass * 100)

# --------------------------------------------------------------------------
# 9. OBJECTIVE QUALITY AMONG EXACTLY-SIX STATES
# --------------------------------------------------------------------------

if mass_exact_selection_171 > 0:

    mean_value_exact_six_171 = float(
        np.sum(
            probabilities_171[correct_selection_count_171]
            * selected_value_171[correct_selection_count_171]
        ) / mass_exact_selection_171
    )

else:
    mean_value_exact_six_171 = np.nan

if mass_all_constraints_171 > 0:

    mean_value_feasible_171 = float(
        np.sum(
            probabilities_171[fully_feasible_171]
            * selected_value_171[fully_feasible_171]
        ) / mass_all_constraints_171
    )

else:
    mean_value_feasible_171 = np.nan

print("\n6. PROBABILITY-WEIGHTED OBJECTIVE QUALITY")

print("Mean objective among exactly-six states:",
      mean_value_exact_six_171)

print("Mean objective among fully feasible states:",
      mean_value_feasible_171)

# --------------------------------------------------------------------------
# 10. TOP PROBABILITY FEASIBLE STATES
# --------------------------------------------------------------------------

feasible_indices_171 = np.flatnonzero(fully_feasible_171)

top_feasible_rows_171 = []

if len(feasible_indices_171) > 0:

    ordered_171 = feasible_indices_171[
        np.argsort(
            probabilities_171[feasible_indices_171]
        )[::-1]
    ][:10]

    for rank, state_index in enumerate(ordered_171, start=1):

        selected_positions = np.flatnonzero(
            customer_bits_171[state_index]
        )

        selected_customers = (
            df_171.iloc[selected_positions]["customer_id"].tolist()
        )

        top_feasible_rows_171.append({
            "rank": rank,
            "state_index": int(state_index),
            "probability": float(
                probabilities_171[state_index]
            ),
            "expected_net_value": float(
                selected_value_171[state_index]
            ),
            "contact_cost": int(
                selected_cost_171[state_index]
            ),
            "capacity": int(
                selected_capacity_171[state_index]
            ),
            "selected_customers": selected_customers
        })

top_feasible_probability_df_171 = pd.DataFrame(
    top_feasible_rows_171
)

print("\n7. TOP 10 FEASIBLE STATES BY QUANTUM PROBABILITY")

if len(top_feasible_probability_df_171) > 0:
    print(top_feasible_probability_df_171.to_string(index=False))
else:
    print("No feasible states found.")

# --------------------------------------------------------------------------
# 11. FINAL AUDIT
# --------------------------------------------------------------------------

checks_171 = {
    "probability_vector_has_correct_size":
        len(probabilities_171) == N_STATES_171,

    "probability_vector_is_normalized":
        np.isclose(probabilities_171.sum(), 1.0, atol=1e-8),

    "selection_count_probability_is_valid":
        0 <= mass_exact_selection_171 <= 1,

    "budget_probability_is_valid":
        0 <= mass_correct_budget_171 <= 1,

    "capacity_probability_is_valid":
        0 <= mass_correct_capacity_171 <= 1,

    "all_feasible_probability_is_valid":
        0 <= mass_all_constraints_171 <= 1,

    "invalid_categories_partition_probability":
        np.isclose(
            sum(
                probabilities_171[mask].sum()
                for mask in invalid_categories_171.values()
            ) + mass_all_constraints_171,
            1.0,
            atol=1e-8
        ),

    "feasible_states_respect_all_constraints":
        np.all(selected_count_171[fully_feasible_171] == 6)
        and np.all(
            selected_cost_171[fully_feasible_171]
            + budget_slack_171[fully_feasible_171] == 25
        )
        and np.all(
            selected_capacity_171[fully_feasible_171]
            + capacity_slack_171[fully_feasible_171] == 22
        )
}

print("\n8. FINAL AUDIT")

for name, passed in checks_171.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

assert all(checks_171.values()), "Cell 171 audit failed."

print("\n" + "=" * 75)
print("CELL 171 COMPLETED SUCCESSFULLY")
print("=" * 75)
```

```python

# ============================================================================
# CELL 173 - QAOA FEASIBLE-STATE PROBABILITY AND CONSTRAINT VALIDATION
# ============================================================================

import numpy as np
import pandas as pd
from qiskit.quantum_info import Statevector

print("=" * 79)
print("CELL 173 - QAOA FEASIBLE-STATE PROBABILITY AND CONSTRAINT VALIDATION")
print("=" * 79)

# 1. VERIFY REQUIRED VARIABLES

required = [
    "build_qaoa_state",
    "optimized_gamma",
    "optimized_beta",
    "costs",
    "capacities",
    "BUDGET",
    "CAPACITY_LIMIT",
    "K_SELECTED"
]

missing = [v for v in required if v not in globals()]

if missing:
    raise RuntimeError(f"Missing required variables: {missing}")

cost_array = np.asarray(costs, dtype=float)
capacity_array = np.asarray(capacities, dtype=float)

assert len(cost_array) == 10, "Expected 10 customer cost values"
assert len(capacity_array) == 10, "Expected 10 customer capacity values"

print("\n1. CONFIGURATION")
print("Number of customer variables:", len(cost_array))
print("Number of total qubits:", N_QUBITS_TOTAL)
print("Budget:", BUDGET)
print("Capacity limit:", CAPACITY_LIMIT)
print("Required selections:", K_SELECTED)

# 2. RECONSTRUCT OPTIMIZED QAOA STATE

print("\n2. RECONSTRUCTING OPTIMIZED QAOA STATE")

circuit_173 = build_qaoa_state(
    float(optimized_gamma),
    float(optimized_beta)
)

state_173 = Statevector.from_instruction(circuit_173)

probabilities_173 = np.asarray(
    state_173.probabilities(),
    dtype=float
)

print("Circuit depth:", circuit_173.depth())
print("Statevector dimension:", len(state_173.data))
print("Probability vector length:", len(probabilities_173))
print("Total probability:", probabilities_173.sum())

assert len(probabilities_173) == 2 ** N_QUBITS_TOTAL
assert np.isclose(probabilities_173.sum(), 1.0, atol=1e-8)

# 3. DECODE CUSTOMER DECISION VARIABLES
# Qiskit uses little-endian computational basis indexing.
# Customer qubits are 0 through 9.

print("\n3. DECODING CUSTOMER DECISION VARIABLES")

state_indices = np.arange(len(probabilities_173), dtype=np.uint64)

decision_matrix_173 = np.column_stack([
    ((state_indices >> q) & 1)
    for q in range(10)
]).astype(np.int8)

selected_count_173 = decision_matrix_173.sum(axis=1)

# 4. CALCULATE CONSTRAINT VALUES

total_cost_173 = decision_matrix_173 @ cost_array
total_capacity_173 = decision_matrix_173 @ capacity_array

count_valid_173 = selected_count_173 == K_SELECTED

budget_valid_173 = total_cost_173 <= BUDGET

capacity_valid_173 = total_capacity_173 <= CAPACITY_LIMIT

feasible_mask_173 = (
    count_valid_173
    & budget_valid_173
    & capacity_valid_173
)

# 5. PROBABILITY DISTRIBUTION

print("\n4. FEASIBILITY DIAGNOSIS")

total_states_173 = len(probabilities_173)

feasible_states_173 = np.count_nonzero(feasible_mask_173)

infeasible_states_173 = total_states_173 - feasible_states_173

feasible_probability_173 = probabilities_173[feasible_mask_173].sum()

infeasible_probability_173 = probabilities_173[~feasible_mask_173].sum()

print("Total quantum states:", total_states_173)
print("Feasible computational states:", feasible_states_173)
print("Infeasible computational states:", infeasible_states_173)

print(
    "Feasible probability:",
    f"{feasible_probability_173:.10f}"
)

print(
    "Feasible probability (%):",
    f"{100 * feasible_probability_173:.6f}%"
)

print(
    "Infeasible probability (%):",
    f"{100 * infeasible_probability_173:.6f}%"
)

# 6. INDIVIDUAL CONSTRAINT PROBABILITIES

constraint_probability_173 = {
    "Exactly K customers": probabilities_173[count_valid_173].sum(),
    "Budget satisfied": probabilities_173[budget_valid_173].sum(),
    "Capacity satisfied": probabilities_173[capacity_valid_173].sum(),
    "All constraints satisfied": feasible_probability_173
}

constraint_df_173 = pd.DataFrame([
    {
        "constraint": name,
        "probability": value,
        "percentage": value * 100
    }
    for name, value in constraint_probability_173.items()
])

print("\n5. INDIVIDUAL CONSTRAINT PROBABILITIES")
print(constraint_df_173.to_string(index=False))

# 7. MOST PROBABLE FEASIBLE STATES

print("\n6. TOP 10 FEASIBLE STATES")

feasible_indices_173 = np.flatnonzero(feasible_mask_173)

if len(feasible_indices_173) > 0:

    sorted_feasible_173 = feasible_indices_173[
        np.argsort(
            probabilities_173[feasible_indices_173]
        )[::-1]
    ]

    top_rows_173 = []

    for rank, idx in enumerate(sorted_feasible_173[:10], start=1):

        selected = np.flatnonzero(
            decision_matrix_173[idx]
        ).tolist()

        top_rows_173.append({
            "rank": rank,
            "state_index": int(idx),
            "selected_customers": selected,
            "selected_count": int(selected_count_173[idx]),
            "total_cost": float(total_cost_173[idx]),
            "total_capacity": float(total_capacity_173[idx]),
            "probability": float(probabilities_173[idx])
        })

    top_feasible_qaoa_173 = pd.DataFrame(top_rows_173)

    print(top_feasible_qaoa_173.to_string(index=False))

else:

    top_feasible_qaoa_173 = pd.DataFrame()

    print("WARNING: No feasible computational states found.")

# 8. PROBABILITY BY NUMBER OF SELECTED CUSTOMERS

print("\n7. PROBABILITY BY CUSTOMER SELECTION COUNT")

selection_distribution_173 = []

for k in range(11):

    mask = selected_count_173 == k

    selection_distribution_173.append({
        "selected_count": k,
        "state_count": int(mask.sum()),
        "probability_mass": float(probabilities_173[mask].sum()),
        "probability_percent": float(
            probabilities_173[mask].sum() * 100
        )
    })

selection_distribution_df_173 = pd.DataFrame(
    selection_distribution_173
)

print(selection_distribution_df_173.to_string(index=False))

# 9. FINAL VALIDATION

print("\n8. FINAL VALIDATION")

print(
    "Statevector normalization:",
    "PASS" if np.isclose(probabilities_173.sum(), 1.0) else "FAIL"
)

print(
    "Decision-variable decoding:",
    "PASS" if decision_matrix_173.shape[1] == 10 else "FAIL"
)

print(
    "Constraint calculations:",
    "PASS" if len(feasible_mask_173) == total_states_173 else "FAIL"
)

print(
    "Feasible states identified:",
    "PASS" if feasible_states_173 > 0 else "WARNING"
)

print("\n" + "=" * 79)
print("CELL 173 DIAGNOSTIC COMPLETED")
print("=" * 79)
```

```python

# ============================================================================
# CELL 174 - CLASSICAL OPTIMUM VS QAOA FEASIBLE SOLUTION COMPARISON
# ============================================================================

import numpy as np
import pandas as pd

print("=" * 79)
print("CELL 174 - CLASSICAL OPTIMUM VS QAOA FEASIBLE SOLUTION COMPARISON")
print("=" * 79)

# 1. VERIFY ORIGINAL OBJECTIVE COEFFICIENTS

required_174 = [
    "quantum_values",
    "quantum_costs",
    "costs",
    "capacities",
    "probabilities_173",
    "BUDGET",
    "CAPACITY_LIMIT",
    "K_SELECTED"
]

missing_174 = [x for x in required_174 if x not in globals()]

if missing_174:
    raise RuntimeError(f"Missing variables: {missing_174}")

qvalues_174 = np.asarray(quantum_values, dtype=float)
qcosts_174 = np.asarray(quantum_costs, dtype=float)

costs_174 = np.asarray(costs, dtype=float)
capacities_174 = np.asarray(capacities, dtype=float)

assert len(qvalues_174) == 10
assert len(qcosts_174) == 10

assert np.all(np.isfinite(qvalues_174))
assert np.all(np.isfinite(qcosts_174))

# Original objective:
# Expected net value = expected value - contact cost

net_coefficients_174 = qvalues_174 - qcosts_174

print("\n1. ORIGINAL OBJECTIVE AUDIT")
print("Customer variables:", len(net_coefficients_174))
print("Objective coefficient range:",
      net_coefficients_174.min(),
      "to",
      net_coefficients_174.max())

# 2. EXHAUSTIVE CLASSICAL FEASIBLE SOLUTIONS

print("\n2. ENUMERATING CLASSICAL CUSTOMER COMBINATIONS")

combinations_174 = np.arange(2**10, dtype=np.uint16)

decision_174 = np.column_stack([
    ((combinations_174 >> q) & 1)
    for q in range(10)
]).astype(np.int8)

selected_174 = decision_174.sum(axis=1)

total_cost_174 = decision_174 @ costs_174
total_capacity_174 = decision_174 @ capacities_174

net_value_174 = decision_174 @ net_coefficients_174

feasible_174 = (
    (selected_174 == K_SELECTED)
    & (total_cost_174 <= BUDGET)
    & (total_capacity_174 <= CAPACITY_LIMIT)
)

classical_df_174 = pd.DataFrame({
    "state_index": combinations_174,
    "selected_count": selected_174,
    "total_cost": total_cost_174,
    "total_capacity": total_capacity_174,
    "expected_net_value": net_value_174,
    "feasible": feasible_174
})

feasible_classical_174 = (
    classical_df_174[classical_df_174["feasible"]]
    .sort_values("expected_net_value", ascending=False)
    .reset_index(drop=True)
)

print("Total customer combinations:", len(classical_df_174))
print("Feasible combinations:", len(feasible_classical_174))

if feasible_classical_174.empty:
    raise RuntimeError("No classical feasible solutions found.")

# 3. CLASSICAL OPTIMAL SOLUTION

print("\n3. CLASSICAL OPTIMAL SOLUTION")

classical_best_174 = feasible_classical_174.iloc[0]

classical_best_mask_174 = int(classical_best_174["state_index"])

classical_best_customers_174 = np.flatnonzero(
    decision_174[classical_best_mask_174]
).tolist()

print("Selected customers:", classical_best_customers_174)
print("Expected net value:",
      classical_best_174["expected_net_value"])
print("Total contact cost:", classical_best_174["total_cost"])
print("Total capacity:", classical_best_174["total_capacity"])

# 4. AGGREGATE QUANTUM PROBABILITY BY CUSTOMER COMBINATION

print("\n4. AGGREGATING QAOA PROBABILITY")

probabilities_array_174 = np.asarray(
    probabilities_173,
    dtype=float
)

customer_masks_174 = (
    np.arange(len(probabilities_array_174), dtype=np.uint64)
    & 1023
).astype(np.int64)

customer_probability_174 = np.bincount(
    customer_masks_174,
    weights=probabilities_array_174,
    minlength=1024
)

assert np.isclose(customer_probability_174.sum(), 1.0, atol=1e-8)

# 5. RANK QAOA PROBABILITY AMONG FEASIBLE COMBINATIONS

feasible_classical_174["qaoa_probability"] = (
    customer_probability_174[
        feasible_classical_174["state_index"].to_numpy(dtype=int)
    ]
)

qaoa_ranked_174 = (
    feasible_classical_174
    .sort_values("qaoa_probability", ascending=False)
    .reset_index(drop=True)
)

qaoa_ranked_174["qaoa_probability_rank"] = (
    np.arange(1, len(qaoa_ranked_174) + 1)
)

qaoa_ranked_174["objective_rank"] = (
    qaoa_ranked_174["expected_net_value"]
    .rank(method="min", ascending=False)
    .astype(int)
)

print("\n5. TOP 10 FEASIBLE SOLUTIONS BY QAOA PROBABILITY")

print(
    qaoa_ranked_174[
        [
            "state_index",
            "expected_net_value",
            "total_cost",
            "total_capacity",
            "qaoa_probability",
            "objective_rank"
        ]
    ].head(10).to_string(index=False)
)

# 6. QUANTUM PROBABILITY OF CLASSICAL OPTIMUM

classical_optimum_probability_174 = (
    customer_probability_174[classical_best_mask_174]
)

classical_optimum_probability_rank_174 = int(
    qaoa_ranked_174.index[
        qaoa_ranked_174["state_index"] == classical_best_mask_174
    ][0] + 1
)

print("\n6. CLASSICAL OPTIMUM REPRESENTATION IN QAOA")

print("Classical optimal customer combination:",
      classical_best_customers_174)

print("Aggregated QAOA probability:",
      classical_optimum_probability_174)

print("QAOA probability (%):",
      classical_optimum_probability_174 * 100)

print("Rank among feasible combinations:",
      classical_optimum_probability_rank_174)

# 7. OBJECTIVE GAP ANALYSIS

qaoa_probability_best_174 = qaoa_ranked_174.iloc[0]

objective_gap_174 = (
    classical_best_174["expected_net_value"]
    - qaoa_probability_best_174["expected_net_value"]
)

relative_gap_174 = (
    objective_gap_174 /
    abs(classical_best_174["expected_net_value"]) * 100
    if classical_best_174["expected_net_value"] != 0
    else np.nan
)

print("\n7. OBJECTIVE COMPARISON")

print("Classical optimal objective:",
      classical_best_174["expected_net_value"])

print("Highest-probability feasible QAOA objective:",
      qaoa_probability_best_174["expected_net_value"])

print("Objective gap:", objective_gap_174)
print("Relative objective gap (%):", relative_gap_174)

# 8. SAVE RESULTS IN MEMORY

classical_optimum_174 = classical_best_174.copy()

qaoa_comparison_df_174 = qaoa_ranked_174.copy()

qaoa_objective_gap_174 = float(objective_gap_174)

qaoa_optimum_probability_174 = float(
    classical_optimum_probability_174
)

# 9. FINAL AUDIT

print("\n8. FINAL VALIDATION")

print("Objective coefficient audit: PASS")
print("Classical exhaustive enumeration: PASS")
print("Feasibility validation: PASS")
print("QAOA probability aggregation:",
      "PASS" if np.isclose(customer_probability_174.sum(), 1.0)
      else "FAIL")

print("Classical optimum identified: PASS")
print("QAOA comparison completed: PASS")

print("\n" + "=" * 79)
print("CELL 174 DIAGNOSTIC COMPLETED")
print("=" * 79)
```

```python

# =============================================================================
# CELL 179 - INDEPENDENT CLASSICAL OPTIMIZATION VERIFICATION
# =============================================================================

import numpy as np
import pandas as pd
import itertools
import os
import json
from datetime import datetime

print("=" * 85)
print("CELL 179 - INDEPENDENT CLASSICAL OPTIMIZATION VERIFICATION")
print("=" * 85)

# ---------------------------------------------------------------------------
# 1. LOAD AND VALIDATE INPUT ARRAYS
# ---------------------------------------------------------------------------

required_arrays = [
    "quantum_values",
    "quantum_costs",
    "quantum_capacities"
]

for name in required_arrays:
    if name not in globals():
        raise NameError(f"Required variable '{name}' is missing.")

values_179 = np.asarray(quantum_values, dtype=float)
costs_179 = np.asarray(quantum_costs, dtype=float)
capacities_179 = np.asarray(quantum_capacities, dtype=float)

n_179 = len(values_179)

if not (len(costs_179) == len(capacities_179) == n_179):
    raise ValueError("Input arrays have inconsistent dimensions.")

if not all(np.all(np.isfinite(x)) for x in [
    values_179, costs_179, capacities_179
]):
    raise ValueError("Non-finite input values detected.")

REQUIRED_SELECTIONS_179 = 6
BUDGET_LIMIT_179 = 25
CAPACITY_LIMIT_179 = 22

print("\n1. INPUT CONFIGURATION")
print("Number of customers:", n_179)
print("Required selections:", REQUIRED_SELECTIONS_179)
print("Budget:", BUDGET_LIMIT_179)
print("Capacity limit:", CAPACITY_LIMIT_179)

# ---------------------------------------------------------------------------
# 2. EXHAUSTIVE ENUMERATION
# ---------------------------------------------------------------------------

print("\n2. ENUMERATING ALL POSSIBLE COMBINATIONS")

classical_solutions_179 = []

for combination in itertools.combinations(range(n_179), REQUIRED_SELECTIONS_179):

    selected_values = values_179[list(combination)]
    selected_costs = costs_179[list(combination)]
    selected_capacities = capacities_179[list(combination)]

    total_value = float(np.sum(selected_values))
    total_cost = float(np.sum(selected_costs))
    total_capacity = float(np.sum(selected_capacities))

    feasible = (
        total_cost <= BUDGET_LIMIT_179
        and total_capacity <= CAPACITY_LIMIT_179
    )

    if feasible:
        objective = total_value - total_cost

        classical_solutions_179.append({
            "selected_indices": tuple(combination),
            "selected_customers": [i + 1 for i in combination],
            "objective": objective,
            "total_value": total_value,
            "total_cost": total_cost,
            "total_capacity": total_capacity
        })

print("Total combinations evaluated:",
      len(list(itertools.combinations(range(n_179), REQUIRED_SELECTIONS_179))))

print("Feasible combinations:",
      len(classical_solutions_179))

# ---------------------------------------------------------------------------
# 3. IDENTIFY OPTIMAL FEASIBLE SOLUTION
# ---------------------------------------------------------------------------

if len(classical_solutions_179) == 0:
    raise ValueError(
        "No feasible solutions found. Check the objective coefficients "
        "and constraints before continuing."
    )

classical_solutions_179 = sorted(
    classical_solutions_179,
    key=lambda x: x["objective"],
    reverse=True
)

best_solution_179 = classical_solutions_179[0]

classical_optimum_179 = best_solution_179["objective"]

print("\n3. INDEPENDENT CLASSICAL OPTIMUM")

print("Selected customers:",
      best_solution_179["selected_customers"])

print("Expected value:",
      round(best_solution_179["total_value"], 6))

print("Total cost:",
      round(best_solution_179["total_cost"], 6))

print("Total capacity:",
      round(best_solution_179["total_capacity"], 6))

print("Optimal objective:",
      round(classical_optimum_179, 6))

# ---------------------------------------------------------------------------
# 4. TOP 10 FEASIBLE SOLUTIONS
# ---------------------------------------------------------------------------

top10_solutions_179 = pd.DataFrame(
    classical_solutions_179[:10]
)

print("\n4. TOP 10 FEASIBLE SOLUTIONS")

display(top10_solutions_179)

# ---------------------------------------------------------------------------
# 5. COMPARE AGAINST PREVIOUS CLASSICAL BENCHMARK
# ---------------------------------------------------------------------------

previous_benchmark_179 = 25.306891

difference_179 = abs(
    classical_optimum_179 - previous_benchmark_179
)

print("\n5. BENCHMARK CONSISTENCY")

print("Previous classical benchmark:",
      previous_benchmark_179)

print("Independently verified optimum:",
      round(classical_optimum_179, 6))

print("Absolute difference:",
      round(difference_179, 8))

if np.isclose(
    classical_optimum_179,
    previous_benchmark_179,
    rtol=1e-5,
    atol=1e-5
):
    print("RESULT: CLASSICAL OPTIMUM VERIFIED")
else:
    print("WARNING: CLASSICAL OBJECTIVE DOES NOT MATCH PREVIOUS BENCHMARK")
    print("Check the original objective coefficients and constraints.")

# ---------------------------------------------------------------------------
# 6. EXPORT RESULTS
# ---------------------------------------------------------------------------

os.makedirs(RESULTS_DIR, exist_ok=True)

top10_path_179 = os.path.join(
    RESULTS_DIR,
    "cell_179_top10_classical_solutions.csv"
)

all_solutions_path_179 = os.path.join(
    RESULTS_DIR,
    "cell_179_all_feasible_classical_solutions.csv"
)

pd.DataFrame(classical_solutions_179).to_csv(
    all_solutions_path_179,
    index=False
)

top10_solutions_179.to_csv(
    top10_path_179,
    index=False
)

verification_report_179 = {
    "cell": 179,
    "timestamp": datetime.now().isoformat(),
    "total_customers": int(n_179),
    "total_combinations": int(
        len(list(itertools.combinations(range(n_179), REQUIRED_SELECTIONS_179)))
    ),
    "feasible_combinations": len(classical_solutions_179),
    "independent_classical_optimum": float(classical_optimum_179),
    "previous_classical_benchmark": previous_benchmark_179,
    "absolute_difference": float(difference_179),
    "selected_customers": best_solution_179["selected_customers"],
    "budget_used": best_solution_179["total_cost"],
    "capacity_used": best_solution_179["total_capacity"]
}

report_path_179 = os.path.join(
    RESULTS_DIR,
    "cell_179_classical_verification.json"
)

with open(report_path_179, "w") as f:
    json.dump(verification_report_179, f, indent=4)

print("\n6. EXPORTS COMPLETED")
print(top10_path_179)
print(all_solutions_path_179)
print(report_path_179)

print("\n" + "=" * 85)
print("CELL 179 EXECUTION COMPLETED")
print("=" * 85)
```

```python

# =============================================================================
# CELL 181 - QAOA FINAL STATEVECTOR AND PROBABILITY AUDIT
# =============================================================================

import numpy as np
import pandas as pd
import json
import os

print("=" * 85)
print("CELL 181 - QAOA FINAL STATEVECTOR AND PROBABILITY AUDIT")
print("=" * 85)

# ---------------------------------------------------------------------------
# 1. EXPECTED QUANTUM DIMENSIONS
# ---------------------------------------------------------------------------

EXPECTED_QUBITS_181 = 20
EXPECTED_STATES_181 = 2 ** EXPECTED_QUBITS_181

print("\n1. EXPECTED QUANTUM DIMENSIONS")
print("Qubits:", EXPECTED_QUBITS_181)
print("Statevector dimension:", EXPECTED_STATES_181)

# ---------------------------------------------------------------------------
# 2. DISCOVER NOTEBOOK ARRAYS
# ---------------------------------------------------------------------------

candidate_arrays_181 = []

for name, value in globals().copy().items():

    if isinstance(value, np.ndarray):

        if value.size in (
            EXPECTED_STATES_181,
            2 ** 10,
            2 ** 20
        ):

            candidate_arrays_181.append({
                "variable": name,
                "shape": str(value.shape),
                "dtype": str(value.dtype),
                "size": int(value.size),
                "is_complex": bool(np.iscomplexobj(value)),
                "finite_values": bool(np.all(np.isfinite(value)))
            })

print("\n2. DISCOVERED QUANTUM ARRAYS")

if candidate_arrays_181:

    quantum_array_inventory_181 = pd.DataFrame(
        candidate_arrays_181
    )

    display(quantum_array_inventory_181)

else:

    quantum_array_inventory_181 = pd.DataFrame()

    print("No matching NumPy arrays discovered.")

# ---------------------------------------------------------------------------
# 3. DISCOVER LIKELY QAOA VARIABLES
# ---------------------------------------------------------------------------

keywords_181 = [
    "statevector",
    "state_vector",
    "probability",
    "probabilities",
    "probs",
    "amplitude",
    "amplitudes",
    "qaoa",
    "optimized_state",
    "final_state"
]

print("\n3. LIKELY QAOA VARIABLES")

likely_variables_181 = []

for name, value in globals().copy().items():

    if any(keyword in name.lower() for keyword in keywords_181):

        if isinstance(value, np.ndarray):

            description = (
                f"NumPy array, shape={value.shape}, "
                f"dtype={value.dtype}"
            )

        else:

            description = type(value).__name__

        likely_variables_181.append({
            "variable": name,
            "description": description
        })

if likely_variables_181:

    likely_variables_table_181 = pd.DataFrame(
        likely_variables_181
    )

    display(likely_variables_table_181)

else:

    likely_variables_table_181 = pd.DataFrame()

    print("No likely QAOA variables discovered.")

# ---------------------------------------------------------------------------
# 4. IDENTIFY NORMALIZED PROBABILITY ARRAYS
# ---------------------------------------------------------------------------

print("\n4. PROBABILITY ARRAY CANDIDATES")

probability_candidates_181 = []

for name, value in globals().copy().items():

    if not isinstance(value, np.ndarray):
        continue

    if value.size not in (
        EXPECTED_STATES_181,
        2 ** 10
    ):
        continue

    if np.iscomplexobj(value):
        continue

    if not np.all(np.isfinite(value)):
        continue

    if np.any(value < -1e-12):
        continue

    total_probability = float(np.sum(value))

    if np.isclose(total_probability, 1.0, atol=1e-5):

        probability_candidates_181.append({
            "variable": name,
            "size": int(value.size),
            "probability_sum": total_probability,
            "minimum": float(np.min(value)),
            "maximum": float(np.max(value))
        })

if probability_candidates_181:

    probability_candidates_table_181 = pd.DataFrame(
        probability_candidates_181
    )

    display(probability_candidates_table_181)

else:

    probability_candidates_table_181 = pd.DataFrame()

    print("No normalized probability arrays identified.")

# ---------------------------------------------------------------------------
# 5. EXPORT AUDIT
# ---------------------------------------------------------------------------

results_dir_181 = "/content/ATMIQ_FINAL_RESULTS"
os.makedirs(results_dir_181, exist_ok=True)

audit_181 = {
    "cell": 181,
    "expected_qubits": EXPECTED_QUBITS_181,
    "expected_states": EXPECTED_STATES_181,
    "discovered_arrays": candidate_arrays_181,
    "likely_qaoa_variables": likely_variables_181,
    "normalized_probability_candidates": probability_candidates_181,
    "probability_array_count": len(probability_candidates_181),
    "purpose": (
        "Identify the final QAOA statevector and probability "
        "variables before subsequent quantum analysis."
    )
}

audit_path_181 = os.path.join(
    results_dir_181,
    "cell_181_qaoa_statevector_audit.json"
)

with open(audit_path_181, "w") as f:
    json.dump(audit_181, f, indent=4)

print("\n5. AUDIT EXPORT")
print(audit_path_181)

print("\n6. AUDIT SUMMARY")
print("Matching arrays:", len(candidate_arrays_181))
print("Likely QAOA variables:", len(likely_variables_181))
print("Normalized probability candidates:",
      len(probability_candidates_181))

print("\n" + "=" * 85)
print("CELL 181 EXECUTION COMPLETED")
print("=" * 85)
```

```python

# =============================================================================
# CELL 182 - FINAL QAOA STATEVECTOR CONSISTENCY VERIFICATION
# =============================================================================

import numpy as np
import pandas as pd
import json
import os

print("=" * 85)
print("CELL 182 - FINAL QAOA STATEVECTOR CONSISTENCY VERIFICATION")
print("=" * 85)

# ---------------------------------------------------------------------------
# 1. EXTRACT FINAL STATEVECTOR
# ---------------------------------------------------------------------------

if "final_statevector" not in globals():
    raise NameError("final_statevector is unavailable.")

if not hasattr(final_statevector, "data"):
    raise TypeError("final_statevector does not expose statevector data.")

final_amplitudes_182 = np.asarray(final_statevector.data)

if final_amplitudes_182.size != 1048576:
    raise ValueError(
        f"Unexpected statevector size: {final_amplitudes_182.size}"
    )

final_probabilities_182 = np.abs(final_amplitudes_182) ** 2

print("\n1. FINAL STATEVECTOR")
print("Amplitude count:", final_amplitudes_182.size)
print("Probability count:", final_probabilities_182.size)
print("Probability sum:", float(np.sum(final_probabilities_182)))
print("Finite amplitudes:",
      bool(np.all(np.isfinite(final_amplitudes_182))))

# ---------------------------------------------------------------------------
# 2. COMPARE AGAINST DISCOVERED PROBABILITY ARRAYS
# ---------------------------------------------------------------------------

candidate_names_182 = [
    "probabilities",
    "state_probabilities",
    "nonzero_probabilities",
    "probabilities_171",
    "probabilities_173",
    "probabilities_array_174"
]

comparison_rows_182 = []

for name in candidate_names_182:

    if name not in globals():
        continue

    candidate = np.asarray(globals()[name])

    if candidate.shape != final_probabilities_182.shape:
        continue

    if np.iscomplexobj(candidate):
        continue

    difference = np.abs(
        candidate.astype(float) - final_probabilities_182
    )

    comparison_rows_182.append({
        "variable": name,
        "probability_sum": float(np.sum(candidate)),
        "maximum_absolute_difference": float(np.max(difference)),
        "mean_absolute_difference": float(np.mean(difference)),
        "matches_final_statevector": bool(
            np.allclose(
                candidate,
                final_probabilities_182,
                atol=1e-12,
                rtol=1e-8
            )
        )
    })

comparison_df_182 = pd.DataFrame(comparison_rows_182)

print("\n2. STATEVECTOR PROBABILITY COMPARISON")

if not comparison_df_182.empty:
    display(comparison_df_182)
else:
    print("No compatible probability arrays found.")

# ---------------------------------------------------------------------------
# 3. IDENTIFY MATCHING ARRAYS
# ---------------------------------------------------------------------------

matching_arrays_182 = []

if not comparison_df_182.empty:

    matching_arrays_182 = comparison_df_182.loc[
        comparison_df_182["matches_final_statevector"],
        "variable"
    ].tolist()

print("\n3. MATCHING PROBABILITY ARRAYS")

if matching_arrays_182:
    for name in matching_arrays_182:
        print("Verified:", name)
else:
    print(
        "WARNING: No discovered probability array exactly matches "
        "the final statevector."
    )

# ---------------------------------------------------------------------------
# 4. VALIDATE PROBABILITY DISTRIBUTION
# ---------------------------------------------------------------------------

probability_sum_182 = float(np.sum(final_probabilities_182))
minimum_probability_182 = float(np.min(final_probabilities_182))
maximum_probability_182 = float(np.max(final_probabilities_182))

normalization_valid_182 = bool(
    np.isclose(probability_sum_182, 1.0, atol=1e-8)
)

nonnegative_valid_182 = bool(
    np.all(final_probabilities_182 >= 0)
)

finite_valid_182 = bool(
    np.all(np.isfinite(final_probabilities_182))
)

print("\n4. FINAL PROBABILITY VALIDATION")
print("Normalized:", normalization_valid_182)
print("Nonnegative:", nonnegative_valid_182)
print("Finite:", finite_valid_182)
print("Minimum probability:", minimum_probability_182)
print("Maximum probability:", maximum_probability_182)

# ---------------------------------------------------------------------------
# 5. CUSTOMER-LEVEL PROBABILITY VALIDATION
# ---------------------------------------------------------------------------

customer_probability_check_182 = None

if "customer_probability_174" in globals():

    customer_probs = np.asarray(customer_probability_174)

    customer_probability_check_182 = {
        "size": int(customer_probs.size),
        "sum": float(np.sum(customer_probs)),
        "normalized": bool(
            np.isclose(np.sum(customer_probs), 1.0, atol=1e-8)
        ),
        "finite": bool(np.all(np.isfinite(customer_probs)))
    }

    print("\n5. CUSTOMER-LEVEL PROBABILITY VALIDATION")
    print("Customer combinations:", customer_probs.size)
    print("Probability sum:", customer_probability_check_182["sum"])
    print("Normalized:",
          customer_probability_check_182["normalized"])
    print("Finite:", customer_probability_check_182["finite"])

# ---------------------------------------------------------------------------
# 6. EXPORT VERIFICATION REPORT
# ---------------------------------------------------------------------------

results_dir_182 = "/content/ATMIQ_FINAL_RESULTS"
os.makedirs(results_dir_182, exist_ok=True)

report_182 = {
    "cell": 182,
    "statevector_dimension": int(final_amplitudes_182.size),
    "probability_sum": probability_sum_182,
    "minimum_probability": minimum_probability_182,
    "maximum_probability": maximum_probability_182,
    "normalization_valid": normalization_valid_182,
    "nonnegative_valid": nonnegative_valid_182,
    "finite_valid": finite_valid_182,
    "matching_probability_arrays": matching_arrays_182,
    "probability_array_comparisons": comparison_rows_182,
    "customer_probability_validation": customer_probability_check_182
}

report_path_182 = os.path.join(
    results_dir_182,
    "cell_182_statevector_consistency_report.json"
)

with open(report_path_182, "w") as f:
    json.dump(report_182, f, indent=4)

print("\n6. EXPORT")
print(report_path_182)

# ---------------------------------------------------------------------------
# 7. FINAL STATUS
# ---------------------------------------------------------------------------

print("\n7. FINAL STATUS")

if normalization_valid_182 and nonnegative_valid_182 and finite_valid_182:
    print("FINAL STATEVECTOR PROBABILITY DISTRIBUTION: VALID")
else:
    print("FINAL STATEVECTOR PROBABILITY DISTRIBUTION: REQUIRES INVESTIGATION")

if matching_arrays_182:
    print("REFERENCE PROBABILITY ARRAYS: IDENTIFIED")
else:
    print("REFERENCE PROBABILITY ARRAYS: NOT IDENTIFIED")

print("\n" + "=" * 85)
print("CELL 182 EXECUTION COMPLETED")
print("=" * 85)
```

```python

# =============================================================================
# CELL 183 - QAOA PROBABILITY OF THE CLASSICAL OPTIMAL SOLUTION
# =============================================================================

import numpy as np
import pandas as pd
import json
import os

print("=" * 85)
print("CELL 183 - QAOA PROBABILITY OF THE CLASSICAL OPTIMAL SOLUTION")
print("=" * 85)

# ---------------------------------------------------------------------------
# 1. VALIDATE REQUIRED INPUTS
# ---------------------------------------------------------------------------

required_variables_183 = [
    "probabilities_array_174",
    "customer_masks_174",
    "customer_probability_174",
    "feasible_174",
    "net_value_174"
]

missing_183 = [
    name for name in required_variables_183
    if name not in globals()
]

if missing_183:
    raise NameError(f"Missing required variables: {missing_183}")

state_probabilities_183 = np.asarray(
    probabilities_array_174, dtype=float
)

customer_masks_183 = np.asarray(
    customer_masks_174, dtype=np.int64
)

customer_probabilities_183 = np.asarray(
    customer_probability_174, dtype=float
)

feasible_masks_183 = np.asarray(feasible_174, dtype=bool)

objective_by_mask_183 = np.asarray(
    net_value_174, dtype=float
)

if state_probabilities_183.size != 1048576:
    raise ValueError("Unexpected state probability dimension.")

if customer_masks_183.size != 1048576:
    raise ValueError("Unexpected customer mask dimension.")

if customer_probabilities_183.size != 1024:
    raise ValueError("Expected 1024 customer combinations.")

if feasible_masks_183.size != 1024:
    raise ValueError("Unexpected feasibility array dimension.")

if objective_by_mask_183.size != 1024:
    raise ValueError("Unexpected objective array dimension.")

if not np.all(np.isfinite(state_probabilities_183)):
    raise ValueError("State probabilities contain non-finite values.")

print("\n1. INPUT VALIDATION")
print("Quantum states:", state_probabilities_183.size)
print("Customer combinations:", customer_probabilities_183.size)
print("State probability sum:",
      float(np.sum(state_probabilities_183)))

# ---------------------------------------------------------------------------
# 2. RECONSTRUCT CUSTOMER-LEVEL PROBABILITIES
# ---------------------------------------------------------------------------

if np.any(customer_masks_183 < 0) or np.any(customer_masks_183 >= 1024):
    raise ValueError("Customer masks are outside the expected range.")

reconstructed_probabilities_183 = np.bincount(
    customer_masks_183,
    weights=state_probabilities_183,
    minlength=1024
)

aggregation_difference_183 = float(
    np.max(
        np.abs(
            reconstructed_probabilities_183 -
            customer_probabilities_183
        )
    )
)

aggregation_verified_183 = bool(
    np.allclose(
        reconstructed_probabilities_183,
        customer_probabilities_183,
        atol=1e-10,
        rtol=1e-8
    )
)

print("\n2. CUSTOMER PROBABILITY AGGREGATION")
print("Reconstructed probability sum:",
      float(np.sum(reconstructed_probabilities_183)))
print("Maximum aggregation difference:",
      aggregation_difference_183)
print("Aggregation verified:", aggregation_verified_183)

if not aggregation_verified_183:
    raise ValueError(
        "Customer probability aggregation does not match "
        "the final statevector distribution."
    )

# ---------------------------------------------------------------------------
# 3. IDENTIFY CLASSICAL OPTIMAL CUSTOMER SELECTION
# ---------------------------------------------------------------------------

# Classical optimum from Cell 179:
# Customer indices 0,1,2,3,4,5 correspond to customers 1 through 6.

optimal_indices_183 = (0, 1, 2, 3, 4, 5)

optimal_customer_mask_183 = sum(
    1 << index for index in optimal_indices_183
)

optimal_customer_ids_183 = [
    index + 1 for index in optimal_indices_183
]

print("\n3. CLASSICAL OPTIMAL CUSTOMER SELECTION")
print("Selected customers:", optimal_customer_ids_183)
print("Customer mask:", optimal_customer_mask_183)
print("Expected number selected:", len(optimal_indices_183))

# Verify the mask represents exactly six customers.
mask_selection_count_183 = int(
    bin(optimal_customer_mask_183).count("1")
)

if mask_selection_count_183 != 6:
    raise ValueError("Optimal customer mask is incorrectly constructed.")

# ---------------------------------------------------------------------------
# 4. CALCULATE OPTIMAL SOLUTION PROBABILITY
# ---------------------------------------------------------------------------

optimal_solution_probability_183 = float(
    customer_probabilities_183[optimal_customer_mask_183]
)

optimal_solution_probability_percent_183 = (
    optimal_solution_probability_183 * 100
)

all_probability_ranks_183 = np.argsort(
    -customer_probabilities_183,
    kind="stable"
)

optimal_probability_rank_183 = int(
    np.where(
        all_probability_ranks_183 == optimal_customer_mask_183
    )[0][0] + 1
)

print("\n4. OPTIMAL SOLUTION PROBABILITY")
print("Optimal solution probability:",
      optimal_solution_probability_183)
print("Optimal solution probability (%):",
      optimal_solution_probability_percent_183)
print("Probability rank among all combinations:",
      optimal_probability_rank_183)

# ---------------------------------------------------------------------------
# 5. FEASIBLE-SOLUTION PROBABILITY MASS
# ---------------------------------------------------------------------------

feasible_probability_mass_183 = float(
    np.sum(customer_probabilities_183[feasible_masks_183])
)

infeasible_probability_mass_183 = float(
    np.sum(customer_probabilities_183[~feasible_masks_183])
)

optimal_is_feasible_183 = bool(
    feasible_masks_183[optimal_customer_mask_183]
)

print("\n5. FEASIBLE PROBABILITY MASS")
print("Feasible probability:",
      feasible_probability_mass_183)
print("Feasible probability (%):",
      feasible_probability_mass_183 * 100)
print("Infeasible probability:",
      infeasible_probability_mass_183)
print("Classical optimum is feasible:", optimal_is_feasible_183)

if not optimal_is_feasible_183:
    raise ValueError(
        "The classical optimum is marked infeasible in the QAOA mask table."
    )

# ---------------------------------------------------------------------------
# 6. TOP 10 MOST PROBABLE FEASIBLE SOLUTIONS
# ---------------------------------------------------------------------------

feasible_mask_indices_183 = np.where(feasible_masks_183)[0]

feasible_ranking_183 = feasible_mask_indices_183[
    np.argsort(
        -customer_probabilities_183[feasible_mask_indices_183],
        kind="stable"
    )
]

top_rows_183 = []

for rank, mask in enumerate(feasible_ranking_183[:10], start=1):

    selected_customers = [
        customer_index + 1
        for customer_index in range(10)
        if int(mask) & (1 << customer_index)
    ]

    top_rows_183.append({
        "rank": rank,
        "customer_mask": int(mask),
        "selected_customers": selected_customers,
        "selection_count": len(selected_customers),
        "objective": float(objective_by_mask_183[mask]),
        "qaoa_probability": float(customer_probabilities_183[mask]),
        "qaoa_probability_percent": float(
            customer_probabilities_183[mask] * 100
        ),
        "is_classical_optimum": bool(
            mask == optimal_customer_mask_183
        )
    })

top_feasible_probability_183 = pd.DataFrame(top_rows_183)

print("\n6. TOP 10 MOST PROBABLE FEASIBLE SOLUTIONS")
display(top_feasible_probability_183)

# ---------------------------------------------------------------------------
# 7. CLASSICAL OPTIMUM PROBABILITY RANK AMONG FEASIBLE SOLUTIONS
# ---------------------------------------------------------------------------

optimal_feasible_rank_183 = int(
    np.where(
        feasible_ranking_183 == optimal_customer_mask_183
    )[0][0] + 1
)

print("\n7. CLASSICAL OPTIMUM QUANTUM RANKING")
print("Rank among feasible combinations:",
      optimal_feasible_rank_183)
print("Number of feasible combinations:",
      len(feasible_mask_indices_183))

# ---------------------------------------------------------------------------
# 8. EXPORT RESULTS
# ---------------------------------------------------------------------------

results_dir_183 = "/content/ATMIQ_FINAL_RESULTS"
os.makedirs(results_dir_183, exist_ok=True)

top_csv_path_183 = os.path.join(
    results_dir_183,
    "cell_183_top_qaoa_feasible_solutions.csv"
)

report_json_path_183 = os.path.join(
    results_dir_183,
    "cell_183_optimal_solution_probability_report.json"
)

top_feasible_probability_183.to_csv(
    top_csv_path_183,
    index=False
)

report_183 = {
    "cell": 183,
    "state_probability_sum": float(
        np.sum(state_probabilities_183)
    ),
    "customer_probability_sum": float(
        np.sum(customer_probabilities_183)
    ),
    "customer_probability_aggregation_verified": aggregation_verified_183,
    "maximum_aggregation_difference": aggregation_difference_183,
    "classical_optimal_customers": optimal_customer_ids_183,
    "classical_optimal_customer_mask": optimal_customer_mask_183,
    "optimal_solution_probability": optimal_solution_probability_183,
    "optimal_solution_probability_percent":
        optimal_solution_probability_percent_183,
    "optimal_probability_rank_all_combinations":
        optimal_probability_rank_183,
    "optimal_probability_rank_feasible_combinations":
        optimal_feasible_rank_183,
    "optimal_solution_is_feasible": optimal_is_feasible_183,
    "feasible_probability_mass": feasible_probability_mass_183,
    "infeasible_probability_mass": infeasible_probability_mass_183,
    "number_of_feasible_combinations":
        int(len(feasible_mask_indices_183))
}

with open(report_json_path_183, "w") as f:
    json.dump(report_183, f, indent=4)

print("\n8. EXPORTS")
print("Top feasible solutions:", top_csv_path_183)
print("Probability report:", report_json_path_183)

# ---------------------------------------------------------------------------
# 9. FINAL SUMMARY
# ---------------------------------------------------------------------------

print("\n9. FINAL SUMMARY")
print("Optimal customers:", optimal_customer_ids_183)
print("Optimal selection probability (%):",
      round(optimal_solution_probability_percent_183, 8))
print("Optimal rank among feasible combinations:",
      optimal_feasible_rank_183)
print("Feasible probability mass (%):",
      round(feasible_probability_mass_183 * 100, 6))
print("Aggregation verified:", aggregation_verified_183)

print("\n" + "=" * 85)
print("CELL 183 EXECUTION COMPLETED")
print("=" * 85)
```

```python

# =============================================================================
# CELL 184 - QAOA CONDITIONAL OBJECTIVE AND CLASSICAL OPTIMUM AUDIT
# =============================================================================

import numpy as np
import pandas as pd
import json
import os

print("=" * 85)
print("CELL 184 - QAOA CONDITIONAL OBJECTIVE AND CLASSICAL OPTIMUM AUDIT")
print("=" * 85)

# ---------------------------------------------------------------------------
# 1. LOAD VERIFIED CUSTOMER-LEVEL ARRAYS
# ---------------------------------------------------------------------------

probabilities_184 = np.asarray(
    customer_probability_174, dtype=float
)

feasible_184 = np.asarray(
    feasible_174, dtype=bool
)

objectives_184 = np.asarray(
    net_value_174, dtype=float
)

if not (
    probabilities_184.shape ==
    feasible_184.shape ==
    objectives_184.shape ==
    (1024,)
):
    raise ValueError("Customer-level arrays must each contain 1024 entries.")

if not np.all(np.isfinite(probabilities_184)):
    raise ValueError("Probabilities contain non-finite values.")

if not np.all(np.isfinite(objectives_184)):
    raise ValueError("Objective array contains non-finite values.")

if np.any(probabilities_184 < 0):
    raise ValueError("Negative probabilities detected.")

print("\n1. INPUT VALIDATION")
print("Customer combinations:", len(probabilities_184))
print("Probability sum:", float(np.sum(probabilities_184)))
print("Feasible combinations:", int(np.sum(feasible_184)))

# ---------------------------------------------------------------------------
# 2. RECOMPUTE CONDITIONAL EXPECTED OBJECTIVE
# ---------------------------------------------------------------------------

feasible_probability_184 = float(
    np.sum(probabilities_184[feasible_184])
)

if feasible_probability_184 <= 0:
    raise ValueError("No probability mass assigned to feasible solutions.")

conditional_probabilities_184 = (
    probabilities_184[feasible_184] /
    feasible_probability_184
)

feasible_objectives_184 = objectives_184[feasible_184]

conditional_expected_objective_184 = float(
    np.sum(
        conditional_probabilities_184 *
        feasible_objectives_184
    )
)

unconditional_feasible_objective_184 = float(
    np.sum(
        probabilities_184[feasible_184] *
        feasible_objectives_184
    )
)

print("\n2. QAOA CONDITIONAL EXPECTED OBJECTIVE")
print("Feasible probability mass:", feasible_probability_184)
print("Conditional probability sum:",
      float(np.sum(conditional_probabilities_184)))
print("Conditional expected objective:",
      conditional_expected_objective_184)
print("Unconditional feasible objective contribution:",
      unconditional_feasible_objective_184)

# ---------------------------------------------------------------------------
# 3. INDEPENDENTLY VERIFY CLASSICAL OPTIMUM
# ---------------------------------------------------------------------------

feasible_indices_184 = np.where(feasible_184)[0]

best_feasible_position_184 = int(
    np.argmax(objectives_184[feasible_184])
)

best_mask_184 = int(
    feasible_indices_184[best_feasible_position_184]
)

best_objective_184 = float(
    objectives_184[best_mask_184]
)

best_customers_184 = [
    index + 1
    for index in range(10)
    if best_mask_184 & (1 << index)
]

expected_classical_optimum_184 = 25.306891

classical_difference_184 = abs(
    best_objective_184 - expected_classical_optimum_184
)

classical_optimum_verified_184 = bool(
    np.isclose(
        best_objective_184,
        expected_classical_optimum_184,
        atol=1e-5
    )
)

print("\n3. CLASSICAL OPTIMUM VERIFICATION")
print("Best feasible mask:", best_mask_184)
print("Best feasible customers:", best_customers_184)
print("Best feasible objective:", best_objective_184)
print("Expected benchmark:", expected_classical_optimum_184)
print("Absolute difference:", classical_difference_184)
print("Classical optimum verified:", classical_optimum_verified_184)

# ---------------------------------------------------------------------------
# 4. VERIFY QAOA BENCHMARK CONSISTENCY
# ---------------------------------------------------------------------------

previous_qaoa_objective_184 = 11.406324

qaoa_difference_184 = abs(
    conditional_expected_objective_184 -
    previous_qaoa_objective_184
)

qaoa_benchmark_verified_184 = bool(
    np.isclose(
        conditional_expected_objective_184,
        previous_qaoa_objective_184,
        atol=1e-4
    )
)

print("\n4. QAOA BENCHMARK VERIFICATION")
print("Recomputed conditional objective:",
      conditional_expected_objective_184)
print("Previous benchmark:",
      previous_qaoa_objective_184)
print("Absolute difference:", qaoa_difference_184)
print("QAOA benchmark verified:", qaoa_benchmark_verified_184)

# ---------------------------------------------------------------------------
# 5. TOP 10 FEASIBLE SOLUTIONS BY OBJECTIVE
# ---------------------------------------------------------------------------

objective_order_184 = feasible_indices_184[
    np.argsort(
        -objectives_184[feasible_indices_184],
        kind="stable"
    )
]

objective_rows_184 = []

for rank, mask in enumerate(objective_order_184[:10], start=1):

    selected = [
        index + 1
        for index in range(10)
        if int(mask) & (1 << index)
    ]

    objective_rows_184.append({
        "objective_rank": rank,
        "customer_mask": int(mask),
        "selected_customers": selected,
        "objective": float(objectives_184[mask]),
        "qaoa_probability": float(probabilities_184[mask]),
        "conditional_probability": float(
            probabilities_184[mask] / feasible_probability_184
        )
    })

top_objectives_184 = pd.DataFrame(objective_rows_184)

print("\n5. TOP 10 FEASIBLE SOLUTIONS BY OBJECTIVE")
display(top_objectives_184)

# ---------------------------------------------------------------------------
# 6. PERFORMANCE METRICS
# ---------------------------------------------------------------------------

qaoa_quality_184 = (
    conditional_expected_objective_184 /
    best_objective_184
) * 100

relative_gap_184 = 100 - qaoa_quality_184

print("\n6. RECOMPUTED PERFORMANCE METRICS")
print("QAOA objective quality (%):", qaoa_quality_184)
print("Relative objective gap (%):", relative_gap_184)

# ---------------------------------------------------------------------------
# 7. EXPORT AUDIT
# ---------------------------------------------------------------------------

results_dir_184 = "/content/ATMIQ_FINAL_RESULTS"
os.makedirs(results_dir_184, exist_ok=True)

top_csv_184 = os.path.join(
    results_dir_184,
    "cell_184_top_feasible_by_objective.csv"
)

report_json_184 = os.path.join(
    results_dir_184,
    "cell_184_objective_consistency_report.json"
)

top_objectives_184.to_csv(top_csv_184, index=False)

report_184 = {
    "cell": 184,
    "feasible_probability_mass": feasible_probability_184,
    "conditional_expected_objective": conditional_expected_objective_184,
    "unconditional_feasible_objective_contribution":
        unconditional_feasible_objective_184,
    "classical_best_objective": best_objective_184,
    "classical_best_customers": best_customers_184,
    "classical_optimum_verified": classical_optimum_verified_184,
    "previous_qaoa_objective": previous_qaoa_objective_184,
    "qaoa_objective_difference": qaoa_difference_184,
    "qaoa_benchmark_verified": qaoa_benchmark_verified_184,
    "qaoa_objective_quality_percent": qaoa_quality_184,
    "relative_objective_gap_percent": relative_gap_184
}

with open(report_json_184, "w") as f:
    json.dump(report_184, f, indent=4)

print("\n7. EXPORTS")
print("Top objective solutions:", top_csv_184)
print("Consistency report:", report_json_184)

# ---------------------------------------------------------------------------
# 8. FINAL STATUS
# ---------------------------------------------------------------------------

print("\n8. FINAL STATUS")
print("Classical optimum verified:", classical_optimum_verified_184)
print("QAOA benchmark verified:", qaoa_benchmark_verified_184)

if classical_optimum_verified_184 and qaoa_benchmark_verified_184:
    print("RESULT: BOTH OBJECTIVE BENCHMARKS CONSISTENT")
else:
    print("RESULT: INVESTIGATION REQUIRED BEFORE FINAL REPORT")

print("\n" + "=" * 85)
print("CELL 184 EXECUTION COMPLETED")
print("=" * 85)
```

```python

# =============================================================================
# CELL 186 - BUDGET AND CAPACITY SENSITIVITY ANALYSIS
# =============================================================================

import numpy as np
import pandas as pd
import itertools
import json
import os
import matplotlib.pyplot as plt

print("=" * 85)
print("CELL 186 - BUDGET AND CAPACITY SENSITIVITY ANALYSIS")
print("=" * 85)

# ---------------------------------------------------------------------------
# 1. INPUT VALIDATION
# ---------------------------------------------------------------------------

costs_186 = np.asarray(costs, dtype=float)
capacities_186 = np.asarray(capacities, dtype=float)
objectives_186 = np.asarray(net_value_174, dtype=float)

N_CUSTOMERS_186 = 10
REQUIRED_SELECTIONS_186 = 6

BASELINE_BUDGET_186 = 25
BASELINE_CAPACITY_186 = 22
BASELINE_OBJECTIVE_186 = 25.30689110902086

if costs_186.shape != (N_CUSTOMERS_186,):
    raise ValueError("Unexpected costs array shape.")

if capacities_186.shape != (N_CUSTOMERS_186,):
    raise ValueError("Unexpected capacities array shape.")

if objectives_186.shape != (1024,):
    raise ValueError("Expected 1024 objective values.")

if not (
    np.all(np.isfinite(costs_186))
    and np.all(np.isfinite(capacities_186))
    and np.all(np.isfinite(objectives_186))
):
    raise ValueError("Non-finite input values detected.")

if np.any(costs_186 < 0) or np.any(capacities_186 < 0):
    raise ValueError("Costs and capacities must be nonnegative.")

print("\n1. INPUT CONFIGURATION")
print("Customers:", N_CUSTOMERS_186)
print("Required selections:", REQUIRED_SELECTIONS_186)
print("Budget baseline:", BASELINE_BUDGET_186)
print("Capacity baseline:", BASELINE_CAPACITY_186)
print("Objective array size:", objectives_186.size)

# ---------------------------------------------------------------------------
# 2. ENUMERATE ALL SIX-CUSTOMER COMBINATIONS
# ---------------------------------------------------------------------------

combinations_186 = list(
    itertools.combinations(range(N_CUSTOMERS_186), REQUIRED_SELECTIONS_186)
)

combination_records_186 = []

for selected in combinations_186:

    mask = sum(1 << index for index in selected)

    total_cost = float(np.sum(costs_186[list(selected)]))
    total_capacity = float(np.sum(capacities_186[list(selected)]))
    objective = float(objectives_186[mask])

    combination_records_186.append({
        "mask": int(mask),
        "selected_indices": selected,
        "selected_customers": [index + 1 for index in selected],
        "total_cost": total_cost,
        "total_capacity": total_capacity,
        "objective": objective
    })

combinations_df_186 = pd.DataFrame(combination_records_186)

print("\n2. COMBINATION ENUMERATION")
print("Total combinations:", len(combinations_df_186))
print("Expected combinations:", 210)

if len(combinations_df_186) != 210:
    raise ValueError("Combination enumeration is incomplete.")

# ---------------------------------------------------------------------------
# 3. DEFINE SENSITIVITY RANGES
# ---------------------------------------------------------------------------

budget_values_186 = list(range(15, 36, 2))
capacity_values_186 = list(range(12, 33, 2))

# Ensure the baseline constraints are included.
if BASELINE_BUDGET_186 not in budget_values_186:
    budget_values_186.append(BASELINE_BUDGET_186)

if BASELINE_CAPACITY_186 not in capacity_values_186:
    capacity_values_186.append(BASELINE_CAPACITY_186)

budget_values_186 = sorted(set(budget_values_186))
capacity_values_186 = sorted(set(capacity_values_186))

print("\n3. SENSITIVITY CONFIGURATION")
print("Budget scenarios:", budget_values_186)
print("Capacity scenarios:", capacity_values_186)
print(
    "Total constraint scenarios:",
    len(budget_values_186) * len(capacity_values_186)
)

# ---------------------------------------------------------------------------
# 4. EVALUATE EVERY BUDGET-CAPACITY SCENARIO
# ---------------------------------------------------------------------------

sensitivity_records_186 = []

for budget_limit in budget_values_186:

    for capacity_limit in capacity_values_186:

        feasible_solutions = combinations_df_186[
            (combinations_df_186["total_cost"] <= budget_limit)
            & (combinations_df_186["total_capacity"] <= capacity_limit)
        ]

        feasible_count = len(feasible_solutions)

        if feasible_count > 0:

            best_row = feasible_solutions.loc[
                feasible_solutions["objective"].idxmax()
            ]

            best_objective = float(best_row["objective"])
            best_customers = list(best_row["selected_customers"])
            best_mask = int(best_row["mask"])
            best_cost = float(best_row["total_cost"])
            best_capacity = float(best_row["total_capacity"])

        else:

            best_objective = None
            best_customers = None
            best_mask = None
            best_cost = None
            best_capacity = None

        sensitivity_records_186.append({
            "budget_limit": budget_limit,
            "capacity_limit": capacity_limit,
            "feasible_combinations": feasible_count,
            "best_objective": best_objective,
            "best_customer_mask": best_mask,
            "best_customers": best_customers,
            "best_total_cost": best_cost,
            "best_total_capacity": best_capacity
        })

sensitivity_df_186 = pd.DataFrame(sensitivity_records_186)

print("\n4. SENSITIVITY SCENARIOS COMPLETED")
print("Scenarios evaluated:", len(sensitivity_df_186))
print(
    "Scenarios with feasible solutions:",
    int((sensitivity_df_186["feasible_combinations"] > 0).sum())
)
print(
    "Scenarios without feasible solutions:",
    int((sensitivity_df_186["feasible_combinations"] == 0).sum())
)

# ---------------------------------------------------------------------------
# 5. VERIFY ORIGINAL BASELINE
# ---------------------------------------------------------------------------

baseline_rows_186 = sensitivity_df_186[
    (sensitivity_df_186["budget_limit"] == BASELINE_BUDGET_186)
    & (sensitivity_df_186["capacity_limit"] == BASELINE_CAPACITY_186)
]

if len(baseline_rows_186) != 1:
    raise ValueError("Baseline scenario was not uniquely identified.")

baseline_row_186 = baseline_rows_186.iloc[0]

baseline_difference_186 = abs(
    float(baseline_row_186["best_objective"])
    - BASELINE_OBJECTIVE_186
)

baseline_verified_186 = bool(
    np.isclose(
        baseline_row_186["best_objective"],
        BASELINE_OBJECTIVE_186,
        atol=1e-5
    )
)

print("\n5. BASELINE VERIFICATION")
print("Baseline budget:", BASELINE_BUDGET_186)
print("Baseline capacity:", BASELINE_CAPACITY_186)
print("Optimal customers:", baseline_row_186["best_customers"])
print("Optimal objective:", baseline_row_186["best_objective"])
print("Expected objective:", BASELINE_OBJECTIVE_186)
print("Absolute difference:", baseline_difference_186)
print("Baseline verified:", baseline_verified_186)

if not baseline_verified_186:
    raise ValueError(
        "Baseline objective does not match the verified classical optimum."
    )

# ---------------------------------------------------------------------------
# 6. DISPLAY SENSITIVITY RESULTS
# ---------------------------------------------------------------------------

print("\n6. SENSITIVITY RESULTS")
display(sensitivity_df_186)

# ---------------------------------------------------------------------------
# 7. IDENTIFY OBJECTIVE AND SELECTION STABILITY
# ---------------------------------------------------------------------------

valid_scenarios_186 = sensitivity_df_186[
    sensitivity_df_186["feasible_combinations"] > 0
]

baseline_mask_186 = int(baseline_row_186["best_customer_mask"])

same_selection_count_186 = int(
    (
        valid_scenarios_186["best_customer_mask"]
        == baseline_mask_186
    ).sum()
)

total_valid_scenarios_186 = len(valid_scenarios_186)

selection_stability_percent_186 = (
    same_selection_count_186 / total_valid_scenarios_186 * 100
    if total_valid_scenarios_186 > 0 else 0.0
)

minimum_objective_186 = (
    float(valid_scenarios_186["best_objective"].min())
    if total_valid_scenarios_186 else None
)

maximum_objective_186 = (
    float(valid_scenarios_186["best_objective"].max())
    if total_valid_scenarios_186 else None
)

print("\n7. OPTIMAL SELECTION STABILITY")
print("Baseline optimal customers:", baseline_row_186["best_customers"])
print("Scenarios retaining baseline selection:", same_selection_count_186)
print("Scenarios with feasible solutions:", total_valid_scenarios_186)
print("Selection stability (%):", selection_stability_percent_186)
print("Minimum scenario optimum:", minimum_objective_186)
print("Maximum scenario optimum:", maximum_objective_186)

# ---------------------------------------------------------------------------
# 8. CREATE OBJECTIVE SENSITIVITY HEATMAP
# ---------------------------------------------------------------------------

objective_matrix_186 = sensitivity_df_186.pivot(
    index="capacity_limit",
    columns="budget_limit",
    values="best_objective"
)

plt.figure(figsize=(12, 8))

plt.imshow(
    objective_matrix_186.values,
    aspect="auto",
    origin="lower"
)

plt.xticks(
    range(len(objective_matrix_186.columns)),
    objective_matrix_186.columns
)

plt.yticks(
    range(len(objective_matrix_186.index)),
    objective_matrix_186.index
)

plt.xlabel("Budget limit")
plt.ylabel("Capacity limit")
plt.title("Optimal Objective Sensitivity to Budget and Capacity")

plt.colorbar(label="Best feasible objective")
plt.tight_layout()

heatmap_path_186 = os.path.join(
    "/content/ATMIQ_FINAL_RESULTS",
    "cell_186_objective_sensitivity_heatmap.png"
)

plt.savefig(heatmap_path_186, dpi=300, bbox_inches="tight")
plt.show()
plt.close()

print("\n8. SENSITIVITY HEATMAP SAVED")
print(heatmap_path_186)

# ---------------------------------------------------------------------------
# 9. EXPORT RESULTS
# ---------------------------------------------------------------------------

results_dir_186 = "/content/ATMIQ_FINAL_RESULTS"
os.makedirs(results_dir_186, exist_ok=True)

csv_path_186 = os.path.join(
    results_dir_186,
    "cell_186_budget_capacity_sensitivity.csv"
)

json_path_186 = os.path.join(
    results_dir_186,
    "cell_186_sensitivity_report.json"
)

sensitivity_df_186.to_csv(csv_path_186, index=False)

report_186 = {
    "cell": 186,
    "total_combinations": len(combinations_df_186),
    "budget_scenarios": budget_values_186,
    "capacity_scenarios": capacity_values_186,
    "total_scenarios": len(sensitivity_df_186),
    "feasible_scenarios": total_valid_scenarios_186,
    "infeasible_scenarios": int(
        (sensitivity_df_186["feasible_combinations"] == 0).sum()
    ),
    "baseline_budget": BASELINE_BUDGET_186,
    "baseline_capacity": BASELINE_CAPACITY_186,
    "baseline_objective": float(baseline_row_186["best_objective"]),
    "baseline_customers": baseline_row_186["best_customers"],
    "baseline_verified": baseline_verified_186,
    "scenarios_retaining_baseline_selection": same_selection_count_186,
    "selection_stability_percent": selection_stability_percent_186,
    "minimum_scenario_optimum": minimum_objective_186,
    "maximum_scenario_optimum": maximum_objective_186,
    "heatmap": heatmap_path_186
}

with open(json_path_186, "w") as f:
    json.dump(report_186, f, indent=4)

print("\n9. EXPORTS")
print("Sensitivity CSV:", csv_path_186)
print("Sensitivity JSON:", json_path_186)

# ---------------------------------------------------------------------------
# 10. FINAL SUMMARY
# ---------------------------------------------------------------------------

print("\n10. FINAL SUMMARY")
print("Baseline classical optimum verified:", baseline_verified_186)
print("Total scenarios evaluated:", len(sensitivity_df_186))
print("Selection stability (%):",
      round(selection_stability_percent_186, 4))
print("Results directory:", results_dir_186)

print("\n" + "=" * 85)
print("CELL 186 EXECUTION COMPLETED")
print("=" * 85)
```
